5.Type Casting

```python
# Integer
num_str = "25"
num_int = int(num_str)

print("String to Integer:", num_int)
print(type(num_int))

print()

# Float
num_float = float(num_int)

print("Integer to Float:", num_float)
print(type(num_float))

print()

# String
text = str(num_float)

print("Float to String:", text)
print(type(text))

print()

# Boolean
boolean_value = bool(1)

print("Integer to Boolean:", boolean_value)
print(type(boolean_value))

print()

# Complex
complex_number = complex(10)

print("Integer to Complex:", complex_number)
print(type(complex_number))
```

7.operators

```python
# Sample Variables
a = 15
b = 4
# 1. Arithmetic Operators
print("Arithmetic Operators")
print("Addition (+):", a + b)
print("Subtraction (-):", a - b)
print("Multiplication (*):", a * b)
print("Division (/):", a / b)
print("Floor Division (//):", a // b)
print("Modulus (%):", a % b)
print("Exponent (**):", a ** b)

# 2. Comparison Operators
print("\nComparison Operators")
print("Equal (==):", a == b)
print("Not Equal (!=):", a != b)
print("Greater Than (>):", a > b)
print("Less Than (<):", a < b)
print("Greater Than or Equal (>=):", a >= b)
print("Less Than or Equal (<=):", a <= b)
```

8.strings

```python
# 1. Creating Strings
single_quote = 'Hello'
double_quote = "Python"
triple_quote = """Welcome to Data Science"""

print("Single Quote :", single_quote)
print("Double Quote :", double_quote)
print("Triple Quote :", triple_quote)
```

```python
text = "Python"
print("First Character :", text[0])
print("Second Character:", text[1])
print("Last Character  :", text[-1])
```

```python
# 3. String Slicing
print(text[0:3])
print(text[:4])
print(text[2:])
print(text[::-1])
```

```python
# 4. String Concatenation
first_name = "Muhammad"
last_name = "Anas"
full_name = first_name + " " + last_name
print("\nFull Name :", full_name)
```

```python
# 5. String Length
print("Length :", len(full_name))
```

```python
# 7. String Formatting
age = 21
print(f"\nMy name is {full_name} and I am {age} years old.")
```

11.Built-in Functions

```python
# 1. print()
print("print()")
print("Hello, Python!")

# 2. type()
print("type()")
print(type(100))
print(type(3.14))
print(type("Python"))
print(type(True))

# 3. int(), float(), str(), bool()
print("\n===== Type Conversion =====")
print(int("25"))
print(float(10))
print(str(100))
print(bool(1))

# 4. len()

print("len()")

name = "Muhammad Anas"

print(len(name))

numbers = [10, 20, 30, 40]

print(len(numbers))
# 5. max(), min(), sum()

print("max(), min(), sum()")

marks = [75, 82, 95, 68, 90]

print("Maximum :", max(marks))
print("Minimum :", min(marks))
print("Sum     :", sum(marks))

# 6. abs(), round(), pow()

print("abs(), round(), pow()")

print(abs(-25))
print(round(3.14159, 2))
print(pow(2, 5))

# 7. sorted()

print("sorted()")

numbers = [9, 5, 3, 7, 1]

print(sorted(numbers))
print(sorted(numbers, reverse=True))

# 8. reversed()

print("reversed()")

text = "Python"

print(list(reversed(text)))

# 9. enumerate()

print("enumerate()")

fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)

# 10. zip()

print("zip()")

names = ["Ali", "Ahmed", "Sara"]
marks = [90, 85, 95]

for student, score in zip(names, marks):
    print(student, score)

# 11. id()


print("id()")

x = 100

print(id(x))

# 12. isinstance()

print("isinstance()")

print(isinstance(10, int))
print(isinstance("Python", str))
print(isinstance(3.14, float))

# 13. chr() and ord()

print("chr() and ord()")

print(chr(65))
print(ord("A"))
```
