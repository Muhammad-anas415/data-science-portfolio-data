1.comparison operators

```python
#comparison operators

a = 10
b = 20
print("Equal (==):", a == b)
print("Not Equal (!=):", a != b)
print("Greater Than (>):", a > b)
print("Less Than (<):", a < b)
print("Greater Than or Equal (>=):", a >= b)
print("Less Than or Equal (<=):", a <= b)
```

2.logical operators

```python
#Logical operators
a = True
b = False
print("Logical AND (and):", a and b)
print("Logical OR (or):", a or b)
print("Logical NOT (not):", not a)
print("Logical XOR (xor):", a ^ b)
print("Logical NAND (nand):", not (a and b))
print("Logical NOR (nor):", not (a or b))
print("Logical XNOR (xnor):", not (a ^ b))
```

3.if else  statement

```python
#if statement
a = 10
b = 20
if a > b:
    print("a is greater than b")
else:
    print("a is less than or equal to b")
```

4.if else elif satatments

```python
#if else eli statments
a = 10
b = 20
if a > b:
    print("a is greater than b")
elif a == b:
    print("a is equal to b")
else:
    print("a is less than b")
```

5.Nested if statements

```python
#nested if statements
a = 10
b = 20
c = 30
if a > b:
    if b > c:
        print("a is greater than b and b is greater than c")
    else:
        print("a is greater than b but b is less than or equal to c")
else:
    print("a is less than or equal to b")
```

6.match case statement

```python
6.#match...case statement
a = 10
b = 20
match a:
    case 10:
        print("a is 10")
    case 20:
        print("a is 20")
    case _:
        print("a is not 10 or 20")
```

7.for loop

```python
#for loop
for i in range(5):
    print(i)
```

8.range function

```python
#range function
a = range(5)
print(a)
print(list(a))
print(tuple(a))
print(set(a))
print(dict(enumerate(a)))
```

9.while loop

```python
#while loop 
i = 0
while i < 5:
    print(i)
    i += 1
```

10.break statement

```python
#break statment
for i in range(5):
    if i == 2:
        break
    print(i)
```

11.continue statement

```python
#continue statment
for i in range(5):
    if i == 2:
        continue
    print(i)
```

12.pass statement

```python
#pass statment
for i in range(5):
    if i == 2:
        pass
    print(i)
```

13.Nested Loop

```python
#nested loop 
for i in range(5):
    for j in range(5):
        print(i, j)
        if i == 2:
            break
    if i == 2:
        break
```
