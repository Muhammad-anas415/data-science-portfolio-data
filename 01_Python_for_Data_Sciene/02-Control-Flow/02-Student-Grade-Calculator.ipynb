# Student Grade Calculator


```python
# Student Grade Calculator
marks = float(input("Enter your marks (0-100): "))
if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "Fail"
print("Result")
print("Marks :", marks)
print("Grade :", grade)
```
