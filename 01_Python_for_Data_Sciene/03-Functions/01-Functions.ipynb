1.Creating a Function

```python
#creating a funtion
def say_hello():
    print("hello")
    print("how")
    print("are")
    print("you")
    
say_hello()
```

2.Calling a Function

```python
say_hello()
```

3.Function Parameters

```python
#functions parameters
def say_hello(name):
    print(f"hello {name}")
    print("how")
    print("are")
    print("you")
    
say_hello("anas")
```

4.Function Arguments

```python
#funciion arguments
def say_hello(name="anas"):
    print(f"hello {name}")
    print("how")
    print("are")
    print("you")
    
say_hello()
say_hello("anas")
```

5.Multiple Parameters

```python
#Multiple Parameters
def say_hello(name="anas",time="morning"):
    print(f"good {time} {name}")
    print("how")
    print("are")
    print("you")
    
say_hello()
say_hello("anas")
say_hello("anas","afternoon")
```

6.Return Statement

```python
#return statement
def say_hello(name="anas",time="morning"):
    return f"good {time} {name}"
    
print(say_hello())
print(say_hello("anas"))
print(say_hello("anas","afternoon"))
```

7.Default Parameters

```python
#Default Parameters
def say_hello(name="anas",time="morning"):
    return f"good {time} {name}"
    
print(say_hello())
print(say_hello("anas"))
print(say_hello("anas","afternoon"))
```

8.Keyword Arguments

```python
#keyword Arguments
def say_hello(name="anas",time="morning"):
    return f"good {time} {name}"
    
print(say_hello())
print(say_hello(name="anas"))
print(say_hello(name="anas",time="afternoon"))
```

9.Local Variables

```python
#local variables
def say_hello(name="anas",time="morning"):
    return f"good {time} {name}"
    
print(say_hello())
print(say_hello(name="anas"))
print(say_hello(name="anas",time="afternoon"))
```

10.Global Variables

```python
#Global Variables
name="anas"
time="morning"

def say_hello(name="anas",time="morning"):
    return f"good {time} {name}"
    
print(say_hello())
print(say_hello(name="anas"))
print(say_hello(name="anas",time="afternoon"))
```

11.Lambda Functions

```python
#lamda function
greet = lambda name="anas",time="morning":f"good {time} {name}"
print(greet())
print(greet(name="anas"))
print(greet(name="anas",time="afternoon"))
```
