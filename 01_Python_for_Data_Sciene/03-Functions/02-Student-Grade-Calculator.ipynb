### Student Grade Calculator

```python
# Student Grade Calculator
def calculate_grade(marks):
    
    if marks >= 90:
        return "A+"
    
    elif marks >= 80:
        return "A"
    
    elif marks >= 70:
        return "B"
    
    elif marks >= 60:
        return "C"
    
    elif marks >= 50:
        return "D"
    
    else:
        return "Fail"

# Student Marks
students = {
    "Ali": 95,
    "Anas": 82,
    "Ahmed": 68,
    "Bilal": 45
}

print("Student Grades")

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name}: {marks} -> Grade {grade}")
```
