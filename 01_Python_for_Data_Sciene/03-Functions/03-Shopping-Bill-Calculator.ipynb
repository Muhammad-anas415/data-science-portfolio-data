### Shopping Bill Calculator

```python
# Shopping Bill Calculator
def calculate_bill(prices, discount):
    
    total = sum(prices)
    
    final_bill = total - (total * discount / 100)
    
    return final_bill
# Product Prices
prices = [250, 400, 150, 300]
discount = 10
bill = calculate_bill(prices, discount)
print("Shopping Bill Summary")
print("Product Prices :", prices)
print("Discount       :", discount, "%")
print("Final Bill     :", bill)
```
