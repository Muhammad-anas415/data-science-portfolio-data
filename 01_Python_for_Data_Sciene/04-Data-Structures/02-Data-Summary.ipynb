#### Data Summary

```python
# Dataset Summary
dataset = {
    "Age": [22, 25, 28, 30, 35],
    "Salary": [30000, 45000, 50000, 60000, 75000]
}
def summarize(column_name, values):

    print(f"\n{column_name} Summary")
    print("Count   :", len(values))
    print("Maximum :", max(values))
    print("Minimum :", min(values))
    print("Average :", round(sum(values) / len(values), 2))
for column, values in dataset.items():
    summarize(column, values)
```
