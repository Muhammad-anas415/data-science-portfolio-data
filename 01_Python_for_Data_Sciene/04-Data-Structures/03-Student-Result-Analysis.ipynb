#### Student Result Analysis

```python
# Student Result Analysis
marks = [85, 72, 90, 65, 88, 95, 78]

def result_summary(data):

    print("Total Students :", len(data))
    print("Highest Marks  :", max(data))
    print("Lowest Marks   :", min(data))
    print("Average Marks  :", round(sum(data) / len(data), 2))

result_summary(marks)
```
