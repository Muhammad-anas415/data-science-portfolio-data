# Introduction to NumPy

#### What is NumPy?

NumPy (Numerical Python) is a powerful Python library used for numerical computing. It provides fast and efficient support for arrays, matrices, and mathematical operations.

NumPy is the foundation of many Data Science, Machine Learning, and Deep Learning libraries.

#### Why Do We Use NumPy?

We use NumPy because it:

- Performs calculations much faster than Python lists.
- Supports multi-dimensional arrays.
- Provides mathematical and statistical functions.
- Offers linear algebra operations.
- Is widely used in Data Science and AI.
