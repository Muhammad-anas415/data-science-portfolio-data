#  NumPy Array Properties

#### Creating Array

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
])

print(arr)
```

#### shape

The `shape` property tells us the number of rows and columns in an array.

```python
print(arr.shape)
```

#### ndim

The `ndim` property returns the number of dimensions of an array.

```python
print(arr.ndim)
```

#### size

The `size` property returns the total number of elements in the array.

```python
print(arr.size)
```

#### dtype

The `dtype` property returns the data type of the array elements.

```python
print(arr.dtype)
```

#### astype()

The `astype()` method converts the data type of an array.

```python
arr1 = np.array([1,2,3,4])

new_arr = arr1.astype(float)

print(new_arr)

print(new_arr.dtype)
```

#### len()

The `len()` function returns the number of rows in the array.

```python
print(len(arr))
```

```python
marks = np.array([
    [85,90,78],
    [88,76,92],
    [95,89,84]
])

print("Student Marks")
print(marks)

print("\nShape      :", marks.shape)
print("Dimensions :", marks.ndim)
print("Size       :", marks.size)
print("Data Type  :", marks.dtype)
```
