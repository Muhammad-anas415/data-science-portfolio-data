# NumPy Indexing and Slicing

#### Importing Numpy in python

```python
import numpy as np
```

#### Creating one dimensional array

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

#### Positive Indexing

Positive indexing starts from **0** and moves from left to right.

```python
print(arr[0])
print(arr[1])
print(arr[2])
```

#### Negative Indexing

Negative indexing starts from **-1** and moves from right to left.

```python
print(arr[-1])
print(arr[-2])
print(arr[-3])
```

#### Slicing

Slicing extracts a range of elements.

Syntax:

array[start : stop]

```python
print(arr[1:4])
print(arr[:3])

print(arr[2:])

print(arr[:])

print(arr[1:5])
```

#### Step Slicing

The third value in slicing specifies the step size.

Syntax:

array[start : stop : step]

```python
print(arr[::2])

print(arr[1::2])

print(arr[::-1])
```

#### Creating a two-dimensional array

```python
matrix = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print(matrix)
```

#### 2D Array Indexing

Use:

array[row, column]

```python
print(matrix[0,0])

print(matrix[1,2])

print(matrix[2,1])
```

#### 2D Array Slicing

```python
print(matrix[0:2,0:2])

print(matrix[:,1])

print(matrix[1,:])

print(matrix[:,0])
```
