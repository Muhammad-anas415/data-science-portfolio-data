# NumPy Array Operations

#### Importing Numpy in python

```python
import numpy as np 
```

#### Creating Array

```python
arr1 = np.array([10, 20, 30, 40])

arr2 = np.array([1, 2, 3, 4])

print(arr1)
print(arr2)
```

#### Addition

The `+` operator adds corresponding elements from two arrays.

```python
print(arr1 + arr2)
```

#### Subtraction

The `-` operator subtracts corresponding elements.

```python
print(arr1 - arr2)
```

#### Multiplication

The `*` operator performs element-wise multiplication.

```python
print(arr1 * arr2)
```

#### Division

The `/` operator divides corresponding elements.

```python
print(arr1 / arr2)
```

#### Floor Division

The `//` operator returns the integer part of the division result.

```python
print(arr1 // arr2)
```

#### Modulus

The `%` operator returns the remainder after division.

```python
print(arr1 % arr2)
```

#### Power

The `**` operator raises each element to a given power.

```python
print(arr2 ** 2)
```
