# Statistics with NumPy

### Importing Numpy in python

```python
import numpy as np
```

#### Creating NumPy Array

```python
marks = np.array([85, 90, 78, 92, 88])

print(marks)
```

#### np.sum()

The `np.sum()` function returns the total sum of all elements in an array.

```python
print(np.sum(marks))
```

#### np.mean()

The `np.mean()` function calculates the average value of all elements.

```python
print(np.mean(marks)) 
```

#### np.median()

The `np.median()` function returns the middle value after sorting the data.

```python
print(np.median(marks))
```

#### np.min()

Returns the smallest value in the array.

```python
print(np.min(marks))
```

#### np.max()

Returns the largest value in the array.

```python
print(np.max(marks))
```

#### np.std()

Standard deviation measures how spread out the values are from the mean.

A small standard deviation means the values are close to the average.

A large standard deviation means the values are more spread out.

```python
print(np.std(marks))
```

#### np.var()

Variance measures how much the data varies from the mean.

It is the square of the standard deviation.

```python
print(np.var(marks))
```

#### np.percentile()

Returns the value below which a given percentage of observations falls.

```python
print(np.percentile(marks, 25))

print(np.percentile(marks, 50))

print(np.percentile(marks, 75))
```

#### np.quantile()

Returns values based on quantiles.

```python
print(np.quantile(marks, 0.25))

print(np.quantile(marks, 0.50))

print(np.quantile(marks, 0.75))
```
