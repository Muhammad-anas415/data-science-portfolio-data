# Linear Algebra with NumPy

#### Importing Numpy in pyhton


```python
import numpy as np 
```

#### Vector

A vector is a one-dimensional array that stores values in a single row or column.

```python
vector = np.array([2, 4, 6])

print(vector)
```

#### Matrix

A matrix is a two-dimensional array consisting of rows and columns.

```python
matrix = np.array([
    [1,2],
    [3,4]
])

print(matrix)
```

#### Matrix Addition

Matrix addition adds the corresponding elements of two matrices having the same shape.

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(A + B)
```

#### Matrix Subtraction

Matrix subtraction subtracts the corresponding elements of two matrices having the same shape.

```python
print(A - B)
```

#### Matrix Multiplication

Matrix multiplication combines rows of the first matrix with columns of the second matrix.

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(np.matmul(A, B))
```

#### Dot Product

The dot product multiplies corresponding elements of two vectors and then adds the results.

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.dot(a,b))
```

#### Transpose

The transpose swaps the rows and columns of a matrix.

```python
A = np.array([
    [1,2,3],
    [4,5,6]
])

print(A.T)
```

#### Determinant

The determinant is a single value that describes certain properties of a square matrix.

A determinant of **0** means the matrix cannot be inverted.

```python
A = np.array([
    [2,3],
    [1,4]
])

print(np.linalg.det(A))
```

#### Inverse Matrix

The inverse of a matrix is used to solve systems of linear equations.

```python
A = np.array([
    [2,1],
    [5,3]
])

print(np.linalg.inv(A))
```

#### Matrix Rank

The rank of a matrix represents the number of linearly independent rows or columns.

```python
A = np.array([
    [1,2],
    [2,4]
])

print(np.linalg.matrix_rank(A))
```

#### Matrix Trace

The trace of a matrix is the sum of the main diagonal elements.

```python
A = np.array([
    [2,3],
    [4,5]
])

print(np.trace(A))
```

#### Eigenvalues

Eigenvalues describe how a matrix stretches or shrinks vectors during a transformation.

```python
A = np.array([
    [4,2],
    [1,3]
])

values, vectors = np.linalg.eig(A)

print(values)
```

#### Eigenvectors

Eigenvectors are special vectors whose direction remains unchanged after a matrix transformation.

```python
print(vectors)
```
