#  NumPy Random Module

#### Import Numpy in python 

```python
import numpy as np
```

#### np.random.rand()

The `np.random.rand()` function generates random decimal numbers between **0 and 1**.

```python
print(np.random.rand())
print(np.random.rand(5))
print(np.random.rand(3,3))
```

#### np.random.randn()

The `np.random.randn()` function generates random numbers from a standard normal distribution with a mean of 0 and a standard deviation of 1.

```python
print(np.random.randn())
print(np.random.randn(5))
print(np.random.randn(3,3))
```

#### np.random.randint()

The `np.random.randint()` function generates random integers within a specified range.

```python
print(np.random.randint(1,10))
print(np.random.randint(1,100,size=5))
print(np.random.randint(1,50,size=(3,4)))
```

#### np.random.choice()

The `np.random.choice()` function randomly selects elements from a list or array.

```python
colors = ["Red","Blue","Green","Black"]

print(np.random.choice(colors))
print(np.random.choice(colors, size=3))
```

#### np.random.shuffle()

The `shuffle()` function randomly rearranges the elements of an existing array.

```python
numbers = np.array([1,2,3,4,5,6])

np.random.shuffle(numbers)

print(numbers)
```
