# Boolean Indexing in NumPy

#### Boolean Indexing

Boolean Indexing filters an array using conditions.

If the condition is **True**, the value is selected.

If the condition is **False**, the value is ignored.

#### Importing Numpy in python

```python
import numpy as np 
```

#### Creating Array

```python
marks = np.array([45, 60, 78, 82, 95, 38, 67])

print(marks)
```

```python
# Greater than 
print(marks[marks > 70])
```

```python
#less than
print(marks[marks < 50])
```

```python
#Greater than or equal to
print(marks[marks >= 50])
```

```python
#less than or equal to 
print(marks[marks <= 60])
```

```python
#Equal too
print(marks[marks == 82])
```

```python
#Not equal to 
print(marks[marks != 60])
```

```python
#And operator
print(marks[(marks >= 60) & (marks <= 90)])
```

```python
#Not Operator
print(marks[~(marks > 70)])
```

```python
#Filtering string
names = np.array(["Ali","Ahmed","Sara","Ayesha","Usman"])

print(names[names == "Sara"])
```

```python
# 2D bolean indexing
marks = np.array([
    [85,90,78],
    [88,45,92],
    [35,89,84]
])

print(marks)
print(marks[marks > 80])
```
