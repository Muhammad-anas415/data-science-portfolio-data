### What is Pandas?

Pandas is an open-source Python library used for **data manipulation**, **data analysis**, and **data cleaning**.

It provides two main data structures:

- **Series** → One-dimensional labeled array.
- **DataFrame** → Two-dimensional table with rows and columns.

Pandas is built on top of **NumPy**, which means it is fast, efficient, and optimized for numerical operations.
