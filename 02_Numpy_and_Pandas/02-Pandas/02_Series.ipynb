# Pandas Series

#### What is a Series?

A Series is a one-dimensional labeled array.

Each element has:

- An index
- A value

Think of it as a single column in an Excel spreadsheet.

#### Importing Pandas in Python

```python
import pandas as pd 
```

#### Creating a series 

```python
marks = pd.Series([85,90,78,92])

print(marks)
```

#### Custom Index

Instead of using default indexes (0,1,2...), we can assign our own labels.

```python
students = pd.Series(
    [85,90,78],
    index=["Ali","Ahmed","Sara"]
)

print(students)
```

#### Series Properties

```python
marks = pd.Series([85,90,78,92])

print("Shape :", marks.shape)
print("Size :", marks.size)
print("Data Type :", marks.dtype)
print("Index :", marks.index)
print("Values :", marks.values)
```

#### Acessing Elements

```python
print(marks[0])

print(marks[2])
```

#### Slicing

```python
print(marks[:2])

print(marks[1:])

print(marks[1:3])
```

#### Mathematical Operations

```python
marks = pd.Series([80,85,90,95])

print(marks + 5)

print(marks * 2)

print(marks - 10)

print(marks / 5)
```

#### Statistical Operations

```python
print("Sum :", marks.sum())

print("Mean :", marks.mean())

print("Median :", marks.median())

print("Maximum :", marks.max())

print("Minimum :", marks.min())

print("Standard Deviation :", marks.std())
```

#### Sorting

```python
print(marks.sort_values())
```

```python
print(marks.sort_values(ascending=False))
```
