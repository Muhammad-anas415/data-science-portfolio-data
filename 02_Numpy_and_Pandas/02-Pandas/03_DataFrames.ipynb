#  Pandas DataFrame

### What is a DataFrame?

A DataFrame is a two-dimensional table made up of rows and columns.

Think of it like an Excel spreadsheet:

- Rows represent records.
- Columns represent features (variables).
- Each column is a Pandas Series.

A DataFrame can store different data types in different columns.

#### Importing pandas in python

```python
import pandas as pd
```

#### Creating a DataFrame

```python
students = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(students)

print(df)
```

#### DataFrame Properties

These properties help us understand the structure of a dataset.

```python
print("Shape:")
print(df.shape)

print("\nDimensions:")
print(df.ndim)

print("\nSize:")
print(df.size)

print("\nColumns:")
print(df.columns)

print("\nIndex:")
print(df.index)

print("\nData Types:")
print(df.dtypes)

print("\nValues:")
print(df.values)
```

#### Selecting a column

```python
print(df["Name"])
```

```python
print(df[["Name", "Marks"]])
```

#### Selecting a row

```python
print(df.loc[0])
```

```python
print(df.iloc[2])
```

```python
print(df.iloc[0:3])
```

#### Adding a new column

```python
df["Department"] = [
    "CS",
    "IT",
    "SE",
    "AI"
]

print(df)
```

#### Deleting a column

```python
df = df.drop("Department", axis=1)

print(df)
```
