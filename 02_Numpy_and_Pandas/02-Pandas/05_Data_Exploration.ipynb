#  Data Exploration with Pandas

Data Exploration is the process of understanding the structure, quality, and contents of a dataset before performing data cleaning, visualization, or machine learning.

#### Importing pandas in python

```python
import pandas as pd 
```

#### Load the data

```python
df = pd.read_csv("students.csv")
df
```

#### head()

The `head()` function displays the first rows of the dataset.

By default, it returns the first **5 rows**.

```python
df.head()
```

#### tail()

The `tail()` function displays the last rows of the dataset.

By default, it returns the last **5 rows**.

```python
df.tail()
```

#### sample()

The `sample()` function returns random rows from the dataset.

It is useful for quickly inspecting random records.

```python
df.sample(3)
```

#### shape

Returns the number of rows and columns.

Format:

(rows, columns)

```python
df.shape
```

#### Columns names 

```python
df.columns
```

#### dtypes

Displays the data type of each column.

```python
df.dtypes
```

#### info()

Displays a summary of the dataset, including:

- Number of rows
- Number of columns
- Column names
- Data types
- Non-null values
- Memory usage

```python
df.info()
```

#### describe()

Generates statistical information for numerical columns.

```python
df.describe()
```

```python
df.describe(include="all")
```

#### value_counts()

Counts how many times each value appears in a column.

```python
df["Department"].value_counts()
```

```python
df["Department"].value_counts(normalize=True)
```

#### Unique values

```python
df["Department"].unique()
```

#### Number of unique values

```python
df["Department"].nunique()
```

#### Missing values

```python
df.isnull().sum()
```

```python
df.isnull().any().sum()
```

#### Duplicated Rows

```python
df.duplicated().sum()
```
