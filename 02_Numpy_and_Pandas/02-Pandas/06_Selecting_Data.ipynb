# Selecting Data in Pandas


#### Importing pandas

```python
import pandas as pd 
```

#### Load the dataset 

```python
df = pd.read_csv("students.csv")
df
```

#### Selecting One Column

A single column can be selected using its column name.

```python
df["Name"]
```

#### Selecting Multiple Columns

Multiple columns can be selected using a list of column names.

```python
df[["Name","Marks"]]
```

#### Selecting Single row

```python
df.iloc[0]
```

#### Selecting Multiple Row

```python
df.iloc[0:3]
```

#### loc[]

`loc[]` selects data using row labels and column names.

```python
df.loc[0,"Marks"]
```

```python
df.loc[0:2,["Name","Marks"]]
```

#### iloc[]

`iloc[]` selects data using row and column positions.

```python
df.iloc[0]
```

```python
df.iloc[0,4]
```

```python
df.iloc[0:3,1:5]
```

#### Conditional Selection

```python
df[df["Marks"]>85]
```

```python
df[df["Department"]=="CS"]
```

```python
df[df["Age"]>20]
```
