# Filtering Data in Pandas

#### Importing pandas 

```python
import pandas as pd 
```

#### Load the dataset 

```python
df = pd.read_csv("students.csv")
df
```

#### Comparison Operators

```python
df[df["Marks"] > 85]
```

```python
df[df["Marks"] < 90]
```

```python
df[df["Marks"] == 90]
```

```python
df[df["Marks"] != 90]
```

#### Logical Operators

```python
df[(df["Marks"] > 80) & (df["Age"] >= 20)]
```

```python
df[(df["Marks"] > 90) | (df["Age"] < 20)]
```

```python
df[~(df["Department"] == "CS")]
```

#### Multiple Conditions

```python
df[(df["Department"] == "CS") & (df["Marks"] > 85)]
```

```python
df[(df["Department"] == "AI") | (df["Department"] == "IT")]
```

#### Using between 

```python
df[df["Marks"].between(80, 90)]
```

```python
df[df["Age"].between(20, 22)]
```

####  Filtering in String

```python
df[df["Name"].str.startswith("A")]
```

```python
df[df["Name"].str.endswith("a")]
```

```python
df[df["Name"].str.contains("ar")]
```
