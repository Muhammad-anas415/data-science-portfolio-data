#  Data Cleaning with Pandas

#### Importing pandas

```python
import pandas as pd 
```

#### Loading dataset 

```python
df=pd.read_csv("students.csv")
df
```

#### Missing Values

Missing values are empty cells represented by **NaN**.

They can reduce model performance if not handled properly.

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

#### Droping missing values

```python
df.dropna()
```

#### Filling missing values

```python
df.fillna(0)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
```

```python
df["City"] = df["City"].fillna("Unknown")
```

#### Find duplicated rows

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

#### Removes the duplicated rows

```python
df = df.drop_duplicates()
df
```

##### Renaming columns

```python
df.rename(columns={"Marks":"Score"})
```

```python
df.rename(columns={
    "Marks":"Score",
    "Age":"Student_Age"
})
```

#### Change the data type

```python
df.dtypes
```

```python
df["Student_ID"] = df["Student_ID"].astype(str)
```

#### Replace values 

```python
df["Department"] = df["Department"].replace("CS","Computer Science")
df
```

```python
df["Department"] = df["Department"].replace({
    "CS":"Computer Science",
    "IT":"Information Technology"
})
df
```

#### String cleaning

```python
df["Name"] = df["Name"].str.strip()
```

```python
df["Name"] = df["Name"].str.lower()
```

```python
df["Name"] = df["Name"].str.upper()
```

```python
df["Name"] = df["Name"].str.title()
```
