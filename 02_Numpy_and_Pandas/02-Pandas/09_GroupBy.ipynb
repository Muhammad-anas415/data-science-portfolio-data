#  GroupBy in Pandas

#### What is GroupBy?

`groupby()` divides the dataset into groups based on one or more columns.

After grouping, we can perform calculations such as:

- Sum
- Mean
- Maximum
- Minimum
- Count
- Standard Deviation
- Median

This follows the **Split → Apply → Combine** approach:

1. Split the data into groups.
2. Apply a function to each group.
3. Combine the results into a new table.

#### Importing pandas and reading the dataset

```python
import pandas as pd 
df = pd.read_csv("students.csv")
df
```

#### Grouopby one column

```python
df.groupby("Department")["Marks"].mean()
```

#### Aggeragation functions

```python
df.groupby("Department")["Marks"].sum()
```

```python
df.groupby("Department")["Marks"].max()
```

```python
df.groupby("Department")["Marks"].min()
```

```python
df.groupby("Department")["Marks"].count()
```

```python
df.groupby("Department")["Marks"].std()
```

```python
df.groupby("Department")["Marks"].median()
```

#### Multiple Aggeragation functions

```python
df.groupby("Department")["Marks"].agg(
    ["mean", "max", "min", "sum", "count"]
)
```

#### Groupby by Multiple columns

```python
df.groupby(
    ["Department", "City"]
)["Marks"].mean()
```
