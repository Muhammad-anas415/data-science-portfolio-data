#  Merging DataFrames in Pandas

#### What is Merging?

Merging combines two DataFrames using a common column.

Common columns are called **keys**.

In this example:

Student_ID → Common Key

#### Importing pandas 

```python
import pandas as pd 
```

#### Creating DataFrames

```python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Department": ["CS", "IT", "SE", "AI"]
})

print(students)
```

```python
marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Marks": [85, 90, 92, 88]
})

print(marks)
```

#### Inner Join

```python
merged_df = pd.merge(
    students,
    marks,
    on="Student_ID"
)

print(merged_df)
```

#### Left Join

```python
left_df = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

print(left_df)
```

#### Right Join

```python
right_df = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)

print(right_df)
```

#### Outer Join

```python
outer_df = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)

print(outer_df)
```
