# Inferential Statistics
Inferential Statistics is the branch of statistics that allows us to use **sample data** to make conclusions, predictions, or decisions about an entire **population**.

#### Population

A **Population** is the complete collection of individuals, objects, or observations that we want to study. It represents the entire group about which we want to draw conclusions.

#### Sample

A **Sample** is a subset of the population selected for analysis. Since studying an entire population is often expensive and time-consuming, a representative sample is used to make conclusions about the population.

```python
import numpy as np
import pandas as pd
# Set seed for reproducibility
np.random.seed(42)
# Create a population of 10,000 exam scores
population = np.random.normal(loc=75, scale=10, size=10000)
# Draw a random sample of 200 students
sample = np.random.choice(population, size=200, replace=False)
print(f"Population Size : {len(population)}")
print(f"Sample Size     : {len(sample)}")
print("\nPopulation Mean :", round(np.mean(population), 2))
print("Sample Mean     :", round(np.mean(sample), 2))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.hist(population, bins=30, alpha=0.6, label="Population")
plt.hist(sample, bins=15, alpha=0.8, label="Sample")
plt.title("Population vs Sample Distribution")
plt.xlabel("Exam Scores")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
