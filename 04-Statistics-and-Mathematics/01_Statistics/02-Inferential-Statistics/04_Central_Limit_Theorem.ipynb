#  Central Limit Theorem (CLT)


The Central Limit Theorem states that:

> As the sample size becomes sufficiently large, the distribution of sample means approaches a normal distribution, regardless of the shape of the original population.




The Central Limit Theorem generally works well when:

- Samples are selected randomly.
- Samples are independent.
- The sample size is sufficiently large.

#### Importing libraries


```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

#### Create a non normal population

```python
# Exponential distribution (clearly right-skewed)
population = np.random.exponential(scale=2, size=10000)
print("Population Size:", len(population))
```

#### Visualize the population distribution

```python
plt.figure(figsize=(8,5))
plt.hist(population, bins=40)
plt.title("Original Population Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

#### Take multiple samples from the population

```python
sample_means = []
sample_size = 30
for i in range(1000):

    sample = np.random.choice(population,
                              size=sample_size,
                              replace=True)

    sample_means.append(np.mean(sample))
```

#### Vislualize the distribution of sample means

```python
plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=30)
plt.title("Distribution of Sample Means")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.show()
```

#### Comparing the population and sample means

```python
print("Population Mean :", round(np.mean(population),2))
print("Average of Sample Means :", round(np.mean(sample_means),2))
```
