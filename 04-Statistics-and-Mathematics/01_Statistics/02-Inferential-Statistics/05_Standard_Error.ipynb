#  Standard Error (SE)
Standard Error (SE) is a statistical measure that describes how much the **sample mean** is expected to vary from the **true population mean**.

#### Formula

The Standard Error of the Mean (SEM) is calculated as:

```text
SE = Standard Deviation / √Sample Size
```

Where:

- SE = Standard Error
- SD = Standard Deviation
- n = Sample Size


#### Importing LIbraries

````python
import pandas as pd 
import numpy as np
````

#### Creating a sample data

````python
np.random.seed(42)
sample = np.random.normal(loc=70, scale=10, size=100)
print(sample[:10])
````

#### Calculating Standard deviation

````python
sample_std = np.std(sample, ddof=1)
print("Standard Deviation:", round(sample_std,2))
````

#### Calculating Standard Error manually

````python
sample_size = len(sample)
standard_error = sample_std / np.sqrt(sample_size)
print("Sample Size:", sample_size)
print("Standard Error:", round(standard_error,2))
````

#### Calculating Standard Error by using `scipy.stats.sem()`

````python
from scipy.stats import sem
print("Standard Error:", round(sem(sample),2))
````
