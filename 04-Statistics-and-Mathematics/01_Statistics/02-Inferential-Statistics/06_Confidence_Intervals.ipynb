# Confidence Intervals (CI)

A **Confidence Interval** is an interval estimate of a population parameter based on sample data.

#### Confidence Interval Formula

For a population mean:

```text
Confidence Interval = Sample Mean ± Margin of Error
```

Where

```text
Margin of Error = Critical Value × Standard Error

#### Importing libraries


````python
import numpy as np
from scipy import stats
````

#### Generating a sample data

````python
np.random.seed(42)
sample = np.random.normal(loc=70, scale=10, size=100)
print(sample[:10])
````

#### Calculate the sample mean

````python
sample_mean = np.mean(sample)
print("Sample Mean:", round(sample_mean,2))
````

#### Calculating Standard Error

````python
standard_error = stats.sem(sample)
print("Standard Error:", round(standard_error,2))
````

#### Calculating confidance interval

````python
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(sample)-1,
    loc=sample_mean,
    scale=standard_error
)
print("95% Confidence Interval:")
print(confidence_interval)
````
