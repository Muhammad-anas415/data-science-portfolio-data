# Hypothesis Testing
A **hypothesis** is a statement or claim about a population parameter.
Hypothesis Testing is the process of using sample data to determine whether there is enough statistical evidence to support or reject that claim.

#### Main types of Hypothesis Testing
1-T_Test

2-Z_Test

3-ANOVA

4- Chi-Square Test

#### Steps of Hypothesis Testing

1. Define the Null Hypothesis (H₀).
2. Define the Alternative Hypothesis (H₁).
3. Choose the significance level (α).
4. Collect sample data.
5. Calculate the test statistic.
6. Calculate the p-value.
7. Compare the p-value with α.
8. Make a decision.
9. Draw a conclusion.

#### Importing libraries

````python
import numpy as np
import scipy.stats as stats
````

#### Genrating a population data

````python
np.random.seed(42)
population = np.random.normal(
    loc=70,
    scale=10,
    size=10000
)
print("Population Size :", len(population))
print("Population Mean :", round(np.mean(population),2))
````

#### Taking a Random Sample

````python
sample = np.random.choice(
    population,
    size=30,
    replace=False
)
print(sample)
````

#### Explore the sample data

````python
print("Sample Size :", len(sample))
print("Sample Mean :", round(np.mean(sample),2))
print("Sample Std :", round(np.std(sample,ddof=1),2))
````

#### Step 1: Define the Hypotheses
##### Null Hypothesis (H₀)
The average exam score is **70 marks**.
```text
H₀ : μ = 70
```
##### Alternative Hypothesis (H₁)
The average exam score is **not equal to 70 marks**.
```text
H₁ : μ ≠ 70
```

#### Step 2: Choose Significance Level

The significance level is

```text
α = 0.05
```

Decision Rule

- If **p-value < 0.05**
  - Reject H₀

- If **p-value ≥ 0.05**
  - Fail to Reject H₀

#### Perform the Hypothesis Test

````python
from scipy.stats import ttest_1samp
t_statistic, p_value = ttest_1samp(sample, popmean=70)
print("T Statistic :", round(t_statistic,3))
print("P Value :", round(p_value,4))
````

#### Decision Making

````python
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis (H₀)")
else:
    print("Fail to Reject the Null Hypothesis (H₀)")
````

#### Conclusion

````python
if p_value < alpha:
    print("There is sufficient evidence that the population mean is different from 70.")
else:
    print("There is insufficient evidence to conclude that the population mean differs from 70.")
````
