# P-Value

The **p-value (Probability Value)** is a statistical measure used in **Hypothesis Testing** to determine whether the observed results are statistically significant.

It helps us decide whether there is enough evidence to **reject the Null Hypothesis (H₀)**.

#### Decision Rule
Compare the **p-value** with the **significance level (α)**.
##### If
```text
p-value < α
```
Reject the Null Hypothesis (H₀)
---
#####  If
```text
p-value ≥ α
```
Fail to Reject the Null Hypothesis (H₀)


#### importing libraries 

````python
import numpy as np
from scipy.stats import ttest_1samp
````

#### Generating a sample data 

````python
np.random.seed(42)
sample = np.random.normal(loc=72, scale=10, size=30)
print(sample)
````

#### Performing one-sample-ttest

````python
t_statistic, p_value = ttest_1samp(sample, popmean=70)
print("T Statistic :", round(t_statistic, 3))
print("P-Value :", round(p_value, 4))
````

#### Make Decision

````python
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis (H₀)")
else:
    print("Fail to Reject the Null Hypothesis (H₀)")
````

#### Summary
The **p-value** measures how likely the observed sample results are if the Null Hypothesis is true.

Key Takeaways:
- A small p-value provides strong evidence against the Null Hypothesis.
- A large p-value indicates that the sample results are consistent with the Null Hypothesis.
- Compare the p-value with the significance level (α) to make a statistical decision.
- The p-value does **not** measure the probability that the Null Hypothesis is true or false.
