#  Type I Error and Type II Error

In Hypothesis Testing, our decision is based on a sample rather than the entire population. Because of this, there is always a possibility of making an incorrect decision.

These incorrect decisions are called **Type I Error** and **Type II Error**.


A **Type I Error** occurs when we **reject the Null Hypothesis (H₀)** even though it is actually **true**.

It is also called a **False Positive**.

#### Decision

```text
Reality:
H₀ is True

Decision:
Reject H₀

 Wrong Decision

````python
import numpy as np
from scipy.stats import ttest_1samp
np.random.seed(42)
population_mean = 70
alpha = 0.05
# Population where H0 is TRUE
population = np.random.normal(loc=70, scale=10, size=100000)
type1_errors = 0
n_experiments = 1000
for _ in range(n_experiments):

    sample = np.random.choice(population, size=30)

    t_stat, p_value = ttest_1samp(sample, popmean=70)

    if p_value < alpha:
        type1_errors += 1
print("Total Experiments :", n_experiments)
print("Type I Errors :", type1_errors)
print("Type I Error Rate :", type1_errors / n_experiments)
````

#### Type II Error (False Negative)

A **Type II Error** occurs when we **fail to reject the Null Hypothesis (H₀)** even though it is actually **false**.

It is also known as a **False Negative** because we incorrectly conclude that there is no significant effect when one actually exists.

#### Decision

```text
Reality:
H₀ is False

Decision:
Fail to Reject H₀

Type II Error (False Negative)
```

````python
np.random.seed(42)
alpha = 0.05
# Population where H0 is FALSE
population = np.random.normal(loc=75, scale=10, size=100000)
type2_errors = 0
n_experiments = 1000
for _ in range(n_experiments):

    sample = np.random.choice(population, size=30)

    t_stat, p_value = ttest_1samp(sample, popmean=70)

    if p_value >= alpha:
        type2_errors += 1
print("Total Experiments :", n_experiments)
print("Type II Errors :", type2_errors)
print("Type II Error Rate :", type2_errors / n_experiments)
````
