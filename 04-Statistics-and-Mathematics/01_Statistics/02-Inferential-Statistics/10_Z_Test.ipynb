#  Z-Test

A **Z-Test** is a statistical hypothesis test used to determine whether there is a significant difference between a **sample mean** and a **population mean** when:

- The population standard deviation (σ) is **known**, or
- The sample size is **large (n ≥ 30)**.

The Z-Test follows the **Standard Normal Distribution (Z-Distribution)**.

#### Z-Test Formula

```text
             x̄ − μ
Z = ------------------------
        σ / √n
```

Where:

- x̄ = Sample Mean
- μ = Population Mean
- σ = Population Standard Deviation
- n = Sample Size

#### Importing Libraries

````python
import numpy as np
from statsmodels.stats.weightstats import ztest
````

#### Generate a population data

````python
np.random.seed(42)
population = np.random.normal(
    loc=70,
    scale=10,
    size=100000
)
print("Population Mean :", round(np.mean(population),2))
print("Population Std :", round(np.std(population),2))
````

#### Take a sample

````python
sample = np.random.choice(
    population,
    size=100,
    replace=False
)
print(sample[:10])
print("Sample Size :", len(sample))
print("Sample Mean :", round(np.mean(sample),2))
print("Sample Std :", round(np.std(sample),2))
````

#### State the hypothesis

````python
population_mean = 70
print("H₀ : μ = 70")
print("H₁ : μ ≠ 70")
````

#### Perform the Z-Test

````python
z_statistic, p_value = ztest(
    sample,
    value=population_mean
)
print("Z Statistic :", round(z_statistic,3))
print("P Value :", round(p_value,4))
````

#### Decision 

````python
alpha = 0.05
if p_value < alpha:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
````

#### Conclusion

````python
if p_value < alpha:
    print("There is sufficient evidence that the population mean differs from 70.")
else:
    print("There is insufficient evidence to conclude that the population mean differs from 70.")
````
