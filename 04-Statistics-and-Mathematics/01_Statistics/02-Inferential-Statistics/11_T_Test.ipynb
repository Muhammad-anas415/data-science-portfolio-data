#  One-Sample T-Test

A **One-Sample T-Test** is used to determine whether the **mean of a sample** is significantly different from a known or hypothesized **population mean**.

Unlike the Z-Test, the T-Test is used when:

- The **population standard deviation (σ) is unknown**.
- The sample size is **small (typically n < 30)**.

####  Formula

```text
              x̄ − μ
t = -----------------------------
        s / √n
```

Where

- x̄ = Sample Mean
- μ = Population Mean
- s = Sample Standard Deviation
- n = Sample Size

#### Importing Libraries 

````python
import numpy as np
from scipy.stats import ttest_1samp
import matplotlib.pyplot as plt
````

#### Generate a population data

````python
np.random.seed(42)
population = np.random.normal(
    loc=70,
    scale=10,
    size=10000
)
print("Population Mean :", round(np.mean(population),2))
````

#### Take a sampl data 

````python
sample = np.random.choice(
    population,
    size=25,
    replace=False
)
sample
````

````python
print("Sample Size :", len(sample))
print("Sample Mean :", round(np.mean(sample),2))
print("Sample Standard Deviation :", round(np.std(sample,ddof=1),2))
````

#### State Hypothesis

````python
print("Null Hypothesis (H₀): μ = 70")
print("Alternative Hypothesis (H₁): μ ≠ 70")
````

#### perform One-Sample T-Test

````python
t_statistic, p_value = ttest_1samp(
    sample,
    popmean=70
)
print("T Statistic :", round(t_statistic,3))
print("P Value :", round(p_value,4))
````

#### Decision Making

````python
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
````

#### Conclusion

````python
if p_value < alpha:
    print("There is sufficient evidence that the average marks are different from 70.")
else:
    print("There is insufficient evidence to conclude that the average marks differ from 70.")
````

#### Visulaize the sample data

````python
plt.figure(figsize=(8,5))
plt.hist(sample,
         bins=8,
         edgecolor="black")

plt.axvline(np.mean(sample),
            color="red",
            linestyle="--",
            label="Sample Mean")

plt.axvline(70,
            color="green",
            linestyle="--",
            label="Population Mean")

plt.legend()

plt.title("One-Sample T-Test")

plt.xlabel("Marks")

plt.ylabel("Frequency")

plt.show()
````
