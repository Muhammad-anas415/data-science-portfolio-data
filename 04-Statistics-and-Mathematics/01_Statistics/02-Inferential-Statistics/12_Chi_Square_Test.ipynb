#  Chi-Square (χ²) Test

The **Chi-Square (χ²) Test** is a statistical hypothesis test used to determine whether there is a **significant association or relationship between two categorical variables**.

Unlike the Z-Test and T-Test, which compare **means**, the Chi-Square Test compares **frequencies (counts)**.

#### Chi-Square Formula

```text
           (Observed − Expected)²
χ² = Σ -----------------------------
              Expected

Where:

- O = Observed Frequency
- E = Expected Frequency


#### Import Libraries 

````python
import pandas as pd 
from scipy.stats import chi2_contingency
````

#### Create a Sample data

````python
data = {
    "Gender": [
        "Male","Male","Male","Male","Male",
        "Female","Female","Female","Female","Female"
    ],
    "Product": [
        "A","A","B","B","A",
        "A","B","B","B","A"
    ]
}
df = pd.DataFrame(data)
df
````

#### Create a Contingency Table

````python
table = pd.crosstab(
    df["Gender"],
    df["Product"]
)
table
````

#### Perform Chi-Square Test

````python
chi2, p_value, dof, expected = chi2_contingency(table)
print("Chi-Square Statistic :", round(chi2,3))
print("P Value :", round(p_value,4))
print("Degrees of Freedom :", dof)
````

#### Expected Frequency

````python
expected_df = pd.DataFrame(
    expected,
    index=table.index,
    columns=table.columns
)
expected_df
````

#### Decision making

````python
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
````

#### Conclusion

````python
if p_value < alpha:
    print("There is a significant relationship between Gender and Product Preference.")
else:
    print("There is no significant relationship between Gender and Product Preference.")
````
