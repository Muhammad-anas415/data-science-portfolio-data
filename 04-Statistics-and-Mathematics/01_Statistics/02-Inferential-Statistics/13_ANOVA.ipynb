#  ANOVA (Analysis of Variance)

**ANOVA (Analysis of Variance)** is a statistical hypothesis test used to determine whether the **means of three or more independent groups** are significantly different.

Instead of performing multiple T-Tests (which increases the chance of making a Type I Error), ANOVA compares all groups at once using a single test.

ANOVA uses the **F-Distribution** to make statistical decisions.

####  F-Statistic

The F-Statistic compares:

- Variation **between** groups.
- Variation **within** groups.

A **large F-value** suggests that at least one group mean is significantly different from the others.

#### Importing Libraries

```python
import numpy as np
import pandas as pd
from scipy.stats import f_oneway
```

#### generating a sample data

```python
np.random.seed(42)
group_A = np.random.normal(70, 5, 30)
group_B = np.random.normal(75, 5, 30)
group_C = np.random.normal(80, 5, 30)
```

#### Calculating means of each group

```python
print("Group A Mean :", round(np.mean(group_A),2))
print("Group B Mean :", round(np.mean(group_B),2))
print("Group C Mean :", round(np.mean(group_C),2))
```

#### Performing one way anova 

```python
f_statistic, p_value = f_oneway(
    group_A,
    group_B,
    group_C
)
print("F Statistic :", round(f_statistic,3))
print("P Value :", round(p_value,5))
```

#### Decision making 

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### Conclusion 

```python
if p_value < alpha:
    print("There is a significant difference between at least one pair of group means.")
else:
    print("There is no significant difference between the group means.")
```
