# Scalars in Linear Algebra

A **Scalar** is the simplest mathematical object in Linear Algebra. It is a **single numerical value** that has **magnitude only** and **no direction**.
Scalars are used to represent quantities such as age, temperature, mass, time, distance, and price.
In Python and NumPy, scalars are represented as single values like integers or floating-point numbers.

A scalar is a **single numerical quantity**.

It contains:

- Magnitude
- No Direction

#### Importing library

```python
import numpy as np 
```

#### Creating Scalars 

```python
integer_scalar = 10
float_scalar = 15.5
negative_scalar = -8
zero_scalar = 0

print(integer_scalar)
print(float_scalar)
print(negative_scalar)
print(zero_scalar)
```

#### Numpy Scalars 

```python
scalar = np.array(25)
print(scalar)
print(type(scalar))
```

#### Scalar  Data Type

```python
scalar = np.array(25)
print("Value :", scalar)
print("Data Type :", scalar.dtype)
print("Dimension :", scalar.ndim)
print("Shape :", scalar.shape)
print("Size :", scalar.size)
```

#### Scalar Arithmetic

```python
a = 15
b = 5

print("Addition :", a + b)
print("Subtraction :", a - b)
print("Multiplication :", a * b)
print("Division :", a / b)
print("Power :", a ** b)
```

#### Scalar with matrix

```python
matrix = np.array([
    [1,2],
    [3,4]
])
scalar = 3
print("Original Matrix")
print(matrix)
print()
print("Scalar Multiplication")
print(scalar * matrix)
```
