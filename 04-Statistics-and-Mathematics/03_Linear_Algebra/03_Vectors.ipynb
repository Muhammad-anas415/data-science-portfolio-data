# Vectors in Linear Algebra

A **Vector** is an ordered collection of numbers arranged in a single row or a single column. Unlike a scalar, a vector has both **magnitude** and **direction**.

Vectors are used to represent data, coordinates, features, and observations in mathematics, physics, and computer science. In Machine Learning, every data sample is commonly represented as a vector of features.

A **vector** is a collection of numerical values that represents both **magnitude** and **direction**.

#### Importing Library 

````python
import numpy as np 
````

#### Creating a simple vector 

````python
vector = np.array([2, 4, 6, 8])
print(vector)
````

#### Vector Properties

````python
print("Vector :", vector)
print("Dimensions :", vector.ndim)
print("Shape :", vector.shape)
print("Size :", vector.size)
print("Data Type :", vector.dtype)
````

#### Creating a Row vector 

````python
row_vector = np.array([[2, 4, 6]])
print(row_vector)
print("Shape :", row_vector.shape)
````

#### Creating a Column vector

````python
column_vector = np.array([[2],
                          [4],
                          [6]])
print(column_vector)
print("Shape :", column_vector.shape)
````

#### Accessing Elements of a Vector

````python
vector = np.array([10,20,30,40,50])
print(vector[0])
print(vector[2])
print(vector[-1])
````

#### Sclicing a Vector

````python
vector = np.array([10,20,30,40,50,60])
print(vector[1:5])
print(vector[:3])
print(vector[3:])
````

#### Creating zero ones and random vectors 

````python
zeros = np.zeros(5)
ones = np.ones(5)
random_vector = np.random.randint(1,10,5)
print("Zeros :", zeros)
print("Ones :", ones)
print("Random :", random_vector)
````
