# Vector Operations

Vector operations are mathematical operations performed on vectors. These operations allow us to manipulate data, calculate similarities, measure distances, and solve mathematical problems.

In Machine Learning, vector operations are performed continuously during data preprocessing, optimization, neural network training, and prediction.

Vector operations are mathematical calculations applied to vectors.

The most common operations include:

- Vector Addition
- Vector Subtraction
- Scalar Multiplication
- Element-wise Multiplication
- Dot Product
- Vector Magnitude
- Vector Normalization

#### Importing Library

```python
import numpy as np
```

#### Creating two Vectors

```python
vector1 = np.array([2, 4, 6])
vector2 = np.array([1, 3, 5])
print("Vector 1 :", vector1)
print("Vector 2 :", vector2)
```

#### Vector Addition

```python
addition = vector1 + vector2
print("Addition :", addition)
```

#### Vector Subtraction

```python
subtraction = vector1 - vector2
print("Subtraction :", subtraction)
```

#### Vector  Multiplication

```python
multiplication = vector1 * vector2
print(multiplication)
```

#### Vector Division 

```python
division = vector1 / vector2
print(division)
```

#### Vector magnitude 

```python
vector = np.array([3,4])
magnitude = np.linalg.norm(vector)
print("Magnitude :", magnitude)
```

#### Unit Vector (Normalized Vector)

```python
vector = np.array([3,4])
unit_vector = vector / np.linalg.norm(vector)
print(unit_vector)
```

#### Dot Product 

```python
vector1 = np.array([2,4,6])
vector2 = np.array([1,3,5])
dot_product = np.dot(vector1, vector2)
print(dot_product)
```

#### Sum of vector elements

```python
vector = np.array([10,20,30,40])
print(np.sum(vector))
```

#### Mean of vector elements

```python
vector = np.array([10,20,30,40])
print(np.mean(vector))
```

#### Maximum and Minimum 

```python
vector = np.array([15,40,12,60,25])
print("Maximum :", np.max(vector))
print("Minimum :", np.min(vector))
```

#### Sort a Vector 

```python
vector = np.array([25,10,50,15,5])
print(np.sort(vector))
```
