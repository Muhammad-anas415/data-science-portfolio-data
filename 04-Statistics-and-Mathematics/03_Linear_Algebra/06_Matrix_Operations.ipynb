#  Matrix Operations

Matrix operations are mathematical operations performed on matrices. They are fundamental to Linear Algebra and are widely used in Data Science, Machine Learning, Deep Learning, Computer Vision, and Artificial Intelligence.

The most commonly used matrix operations are:

- Matrix Addition
- Matrix Subtraction
- Scalar Multiplication
- Element-wise Multiplication
- Matrix Multiplication
- Matrix Transpose
- Matrix Sum
- Matrix Mean
- Matrix Maximum
- Matrix Minimum

#### Import Libraries 

```python
import numpy as np
```

#### Creating two Matrices 

```python
matrix1 = np.array([
    [1, 2],
    [3, 4]
])
matrix2 = np.array([
    [5, 6],
    [7, 8]
])
print("Matrix 1:\n", matrix1)
print("\nMatrix 2:\n", matrix2)
```

#### Matrix Addition 

```python
addition = matrix1 + matrix2
print(addition)
```

#### Matrix Subtraction 

```python
subtraction = matrix2 - matrix1
print(subtraction)
```

#### Scalar Multiplication 

```python
scalar = 3
result = scalar * matrix1
print(result)
```

#### Matrix Multiplication 

```python
matrix_product = np.matmul(matrix1, matrix2)
print(matrix_product)
```

#### Matrix Division 

```python
division = matrix2 / matrix1
print(division)
```

#### Matrix Transpose 

```python
transpose = matrix1.T
print(transpose)
```

#### Matrix sum 

```python
print(np.sum(matrix1))
```

#### Matrix mean 

```python
print(np.mean(matrix1))
```

#### Matrix min and max 

```python
print("Maximum :", np.max(matrix1))
print("Minimum :", np.min(matrix1))
```

#### Matrix shape 

```python
print(matrix1.shape)
```

#### Matrix dimension 

```python
print(matrix1.ndim)
```

#### Matrix Size 

```python
print(matrix1.size)
```
