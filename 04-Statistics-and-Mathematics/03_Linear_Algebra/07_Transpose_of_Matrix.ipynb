# Transpose of a Matrix

The **transpose of a matrix** is a new matrix obtained by **interchanging its rows and columns**. In other words, every row of the original matrix becomes a column in the transposed matrix, and every column becomes a row.

The transpose of a matrix is represented by the symbol **Aᵀ** (read as "A transpose").

#### Import library 

```python
import numpy as np
```

#### Creating a matrix using NumPy 

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matrix)
```

#### Find the transpose of the matrix

```python
transpose = matrix.T
print(transpose)
```

##### Checking shape of the matrix 

```python
print("Original Shape :", matrix.shape)
print("Transpose Shape :", transpose.shape)
```

#### Using the `numpy.transpose()` function

```python
transpose_matrix = np.transpose(matrix)
print(transpose_matrix)
```

#### Transpose of square matrix

```python
square_matrix = np.array([
    [1, 2],
    [3, 4]
])
print("Original Matrix:\n", square_matrix)
print("\nTranspose:\n", square_matrix.T)
```

#### Transpose of a row vector 

```python
row_vector = np.array([[10, 20, 30]])
print("Row Vector:\n", row_vector)
print("\nTranspose:\n", row_vector.T)
```

#### Transpose of a column vector

```python
column_vector = np.array([
    [10],
    [20],
    [30]
])
print("Column Vector:\n", column_vector)
print("\nTranspose:\n", column_vector.T)
```
