#  Determinant of a Matrix

The **determinant** is a special numerical value that can be calculated from a **square matrix**. It provides important information about the matrix, such as whether the matrix is invertible, whether its rows or columns are linearly independent, and how the matrix transforms space.

The determinant is commonly denoted as:

```text
det(A)
or

|A|

#### Types of Determinant Values

##### Positive Determinant (> 0)

- Matrix is invertible.
- Orientation is preserved.

##### Negative Determinant (< 0)

- Matrix is invertible.
- Orientation is reversed (reflection).


##### Zero Determinant (= 0)

- Matrix is singular.
- Matrix cannot be inverted.
- Rows or columns are linearly dependent.

#### Import Library 

````python
import numpy as np
````

#### Create a 2x2 Matrix

````python
matrix = np.array([
    [4, 7],
    [2, 6]
])
print(matrix)
````

#### Calculate the Determinant of a 2x2 Matrix

````python
det = np.linalg.det(matrix)
print("Determinant:", det)
````

#### Determinant of a 3x3 Matrix

````python
matrix = np.array([
    [1, 2, 3],
    [0, 4, 5],
    [1, 0, 6]
])
det = np.linalg.det(matrix)
print(det)
````

#### Singular Matrix (Determinant = 0)

````python
matrix = np.array([
    [2, 4],
    [1, 2]
])
print(np.linalg.det(matrix))
````

#### Non-Singular Matrix (Determinant ≠ 0)

````python
matrix = np.array([
    [5, 2],
    [3, 4]
])
print(np.linalg.det(matrix))
````

#### Determinant of Identity Matrix

````python
identity = np.eye(3)
print(identity)
print("Determinant:", np.linalg.det(identity))
````

#### determinant of a Diagonal Matrix

````python
matrix = np.array([
    [3, 0, 0],
    [0, 5, 0],
    [0, 0, 2]
])
print(np.linalg.det(matrix))
````
