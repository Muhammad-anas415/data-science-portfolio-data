#  Inverse of a Matrix

The **inverse of a matrix** is another matrix that, when multiplied by the original matrix, produces the **Identity Matrix**.

The inverse of a matrix **A** is denoted by:

```text
A⁻¹

If

```text
A × A⁻¹ = I
```

where **I** is the Identity Matrix, then **A⁻¹** is called the inverse of **A**.

#### Importing Library 

````python
import numpy as np 
````

#### Create a matrix using NumPy

````python
matrix = np.array([
    [4, 7],
    [2, 6]
])
print(matrix)
````

#### Find the determinant of the matrix

````python
det = np.linalg.det(matrix)
print("Determinant:", det)
````

#### Calculate the Inverse of the matrix

````python
inverse = np.linalg.inv(matrix)
print(inverse)
````

#### Verify the result by multiplying the original matrix with its inverse to get the Identity Matrix.

````python
identity = matrix @ inverse
print(identity)
````

#### Inverse of 3x3 Matrix

````python
matrix = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])
inverse = np.linalg.inv(matrix)
print(inverse)
````
