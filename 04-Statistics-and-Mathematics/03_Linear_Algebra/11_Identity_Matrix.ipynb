#  Identity Matrix

An **Identity Matrix** is a special type of **square matrix** in which:

- Every element on the **main diagonal** is **1**.
- Every other element is **0**.

The identity matrix is represented by the symbol:

```text
I
```

or

```text
Iₙ

where **n** is the size of the matrix.

The identity matrix plays the same role in matrix multiplication that the number **1** plays in ordinary multiplication.


#### Import library 

````python
import numpy as np 
````

#### Create a 3x3 identity matrix

````python
identity = np.eye(3)
print(identity)
````

#### Create a 5x5 identity matrix

````python
identity = np.eye(5)
print(identity)
````

#### Identity Matrix Multiplication

````python
A = np.array([
    [2,4],
    [6,8]
])
I = np.eye(2)
print("Matrix A")
print(A)
print("\nIdentity Matrix")
print(I)
print("\nA @ I")
print(A @ I)
````

#### determinant of a Identity Matrix

````python
I = np.eye(4)
print(np.linalg.det(I))
````

#### Rank of a identity Matrix

````python
I = np.eye(5)
print(np.linalg.matrix_rank(I))
````

#### Inverse of a Identity Matrix

````python
I = np.eye(4)
print(np.linalg.inv(I))
````
