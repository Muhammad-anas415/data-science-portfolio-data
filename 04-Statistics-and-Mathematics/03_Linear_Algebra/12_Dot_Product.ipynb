#  Dot Product

The **Dot Product**, also known as the **Scalar Product** or **Inner Product**, is an operation performed between two vectors of the same length. The result of a dot product is always a **single scalar value**.

The dot product multiplies the corresponding elements of two vectors and then adds the results together.

For two vectors:

A = [a₁, a₂, ..., aₙ]

B = [b₁, b₂, ..., bₙ]

The dot product is:

A · B = a₁b₁ + a₂b₂ + ... + aₙbₙ

The result is always a **scalar (single number)**.

#### Import Library 

```python
import numpy as np 
```

#### Creating two vectors 

```python
vector1 = np.array([2, 4, 6])
vector2 = np.array([1, 3, 5])
print("Vector 1:", vector1)
print("Vector 2:", vector2)
```

#### Calculate the dot product using numpy's `np.dot()` function

```python
result = np.dot(vector1, vector2)
print("Dot Product:", result)
```

#### Manual Calculation of Dot Product

```python
manual = (2*1) + (4*3) + (6*5)
print(manual)
```

#### using the @ operator

```python
result = vector1 @ vector2
print(result)
```

#### Dor product of matrices 

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
print(np.dot(A, B))
```

#### Dot product with floating numbers 

```python
A = np.array([1.5, 2.5, 3.5])
B = np.array([4.0, 5.0, 6.0])
print(np.dot(A, B))
```
