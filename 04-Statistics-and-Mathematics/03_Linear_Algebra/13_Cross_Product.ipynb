# Cross Product

The **Cross Product**, also known as the **Vector Product**, is an operation performed between **two 3-dimensional vectors**. The result of a cross product is another **vector** that is **perpendicular** to both input vectors.

The cross product is represented as:

```text
A × B

#### Import Library 

````python
import numpy as np 
````

#### Create two vetors 

````python
A = np.array([2, 3, 4])
B = np.array([5, 6, 7])
print("Vector A:", A)
print("Vector B:", B)
````

#### Calculate cross product using numpy's `np.cross()` function

````python
cross_product = np.cross(A, B)
print("Cross Product:", cross_product)
````

#### Cross product of parallel vectors

````python
A = np.array([2, 4, 6])
B = np.array([1, 2, 3])
print(np.cross(A, B))
````

#### Cross product of unit vectors

````python
i = np.array([1, 0, 0])
j = np.array([0, 1, 0])
k = np.array([0, 0, 1])
print("i × j =", np.cross(i, j))
print("j × k =", np.cross(j, k))
print("k × i =", np.cross(k, i))
````

#### Compare cross product with dot product

````python
A = np.array([2, 3, 4])

B = np.array([5, 6, 7])

print("Dot Product:", np.dot(A, B))

print("Cross Product:", np.cross(A, B))
````
