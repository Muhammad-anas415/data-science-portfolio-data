# Linear Combination

A **Linear Combination** is the process of combining two or more vectors using **scalar multiplication** and **vector addition**.

A **linear combination** is formed by multiplying vectors with scalars and then adding them.

Mathematically,

```
w = a·u + b·v
```

where:

- **u** and **v** are vectors.
- **a** and **b** are scalars.
- **w** is the resulting vector.

#### Import LIbrary 

````python
import numpy as np 
````

#### Creating two vectors

````python
u = np.array([2, 1])
v = np.array([1, 3])
print("Vector u:", u)
print("Vector v:", v)
````

#### Define Scalar 

````python
a = 2
b = 3
print("Scalar a:", a)
print("Scalar b:", b)
````

#### Compute the linear combination

````python
w = a * u + b * v
print("Linear Combination:", w)
````
