#  Eigenvalues and Eigenvectors

An **Eigenvalue** is the scalar that tells us **how much an Eigenvector is stretched or compressed** during the transformation.

- Positive Eigenvalue → Same direction.
- Negative Eigenvalue → Opposite direction.
- Large magnitude → Greater stretching.
- Small magnitude → Less stretching.


#### Mathematical Equation

The relationship between a matrix, its Eigenvector, and its Eigenvalue is:

```text
A · v = λ · v
```

Where:

- **A** = Matrix
- **v** = Eigenvector
- **λ (lambda)** = Eigenvalue

#### Import library 

````python
import numpy as np
````

#### Create a Matrix 

````python
A = np.array([
    [4, 2],
    [1, 3]
])

print(A)
````

#### Compute Eigenvalues and Eigenvectors

````python
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n")

print(eigenvectors)
````
