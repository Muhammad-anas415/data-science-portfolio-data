# Singular Value Decomposition (SVD)

**Singular Value Decomposition (SVD)** is one of the most powerful matrix decomposition techniques in Linear Algebra. It decomposes any matrix into three simpler matrices, making it easier to analyze, compress, and process data.

Unlike Eigenvalue Decomposition, which only works directly with square matrices, **SVD works with any matrix**, whether it is square or rectangular.

#### Import Library 

```python
import numpy as np 
```

#### Ceate a Matrix 

```python
A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print(A)
```

#### COmpute SVD

```python
U, S, VT = np.linalg.svd(A)

print("U Matrix:\n")
print(U)

print("\nSingular Values:\n")
print(S)

print("\nV Transpose:\n")
print(VT)
```

#### Shape of the matrix 

```python
print("Shape of A :", A.shape)
print("Shape of U :", U.shape)
print("Shape of S :", S.shape)
print("Shape of VT:", VT.shape)
```
