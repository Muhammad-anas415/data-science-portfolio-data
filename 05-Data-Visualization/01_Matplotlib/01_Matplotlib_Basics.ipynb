#  Matplotlib Basics
Matplotlib is one of the most popular and widely used data visualization libraries in Python. It provides a flexible and powerful interface for creating static, animated, and interactive visualizations.

Built on top of NumPy, Matplotlib allows users to create publication-quality charts with complete control over every element of a figure.

Whether you are performing Exploratory Data Analysis (EDA), presenting business insights, or visualizing Machine Learning results, Matplotlib serves as the foundation for most Python visualizations.

Many other visualization libraries, such as Seaborn, are built on top of Matplotlib.

Matplotlib is an open-source Python library used for creating high-quality data visualizations.

It supports:

- Line Charts
- Bar Charts
- Scatter Plots
- Histograms
- Pie Charts
- Box Plots
- Heatmaps
- Area Charts
- Subplots
- 3D Plots

It is considered the foundation of Python data visualization.

#### Importing Libraries

```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Matplotlib Version:", plt.matplotlib.__version__)
```

####  Creating Your First Plot

The simplest visualization in Matplotlib is a line plot.

A plot consists of two main components:

- X-axis
- Y-axis

The `plot()` function is used to draw a line graph.

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

####  Adding Titles and Labels

A good visualization should always contain:

- Title
- X-axis Label
- Y-axis Label

These help readers understand what the graph represents.

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

#### Changing Figure Size

The `figure()` function allows us to control the width and height of the visualization.

```python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.title("Figure Size Example")
plt.show()
```

#### Adding Grid Lines

Grid lines improve readability and help interpret values more accurately.

```python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.grid(True)
plt.show()
```

#### Changing Line Color, Style and Width

Matplotlib provides many customization options such as:

- Color
- Line Width
- Line Style
- Marker Style

```python
plt.figure(figsize=(8,5))
plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=3,
    marker="o"
)
plt.title("Customized Line")
plt.grid(True)
plt.show()
```

# Saving a Figure

The `savefig()` function saves the visualization as an image.

Supported formats include:

- PNG
- JPG
- PDF
- SVG

```python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.title("Save Figure Example")
plt.savefig("my_plot.png")
plt.show()
```
