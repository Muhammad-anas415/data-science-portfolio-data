#  Figure and Axes in Matplotlib

#### What is a Figure?

A **Figure** is the outermost container in Matplotlib.

It acts as the entire canvas on which one or more plots are drawn.

A Figure can contain:

- One Axes
- Multiple Axes
- Titles
- Legends
- Text
- Images

Think of a Figure as a blank sheet of paper where you can draw one or more graphs.

####  What is an Axes?

An **Axes** is the actual plotting area inside a Figure.

It contains:

- X-axis
- Y-axis
- Data
- Grid
- Labels
- Titles
- Legends

Every chart you create in Matplotlib is drawn on an Axes object.

A Figure may contain one or multiple Axes.

#### Importing libraries 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Creating a Figure
fig = plt.figure(figsize=(8,5))
print(fig)
```

The code above creates an empty **Figure**. Since no Axes have been added yet, no graph will be displayed.

```python
# Creating Figure and Axes
fig, ax = plt.subplots(figsize=(8,5))
print("Figure Object :", fig)
print("Axes Object :", ax)
```

`plt.subplots()` is the recommended way to create both a Figure and its Axes simultaneously.

- `fig` represents the Figure.
- `ax` represents the plotting area (Axes).

```python
x = np.arange(1,6)
y = x**2
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,y)
plt.show()
```

Notice that we are using **ax.plot()** instead of **plt.plot()**.

This is known as the **Object-Oriented (OO) interface**, which is the preferred approach for creating professional Matplotlib visualizations.

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,y)
ax.set_title("Square Function")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")
plt.show()
```

Instead of using:

- `plt.title()`
- `plt.xlabel()`
- `plt.ylabel()`

the Object-Oriented approach uses:

- `ax.set_title()`
- `ax.set_xlabel()`
- `ax.set_ylabel()`

This provides better control, especially when working with multiple plots.

```python
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(x,y)
ax.grid(True)
ax.set_facecolor("#f5f5f5")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,y,color="red",linewidth=3)
print("Figure Size:",fig.get_size_inches())
print("Figure DPI:",fig.dpi)
```

The Figure object also stores useful information such as:

- Figure Size
- Resolution (DPI)
- Background Color
- Overall Layout

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,y,label="Square")
ax.legend()
ax.grid(True)
plt.show()
```
