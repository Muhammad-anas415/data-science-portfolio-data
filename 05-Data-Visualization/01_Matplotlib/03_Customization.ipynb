# Matplotlib Customization

Creating a visualization is only the first step. A professional visualization should also be clear, readable, and visually appealing.

Matplotlib provides powerful customization options that allow us to modify almost every component of a chart, including titles, labels, colors, line styles, markers, fonts, legends, grids, axis limits, and background colors.

Good customization improves readability, highlights important information, and makes visualizations suitable for reports, presentations, dashboards, and research publications.


#### Importing libraries and creating a data 

````python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1,11)
y = x**2
````

#### Figure Size

The `figsize` parameter controls the width and height of the figure.

Syntax:

```python
plt.figure(figsize=(width,height))
```

````python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.title("Figure Size Example")
plt.show()
````

#### Line Color

The `color` parameter changes the color of the plotted line.

Some common colors:

- blue
- red
- green
- orange
- purple
- black
- cyan
- magenta

````python
plt.figure(figsize=(8,5))
plt.plot(x,y,color="red")
plt.title("Line Color")
plt.show()
````

#### Line Styles

The `linestyle` parameter changes the appearance of the line.

Common styles include:

- Solid (`-`)
- Dashed (`--`)
- Dash-dot (`-.`)
- Dotted (`:`)

````python
plt.figure(figsize=(8,5))
plt.plot(x,y,linestyle="--")
plt.title("Line Style")
plt.show()
````

#### Line Width

The `linewidth` parameter controls the thickness of the line.

A larger value produces a thicker line.

````python
plt.figure(figsize=(8,5))
plt.plot(x,y,linewidth=4)
plt.title("Line Width")
plt.show()
````

#### Markers

Markers highlight individual data points on the graph.

Common markers:

- o
- s
- ^
- *
- D
- x
- +

````python
plt.figure(figsize=(8,5))
plt.plot(
    x,
    y,
    marker="o",
    markersize=8
)
plt.title("Markers")
plt.show()
````

#### Titles and Axis Labels

Titles and labels help explain what the graph represents.

Always include:

- Chart Title
- X-axis Label
- Y-axis Label

````python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.title("Sales Growth")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
````

#### Grid Lines

Grid lines improve readability by making it easier to estimate values.

````python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.grid(True)
plt.show()
````

####  Legends

Legends identify multiple plotted datasets.

````python
plt.figure(figsize=(8,5))
plt.plot(x,y,label="Square Function")
plt.legend()
plt.show()
````

#### Axis Limits

Use `xlim()` and `ylim()` to control the visible range of the axes.

````python
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.xlim(0,12)
plt.ylim(0,120)
plt.show()
````

#### Background Color

The plotting area's background color can be customized using the Axes object.

````python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x,y)
ax.set_facecolor("red")
plt.show()
````
