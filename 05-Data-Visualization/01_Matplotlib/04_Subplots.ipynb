#  Subplots in Matplotlib

A subplot is an individual plotting area (Axes) inside a Figure.

Instead of displaying one graph per figure, multiple graphs can be arranged inside the same figure.

Benefits of subplots include:

- Better comparison of multiple datasets.
- Improved use of screen or page space.
- Professional dashboards.
- Cleaner reports.
- Easier analysis of multiple variables.

#### Importing libraries and creating a data

````python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1,11)
y1 = x
y2 = x**2
y3 = x**3
y4 = np.sqrt(x)
````

#### Creating a Single Subplot

A subplot with one row and one column is equivalent to a normal plot.

Syntax:

```python
fig, ax = plt.subplots()
```

````python
fig, ax = plt.subplots(figsize=(7,5))

ax.plot(x,y1)

ax.set_title("Single Subplot")

plt.show()
````

#### Creating Multiple Subplots

The `subplots()` function accepts:

- Number of rows
- Number of columns

Example:

```python
fig, ax = plt.subplots(2,2)
```

This creates four plotting areas.

````python
fig, ax = plt.subplots(2,2,figsize=(10,8))

plt.show()
````

#### Accessing Individual Subplots

Each subplot can be accessed using its row and column index.

Example:

```python
ax[0,0]
ax[0,1]
ax[1,0]
ax[1,1]
```

````python
fig, ax = plt.subplots(2,2,figsize=(10,8))

ax[0,0].plot(x,y1)

ax[0,1].plot(x,y2)

ax[1,0].plot(x,y3)

ax[1,1].plot(x,y4)

plt.show()
````

#### Adding Titles

Each subplot can have its own title.

````python
fig, ax = plt.subplots(2,2,figsize=(10,8))

ax[0,0].plot(x,y1)
ax[0,0].set_title("Linear")

ax[0,1].plot(x,y2)
ax[0,1].set_title("Quadratic")

ax[1,0].plot(x,y3)
ax[1,0].set_title("Cubic")

ax[1,1].plot(x,y4)
ax[1,1].set_title("Square Root")

plt.tight_layout()

plt.show()
````

#### Customizing Each Subplot

Every subplot is an independent Axes object.

Each subplot can have its own:

- Color
- Labels
- Grid
- Legend
- Background
- Limits

````python
fig, ax = plt.subplots(2,2,figsize=(10,8))

ax[0,0].plot(x,y1,color="red")
ax[0,0].grid(True)

ax[0,1].plot(x,y2,color="blue")
ax[0,1].grid(True)

ax[1,0].plot(x,y3,color="green")
ax[1,0].grid(True)

ax[1,1].plot(x,y4,color="purple")
ax[1,1].grid(True)

plt.tight_layout()

plt.show()
````
