#  Seaborn Themes and Styles

A well-designed visualization is not only informative but also visually appealing. Seaborn provides built-in themes and styles that improve the appearance of plots with minimal effort.

Themes control the overall look of a figure, including the background, grid lines, fonts, and colors. By applying different themes, you can create visualizations that are suitable for presentations, reports, dashboards, and research publications.

#### What is a Theme?

A theme is a predefined collection of visual settings that controls the appearance of plots.

A theme automatically adjusts:

- Background color
- Grid lines
- Fonts
- Tick marks
- Overall layout

Using themes helps create clean, consistent, and professional visualizations with very little code.

#### Importing libraries and loading data

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
```

#### Default Theme

By default, Seaborn applies its own modern styling, making charts more attractive than the default Matplotlib style.

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Default Seaborn Theme")

plt.show()
```

#### Available Themes

Seaborn provides five built-in themes:

- darkgrid
- whitegrid
- dark
- white
- ticks

Each theme is suitable for different visualization needs.

```python
sns.set_theme(style="darkgrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Darkgrid Theme")

plt.show()
```

```python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Whitegrid Theme")

plt.show()
```

```python
sns.set_theme(style="dark")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Dark Theme")

plt.show()
```

```python
sns.set_theme(style="white")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("White Theme")

plt.show()
```

```python
sns.set_theme(style="ticks")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Ticks Theme")

plt.show()
```
