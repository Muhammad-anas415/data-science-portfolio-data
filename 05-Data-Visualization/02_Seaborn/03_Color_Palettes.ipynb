# Seaborn Color Palettes


A color palette is a collection of colors used in a visualization.

Color palettes help us:

- Differentiate categories.
- Highlight important information.
- Improve readability.
- Create visually appealing charts.
- Maintain consistency across visualizations.

Choosing an appropriate palette makes visualizations easier to interpret and more effective at communicating insights.

#### Importing libraries and load the data

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
```

#### Viewing the Default Color Palette

Seaborn uses a default color palette that provides visually balanced colors for most plots.

```python
sns.color_palette()
```

#### Displaying a Color Palette

The `sns.palplot()` function displays all colors in a palette.

```python
sns.palplot(sns.color_palette())

plt.show()
```

#### Qualitative Color Palettes

Qualitative palettes are best suited for **categorical data**.

Common qualitative palettes include:

- deep
- muted
- bright
- pastel
- dark
- colorblind

```python
sns.palplot(sns.color_palette("deep"))

plt.show()
```

```python
sns.palplot(sns.color_palette("pastel"))

plt.show()
```

```python
sns.palplot(sns.color_palette("bright"))

plt.show()
```

#### Sequential Color Palettes

Sequential palettes are used when values increase from low to high.

Examples include:

- Blues
- Greens
- Reds
- Purples
- Oranges

```python
sns.palplot(sns.color_palette("Blues"))

plt.show()
```

```python
sns.palplot(sns.color_palette("Greens"))

plt.show()
```

#### Diverging Color Palettes

Diverging palettes are useful when data has a meaningful midpoint, such as positive vs. negative values.

Examples include:

- coolwarm
- RdBu
- Spectral
- BrBG

```python
sns.palplot(sns.color_palette("coolwarm"))

plt.show()
```

```python
sns.palplot(sns.color_palette("Spectral"))

plt.show()
```

#### Applying a Color Palette

The `palette` parameter changes the colors used in categorical plots.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="Set2"
)

plt.title("Bar Plot with Set2 Palette")

plt.show()
```

#### Creating a Custom Color Palette

Custom palettes allow complete control over visualization colors.

```python
custom_palette = ["#FF5733", "#3498DB", "#2ECC71"]

sns.palplot(custom_palette)

plt.show()
```

#### Setting a Global Color Palette

The `set_palette()` function changes the default palette for all future plots.

```python
sns.set_palette("pastel")

plt.figure(figsize=(8,5))

sns.barplot(
    data=tips,
    x="day",
    y="tip"
)

plt.title("Global Pastel Palette")

plt.show()
```

```python
sns.set_theme(style="whitegrid")

sns.set_palette("colorblind")

plt.figure(figsize=(9,5))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Professional Color Palette Example", fontsize=14)

plt.show()
```
