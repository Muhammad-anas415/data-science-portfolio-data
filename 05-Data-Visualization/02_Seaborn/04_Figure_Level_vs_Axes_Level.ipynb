# Figure-Level vs Axes-Level Functions in Seaborn

Seaborn provides two different interfaces for creating visualizations:

1. **Axes-Level Functions**
2. **Figure-Level Functions**

Although both are used to create visualizations, they differ in how they create and manage figures.

An **Axes-Level Function** draws a plot on an existing Matplotlib Axes object, giving you greater control and flexibility.

A **Figure-Level Function** automatically creates its own Figure and one or more Axes objects. It is especially useful for creating multiple plots using faceting and complex layouts with very little code.

#### What are Axes-Level Functions?

Axes-Level functions draw a visualization on a single Matplotlib **Axes** object.

Characteristics:

- Draw one plot at a time.
- Can be combined with Matplotlib subplots.
- Offer greater customization.
- Return a Matplotlib Axes object.
- Ideal for dashboards and custom layouts.

Common Axes-Level Functions:

- scatterplot()
- lineplot()
- barplot()
- histplot()
- boxplot()
- violinplot()
- heatmap()

#### What are Figure-Level Functions?

Figure-Level functions automatically create a Figure and one or more Axes objects.

Characteristics:

- Automatically manage layout.
- Support faceting using rows and columns.
- Return a Seaborn Grid object.
- Easier for grouped visualizations.
- Require less code for multiple plots.

Common Figure-Level Functions:

- relplot()
- catplot()
- displot()
- lmplot()
- pairplot()
- jointplot()
- clustermap()

#### Importing libraries and loading the data

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
````

#### Example of an Axes-Level Function

The `scatterplot()` function is an Axes-Level function.

It draws a plot on the current Matplotlib Axes.

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Axes-Level Function")

plt.show()
````

#### Using Axes-Level Functions with Subplots

Axes-Level functions work perfectly with Matplotlib subplots because they allow us to specify the Axes object.

````python
fig, ax = plt.subplots(1,2,figsize=(12,5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=ax[0]
)

ax[0].set_title("Scatter Plot")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=ax[1]
)

ax[1].set_title("Box Plot")

plt.tight_layout()

plt.show()
````

#### Example of a Figure-Level Function

The `relplot()` function is a Figure-Level function.

Unlike `scatterplot()`, it automatically creates its own Figure and Axes.

````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="scatter"
)

plt.show()
````

#### Faceting with Figure-Level Functions

One of the biggest advantages of Figure-Level functions is the ability to create multiple plots automatically using categories.

````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="time"
)

plt.show()
````

````python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    col="sex",
    kind="bar"
)

plt.show()
````

#### Return Types

Axes-Level Functions return:

```python
matplotlib.axes.Axes
```

Figure-Level Functions return:

```python
seaborn.axisgrid.FacetGrid
```

This is why Figure-Level functions provide additional features such as faceting and automatic layouts.
