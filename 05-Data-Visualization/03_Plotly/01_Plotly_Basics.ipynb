#### Loading a Built-in Dataset

Plotly Express provides several built-in datasets that are useful for learning and experimentation.

In this notebook, we will use the **tips** dataset.

```python
tips = px.data.tips()

tips.head()
```

```python
tips.info()
```

```python
tips.describe()
```

#### Creating Your First Interactive Plot

The `scatter()` function creates an interactive scatter plot.

Unlike Matplotlib, Plotly charts allow users to zoom, pan, and hover over data points without additional code.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip"
)

fig.show()
```

#### Adding Colors

The `color` parameter groups observations using different colors.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex"
)

fig.show()
```

#### Adding Hover Information

Hover labels display additional information when the mouse moves over a data point.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    hover_data=["day", "time", "size"]
)

fig.show()
```

#### Customizing Titles and Labels

Plotly Express allows us to customize chart titles and axis labels using simple parameters.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    title="Total Bill vs Tip",
    labels={
        "total_bill": "Total Bill ($)",
        "tip": "Tip ($)"
    }
)

fig.show()
```
