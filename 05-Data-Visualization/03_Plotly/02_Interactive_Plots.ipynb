# Interactive Features in Plotly Express

One of the biggest advantages of Plotly Express is that every visualization is interactive by default. Unlike traditional plotting libraries, Plotly allows users to interact with charts without writing additional code.

Users can zoom into specific areas, pan across the plot, hover over data points to view detailed information, toggle legend items, and export charts as images directly from the visualization toolbar.

These interactive features make Plotly an excellent choice for Exploratory Data Analysis (EDA), dashboards, business intelligence, and data storytelling.

#### Importing library and load the data

```python
import plotly.express as px

tips = px.data.tips()
```

#### Creating an Interactive Plot

Every Plotly Express chart is interactive by default.

Users can:

- Hover over data points
- Zoom in and out
- Pan across the chart
- Select regions
- Download the figure

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    title="Interactive Scatter Plot"
)

fig.show()
```

#### Hover Information

When the mouse pointer moves over a data point, Plotly automatically displays useful information such as the x-value, y-value, and any additional variables.

Hover labels make it easy to inspect individual observations without cluttering the visualization.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    hover_data=["day", "time", "size"]
)

fig.show()
```

#### Hover Name

The `hover_name` parameter highlights an important variable at the top of the tooltip.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    hover_name="day",
    color="sex"
)

fig.show()
```

#### Interactive Legend

Plotly legends are interactive.

You can:

- Click a legend item to hide or show a category.
- Double-click a legend item to display only that category.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="day",
    title="Interactive Legend"
)

fig.show()
```

#### Zooming and Panning

Every Plotly chart includes built-in tools for:

- Zoom In
- Zoom Out
- Pan
- Box Select
- Lasso Select
- Reset Axes

These controls are available in the toolbar displayed when you hover over the chart.

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    size="size"
)

fig.show()
```

#### Exporting Charts

The Plotly toolbar allows users to download the current visualization as a PNG image.

This makes it easy to include charts in reports, presentations, and documentation.

```python
fig = px.bar(
    tips,
    x="day",
    y="total_bill",
    color="sex",
    title="Download This Chart from the Toolbar"
)

fig.show()
```
