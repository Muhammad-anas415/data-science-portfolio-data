#### Line Plot

The following example demonstrates how to create a **Line Plot** to visualize trends or changes over an ordered sequence, such as time, months, or years.

**Purpose:**
- Display trends and patterns.
- Show changes over time.
- Compare continuous numerical values.

**X-axis:** Represents the independent variable (e.g., Month, Year, Time).

**Y-axis:** Represents the dependent numerical variable (e.g., Sales, Revenue, Temperature).


#### Matplotlib 

```python
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Create Figure
plt.figure(figsize=(8,5))

# Line Plot
plt.plot(
    months,
    sales,
    color="blue",
    marker="o"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Line Plot
sns.lineplot(
    x=months,
    y=sales,
    color="green",
    marker="o"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 200, 230]
})

# Line Plot
fig = px.line(
    df,
    x="Month",
    y="Sales",
    title="Monthly Sales",
    markers=True,
    color_discrete_sequence=["red"]
)

fig.show()
```
