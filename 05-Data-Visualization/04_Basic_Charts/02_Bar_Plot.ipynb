#### Bar Plot

The following example demonstrates how to create a **Bar Plot** to compare numerical values across different categories.

**Purpose:**
- Compare values between different categories.
- Identify the highest and lowest values.
- Visualize categorical data clearly.

**X-axis:** Represents the categorical variable (e.g., Product, Month, Department, Category).

**Y-axis:** Represents the numerical variable (e.g., Sales, Revenue, Count, Profit).

#### Matplotlib

```python
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Create Figure
plt.figure(figsize=(8, 5))

# Bar Plot
plt.bar(
    months,
    sales,
    color="steelblue"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Bar Plot
sns.barplot(
    x=months,
    y=sales,
    color="orange"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 200, 230]
})

# Bar Plot
fig = px.bar(
    df,
    x="Month",
    y="Sales",
    title="Monthly Sales",
    color="Month",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.show()
```
