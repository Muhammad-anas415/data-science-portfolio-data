#### Horizontal Bar Plot

The following example demonstrates how to create a **Horizontal Bar Plot** to compare numerical values across different categories.

**Purpose:**
- Compare values between categories when category names are long.
- Improve readability for datasets with many categories.
- Clearly identify the highest and lowest values.

**X-axis:** Represents the numerical variable (e.g., Sales, Revenue, Profit, Count).

**Y-axis:** Represents the categorical variable (e.g., Product, Department, Category, Month).

#### Matplotlib

```python
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Create Figure
plt.figure(figsize=(8,5))

# Horizontal Bar Plot
plt.barh(
    months,
    sales,
    color="teal"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Sales")
plt.ylabel("Month")

plt.show()
```

#### Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 230]

# Horizontal Bar Plot
sns.barplot(
    x=sales,
    y=months,
    color="coral"
)

# Labels
plt.title("Monthly Sales")
plt.xlabel("Sales")
plt.ylabel("Month")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 200, 230]
})

# Horizontal Bar Plot
fig = px.bar(
    df,
    x="Sales",
    y="Month",
    orientation="h",
    color="Month",
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_layout(
    title="Monthly Sales",
    xaxis_title="Sales",
    yaxis_title="Month"
)

fig.show()
```
