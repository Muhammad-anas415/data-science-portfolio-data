#### Grouped Bar Plot

The following example demonstrates how to create a **Grouped Bar Plot** to compare multiple categories side by side.

**Purpose:**
- Compare two or more groups within the same category.
- Identify differences between categories.
- Make comparisons easier using grouped bars.

**X-axis:** Represents the main categorical variable (e.g., Month, Department, Product).

**Y-axis:** Represents the numerical variable (e.g., Sales, Revenue, Profit).

**Grouped Variable:** Represents the second category that separates the bars into different groups (e.g., Product Type, Gender, Region).

#### Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a = [120, 150, 170, 160, 200, 230]
product_b = [100, 140, 160, 180, 190, 210]

# Position of bars
x = np.arange(len(months))
width = 0.35

plt.figure(figsize=(8,5))

plt.bar(
    x - width/2,
    product_a,
    width=width,
    color="steelblue",
    label="Product A"
)

plt.bar(
    x + width/2,
    product_b,
    width=width,
    color="orange",
    label="Product B"
)

plt.xticks(x, months)

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.show()
```

#### Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan","Jan","Feb","Feb","Mar","Mar",
              "Apr","Apr","May","May","Jun","Jun"],
    "Product": ["A","B"]*6,
    "Sales": [120,100,150,140,170,160,160,180,200,190,230,210]
})

plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Month",
    y="Sales",
    hue="Product",
    palette="Set2"
)

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan","Jan","Feb","Feb","Mar","Mar",
              "Apr","Apr","May","May","Jun","Jun"],
    "Product": ["A","B"]*6,
    "Sales": [120,100,150,140,170,160,160,180,200,190,230,210]
})

fig = px.bar(
    df,
    x="Month",
    y="Sales",
    color="Product",
    barmode="group",
    title="Monthly Sales Comparison",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.show()
```
