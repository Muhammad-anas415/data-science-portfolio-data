#### Stacked Bar Plot

The following example demonstrates how to create a **Stacked Bar Plot** to visualize the total value of each category while also showing the contribution of different groups.

**Purpose:**
- Show the total value of each category.
- Display the contribution of multiple groups within each category.
- Compare both individual group contributions and overall totals.

**X-axis:** Represents the categorical variable (e.g., Month, Department, Product).

**Y-axis:** Represents the numerical variable (e.g., Sales, Revenue, Profit).

**Stacked Variable:** Represents the different groups that are stacked together (e.g., Product A, Product B, Region, Gender).

#### Matplotlib

```python
import matplotlib.pyplot as plt

# Sample Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = [120, 150, 170, 160, 200, 230]
product_b = [100, 140, 160, 180, 190, 210]

plt.figure(figsize=(8,5))

plt.bar(
    months,
    product_a,
    color="steelblue",
    label="Product A"
)

plt.bar(
    months,
    product_b,
    bottom=product_a,
    color="orange",
    label="Product B"
)

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.show()
```

#### Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Data
df = pd.DataFrame({
    "Product A": [120,150,170,160,200,230],
    "Product B": [100,140,160,180,190,210]
},
index=["Jan","Feb","Mar","Apr","May","Jun"])

df.plot(
    kind="bar",
    stacked=True,
    figsize=(8,5),
    color=["steelblue","orange"]
)

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Month": ["Jan","Jan","Feb","Feb","Mar","Mar",
              "Apr","Apr","May","May","Jun","Jun"],
    "Product": ["A","B"]*6,
    "Sales": [120,100,150,140,170,160,160,180,200,190,230,210]
})

fig = px.bar(
    df,
    x="Month",
    y="Sales",
    color="Product",
    barmode="stack",
    title="Monthly Sales Comparison",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.show()
```
