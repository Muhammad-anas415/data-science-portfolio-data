#### Pie Chart

The following example demonstrates how to create a **Pie Chart** to show how each category contributes to the whole.

**Purpose:**
- Display the proportion or percentage of each category.
- Visualize part-to-whole relationships.
- Compare the relative size of different categories.

**Categories:** Represent different groups (e.g., Products, Departments, Regions).

**Values:** Represent the numerical contribution of each category (e.g., Sales, Revenue, Population).

#### Matplotlib

```python
import matplotlib.pyplot as plt

# Sample Data
products = ["Product A", "Product B", "Product C", "Product D"]
sales = [35, 25, 20, 20]

plt.figure(figsize=(6,6))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    colors=["skyblue", "orange", "lightgreen", "tomato"],
    startangle=90
)

plt.title("Sales Distribution by Product")

plt.show()
```

#### Plotly

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Product": ["Product A", "Product B", "Product C", "Product D"],
    "Sales": [35, 25, 20, 20]
})

fig = px.pie(
    df,
    names="Product",
    values="Sales",
    title="Sales Distribution by Product",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.show()
```
