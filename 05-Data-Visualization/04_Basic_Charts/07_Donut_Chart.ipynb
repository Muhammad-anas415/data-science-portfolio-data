#### Doughnut Chart

The following example demonstrates how to create a **Doughnut Chart** to visualize the proportion of different categories.

**Purpose:**
- Display part-to-whole relationships.
- Show the percentage contribution of each category.
- Provide a cleaner alternative to a traditional pie chart.

**Categories:** Represent different groups (e.g., Products, Departments, Regions).

**Values:** Represent the numerical contribution of each category (e.g., Sales, Revenue, Population).

#### Matplotlib

```python
import matplotlib.pyplot as plt

# Sample Data
products = ["Product A", "Product B", "Product C", "Product D"]
sales = [35, 25, 20, 20]

plt.figure(figsize=(6,6))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90,
    colors=["skyblue", "orange", "lightgreen", "tomato"],
    wedgeprops={"width": 0.4}
)

plt.title("Sales Distribution by Product")

plt.show()
```

#### Seaborn

```python
import plotly.express as px
import pandas as pd

# Sample Data
df = pd.DataFrame({
    "Product": ["Product A", "Product B", "Product C", "Product D"],
    "Sales": [35, 25, 20, 20]
})

fig = px.pie(
    df,
    names="Product",
    values="Sales",
    hole=0.4,
    title="Sales Distribution by Product",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.show()
```
