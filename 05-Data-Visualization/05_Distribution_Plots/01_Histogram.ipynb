#### Histogram

The following example demonstrates how to create a **Histogram** to visualize the distribution of a numerical variable.

**Purpose:**
- Understand the distribution of data.
- Identify the frequency of values.
- Detect skewness, spread, and possible outliers.
- Observe the shape of the distribution (Normal, Left-Skewed, Right-Skewed, etc.).

**X-axis:** Represents the numerical variable (e.g., Total Bill, Age, Salary).

**Y-axis:** Represents the frequency (number of observations) in each interval (bin).

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

# Create Figure
plt.figure(figsize=(8,5))

# Histogram
plt.hist(
    tips["total_bill"],
    bins=15,
    color="steelblue",
    edgecolor="black"
)

# Labels
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")

plt.show()
```

####  Seaborn

```python
# Histogram
sns.histplot(
    data=tips,
    x="total_bill",
    bins=15,
    color="orange",
    edgecolor="black"
)

# Labels
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")

plt.show()
```

#### Plotly

```python
import plotly.express as px

# Histogram
fig = px.histogram(
    tips,
    x="total_bill",
    nbins=15,
    title="Distribution of Total Bill",
    color_discrete_sequence=["seagreen"]
)

fig.update_layout(
    xaxis_title="Total Bill ($)",
    yaxis_title="Frequency"
)

fig.show()
```
