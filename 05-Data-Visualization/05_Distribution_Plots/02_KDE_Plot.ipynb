#### KDE (Kernel Density Estimation) Plot

The following example demonstrates how to create a **Kernel Density Estimation (KDE) Plot** to visualize the distribution of a numerical variable using a smooth density curve.

**Purpose:**
- Understand the overall distribution of data.
- Identify peaks and patterns in the data.
- Compare multiple distributions.
- Estimate the probability density of a continuous variable.

**X-axis:** Represents the numerical variable (e.g., Total Bill, Age, Salary).

**Y-axis:** Represents the estimated probability density of the data.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import gaussian_kde

# Load Dataset
tips = sns.load_dataset("tips")

# KDE Calculation
x = tips["total_bill"]
kde = gaussian_kde(x)
x_range = np.linspace(x.min(), x.max(), 200)

# Plot
plt.figure(figsize=(8,5))

plt.plot(
    x_range,
    kde(x_range),
    color="steelblue",
    linewidth=2
)

plt.fill_between(
    x_range,
    kde(x_range),
    color="steelblue",
    alpha=0.3
)

plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Density")

plt.show()
```

#### Seaborn

```python
import seaborn as sns

plt.figure(figsize=(8,5))

sns.kdeplot(
    data=tips,
    x="total_bill",
    fill=True,
    color="darkorange"
)

plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Density")

plt.show()
```

#### Plotly

```python
import plotly.figure_factory as ff

fig = ff.create_distplot(
    [tips["total_bill"]],
    group_labels=["Total Bill"],
    show_hist=False,
    show_rug=False
)

fig.update_layout(
    title="KDE Plot of Total Bill",
    xaxis_title="Total Bill ($)",
    yaxis_title="Density"
)

fig.show()
```
