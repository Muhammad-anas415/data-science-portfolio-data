####  Box Plot

The following example demonstrates how to create a **Box Plot** to summarize the distribution of a numerical variable.

**Purpose:**
- Visualize the distribution of data.
- Identify the median and quartiles.
- Detect outliers.
- Compare distributions across different categories.

**X-axis:** Represents the categorical variable (e.g., Day, Gender, Smoker).

**Y-axis:** Represents the numerical variable (e.g., Total Bill, Tip, Salary).

A box plot displays the **minimum value, first quartile (Q1), median (Q2), third quartile (Q3), maximum value, and potential outliers**, making it one of the most informative charts for exploratory data analysis (EDA).

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

# Create Figure
plt.figure(figsize=(8,5))

# Box Plot
plt.boxplot(
    tips["total_bill"],
    patch_artist=True,
    boxprops=dict(facecolor="skyblue")
)

# Labels
plt.title("Box Plot of Total Bill")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Seaborn

```python
import seaborn as sns
plt.figure(figsize=(8,5))

# Box Plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    color="lightgreen"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Plotly

```python
import plotly.express as px

# Box Plot
fig = px.box(
    tips,
    x="day",
    y="total_bill",
    color="day",
    title="Total Bill by Day",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    xaxis_title="Day",
    yaxis_title="Total Bill ($)"
)

fig.show()
```
