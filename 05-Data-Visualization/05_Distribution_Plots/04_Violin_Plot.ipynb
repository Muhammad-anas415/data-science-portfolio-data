####  Violin Plot

The following example demonstrates how to create a **Violin Plot** to visualize both the distribution and density of a numerical variable across different categories.

**Purpose:**
- Compare the distribution of multiple groups.
- Visualize the probability density of the data.
- Display the median and spread of the data.
- Identify whether the distribution is symmetric, skewed, or multimodal.

**X-axis:** Represents the categorical variable (e.g., Day, Gender, Department).

**Y-axis:** Represents the numerical variable (e.g., Total Bill, Salary, Revenue).

A violin plot combines the features of a **Box Plot** and a **Kernel Density Estimation (KDE) Plot**, making it one of the most informative distribution visualizations.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

# Prepare Data
data = [
    tips[tips["day"] == "Thur"]["total_bill"],
    tips[tips["day"] == "Fri"]["total_bill"],
    tips[tips["day"] == "Sat"]["total_bill"],
    tips[tips["day"] == "Sun"]["total_bill"]
]

# Create Figure
plt.figure(figsize=(8,5))

# Violin Plot
plt.violinplot(
    data,
    showmeans=True,
    showmedians=True
)

plt.xticks([1,2,3,4], ["Thur", "Fri", "Sat", "Sun"])

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Seaborn

```python
import seaborn as sns
plt.figure(figsize=(8,5))

# Violin Plot
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    palette="Set2"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### PLotly

```python
import plotly.express as px
# Violin Plot
fig = px.violin(
    tips,
    x="day",
    y="total_bill",
    color="day",
    box=True,
    points="all",
    title="Total Bill by Day",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    xaxis_title="Day",
    yaxis_title="Total Bill ($)"
)

fig.show()
```
