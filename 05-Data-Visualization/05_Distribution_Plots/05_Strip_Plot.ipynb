#### Strip Plot

The following example demonstrates how to create a **Strip Plot** to visualize individual observations across different categories.

**Purpose:**
- Display every data point.
- Compare the distribution of observations across categories.
- Identify clusters, gaps, and potential outliers.
- Explore the spread of small to medium-sized datasets.

**X-axis:** Represents the categorical variable (e.g., Day, Gender, Department).

**Y-axis:** Represents the numerical variable (e.g., Total Bill, Salary, Revenue).

A strip plot displays each observation as a separate point. To reduce overlapping points, a small amount of random displacement (jitter) can be added.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

plt.figure(figsize=(8,5))

# Strip Plot
days = ["Thur", "Fri", "Sat", "Sun"]

for i, day in enumerate(days):
    data = tips[tips["day"] == day]["total_bill"]
    plt.scatter(
        [i] * len(data),
        data,
        color="steelblue",
        alpha=0.7
    )

plt.xticks(range(len(days)), days)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

####  Seaborn

```python
import seaborn as sns
plt.figure(figsize=(8,5))

# Strip Plot
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    jitter=True,
    palette="Set2"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Plotly

```python
import plotly.express as px
# Strip Plot
fig = px.strip(
    tips,
    x="day",
    y="total_bill",
    color="sex",
    title="Total Bill by Day",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    xaxis_title="Day",
    yaxis_title="Total Bill ($)"
)

fig.show()
```
