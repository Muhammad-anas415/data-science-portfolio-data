#### Swarm Plot

The following example demonstrates how to create a **Swarm Plot** to visualize individual observations across different categories without overlapping points.

**Purpose:**
- Display every individual observation.
- Prevent overlapping points for better readability.
- Compare distributions across different categories.
- Identify clusters, gaps, and potential outliers.

**X-axis:** Represents the categorical variable (e.g., Day, Gender, Department).

**Y-axis:** Represents the numerical variable (e.g., Total Bill, Salary, Revenue).

A swarm plot arranges data points so that they do not overlap, making it easier to observe the distribution of values compared to a strip plot.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

# Prepare Data
days = ["Thur", "Fri", "Sat", "Sun"]

plt.figure(figsize=(8,5))

for i, day in enumerate(days):
    data = tips[tips["day"] == day]["total_bill"]
    plt.scatter(
        [i] * len(data),
        data,
        color="steelblue",
        alpha=0.7
    )

plt.xticks(range(len(days)), days)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Seaborn

```python
plt.figure(figsize=(8,5))

sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="Set2"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

#### Plotly

```python
import plotly.express as px
fig = px.strip(
    tips,
    x="day",
    y="total_bill",
    color="sex",
    title="Total Bill by Day",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_traces(jitter=0)

fig.update_layout(
    xaxis_title="Day",
    yaxis_title="Total Bill ($)"
)

fig.show()
```
