#### Scatter Plot

The following example demonstrates how to create a **Scatter Plot** to visualize the relationship between two numerical variables.

**Purpose:**
- Identify relationships between variables.
- Detect positive or negative correlations.
- Find clusters and patterns.
- Detect potential outliers.

**X-axis:** Represents the first numerical variable (e.g., Total Bill, Age, Salary).

**Y-axis:** Represents the second numerical variable (e.g., Tip, Income, Revenue).

Each point in the scatter plot represents one observation in the dataset.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
tips = sns.load_dataset("tips")

# Create Figure
plt.figure(figsize=(8,5))

# Scatter Plot
plt.scatter(
    tips["total_bill"],
    tips["tip"],
    color="steelblue",
    alpha=0.7
)

# Labels
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

plt.show()
```

#### Seaborn

```python
plt.figure(figsize=(8,5))
# Scatter Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    palette="Set2"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

plt.show()
```

#### Pyplot

```python
import plotly.express as px
# Scatter Plot
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    title="Total Bill vs Tip",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    xaxis_title="Total Bill ($)",
    yaxis_title="Tip ($)"
)

fig.show()
```
