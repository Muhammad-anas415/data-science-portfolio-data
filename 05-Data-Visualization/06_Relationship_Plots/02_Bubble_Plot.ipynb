#### Bubble Plot

The following example demonstrates how to create a **Bubble Plot** to visualize the relationship between two numerical variables while representing an additional variable using bubble size.

**Purpose:**
- Visualize relationships between multiple numerical variables.
- Compare the magnitude of a third variable using bubble size.
- Identify patterns and trends in the dataset.
- Understand how different groups affect the relationship between variables.

**X-axis:** Represents the first numerical variable (e.g., Total Bill).

**Y-axis:** Represents the second numerical variable (e.g., Tip Amount).

**Bubble Size:** Represents the magnitude of another numerical variable (e.g., Number of people in the group).

**Color:** Represents different categories in the dataset (e.g., Gender, Day, or Time).


#### Matplotlib


```python
import matplotlib.pyplot as plt
import seaborn as sns
# Load Dataset
tips = sns.load_dataset("tips")
# Create Figure
plt.figure(figsize=(8,5))
# Bubble Plot
plt.scatter(
    tips["total_bill"],
    tips["tip"],
    s=tips["size"]*100,
    c=tips["size"],
    cmap="viridis",
    alpha=0.7
)
# Labels
plt.title("Total Bill vs Tip - Bubble Plot")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
# Color Bar
plt.colorbar(
    label="Group Size"
)
plt.show()
```

#### Seaborn

```python
# Create Figure
plt.figure(figsize=(8,5))
# Bubble Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    size="size",
    hue="sex",
    sizes=(50,500),
    alpha=0.7
)
# Labels
plt.title("Total Bill vs Tip - Seaborn Bubble Plot")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

#### Plotly

```python
import plotly.express as px
# Bubble Plot
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    size="size",
    color="sex",
    hover_name="day",
    title="Total Bill vs Tip - Plotly Bubble Plot"
)
# Display Plot
fig.show()
```
