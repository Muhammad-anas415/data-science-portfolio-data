#### Line Plot

The following example demonstrates how to create a **Line Plot** to visualize the relationship and trends between two variables.

**Purpose:**
- Analyze trends and patterns over continuous values.
- Understand how one variable changes with another variable.
- Identify increasing or decreasing relationships.
- Compare changes over time or ordered data.


**X-axis:** Represents the independent variable (e.g., Day, Time, Age, Year).

**Y-axis:** Represents the dependent variable (e.g., Sales, Revenue, Average Tip).


A line plot connects data points to show the direction and pattern of change between variables.

Line plots are commonly used for:
- Time-series analysis.
- Trend analysis.
- Performance monitoring.
- Data exploration.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Load Dataset
tips = sns.load_dataset("tips")
# Group data by day
daily_tip = tips.groupby("day")["tip"].mean()
# Create Figure
plt.figure(figsize=(8,5))
# Line Plot
plt.plot(
    daily_tip.index,
    daily_tip.values,
    marker="o",
    color="steelblue",
    linewidth=2
)
# Labels
plt.title("Average Tip Amount by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip ($)")
# Grid
plt.grid(True)
plt.show()
```

#### Seaborn

```python
# Create Figure
plt.figure(figsize=(8,5))
# Line Plot
sns.lineplot(
    data=tips,
    x="day",
    y="tip",
    marker="o",
    errorbar=None
)
# Labels
plt.title("Average Tip Amount by Day - Seaborn")
plt.xlabel("Day")
plt.ylabel("Tip ($)")
# Grid
plt.grid(True)
plt.show()
```

#### Plotly

```python
import plotly.express as px
# Calculate average tip per day
daily_tip = (
    tips.groupby("day")["tip"]
    .mean()
    .reset_index()
)
# Create Line Plot
fig = px.line(
    daily_tip,
    x="day",
    y="tip",
    markers=True,
    title="Average Tip Amount by Day - Plotly"
)
# Display Plot
fig.show()
```
