#### Regression Plot

The following example demonstrates how to create a **Regression Plot** to visualize the relationship between two numerical variables along with a fitted regression line.

A regression plot helps to understand the **linear relationship** between variables and shows the general trend in the data.


**Purpose:**
- Identify the relationship between two numerical variables.
- Understand positive or negative linear trends.
- Estimate the direction of change between variables.
- Observe how well data points fit a regression line.
- Detect possible outliers.


**X-axis:** Represents the independent variable (e.g., Total Bill).

**Y-axis:** Represents the dependent variable (e.g., Tip Amount).


The regression line represents the best-fit line that shows the overall trend of the relationship between variables.

Regression plots are commonly used in:
- Exploratory Data Analysis (EDA).
- Correlation analysis.
- Machine Learning regression problems.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
# Load Dataset
tips = sns.load_dataset("tips")
# Select Variables
x = tips["total_bill"]
y = tips["tip"]
# Calculate Regression Line
m, b = np.polyfit(x, y, 1)
# Create Figure
plt.figure(figsize=(8,5))
# Scatter Points
plt.scatter(
    x,
    y,
    color="steelblue",
    alpha=0.6
)
# Regression Line
plt.plot(
    x,
    m*x+b,
    color="red",
    linewidth=2
)
# Labels
plt.title("Total Bill vs Tip - Regression Plot")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

#### Seaborn

```python
# Create Figure
plt.figure(figsize=(8,5))
# Regression Plot
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter_kws={
        "color":"steelblue",
        "alpha":0.6
    },
    line_kws={
        "color":"red"
    }
)
# Labels
plt.title("Total Bill vs Tip - Seaborn Regression Plot")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

####  Plotly

```python
import plotly.express as px
# Create Scatter Plot
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    trendline="ols",
    title="Total Bill vs Tip - Plotly Regression Plot"
)
# Display Plot
fig.show()
```
