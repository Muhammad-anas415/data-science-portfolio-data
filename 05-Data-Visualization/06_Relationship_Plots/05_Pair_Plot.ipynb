#### Pair Plot

The following example demonstrates how to create a **Pair Plot** to visualize relationships between multiple numerical variables in a dataset.

A Pair Plot creates multiple scatter plots between numerical features and also displays the distribution of each variable on the diagonal.


**Purpose:**
- Analyze relationships between multiple numerical variables.
- Identify correlations between features.
- Detect patterns, clusters, and outliers.
- Understand feature distributions.
- Perform Exploratory Data Analysis (EDA).


**Diagonal plots:** Represent the distribution of individual numerical variables.

**Off-diagonal plots:** Represent relationships between two numerical variables.


Pair plots are commonly used during the initial stage of data analysis to understand the structure of a dataset.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
# Load Dataset
tips = sns.load_dataset("tips")
# Select Numerical Features
data = tips[
    [
        "total_bill",
        "tip",
        "size"
    ]
]
# Create Pair Plot using pandas plotting
pd.plotting.scatter_matrix(
    data,
    figsize=(8,8),
    alpha=0.7
)
# Title
plt.suptitle(
    "Pair Plot Using Matplotlib"
)
plt.show()
```

#### Seaborn

```python
# create Pair Plot
sns.pairplot(
    tips,
    vars=[
        "total_bill",
        "tip",
        "size"
    ],
    hue="sex"
)
# Title
plt.suptitle(
    "Pair Plot Using Seaborn",
    y=1.02
)
plt.show()
```

#### Plotly

```python
import plotly.express as px

# Create Scatter Matrix

fig = px.scatter_matrix(
    tips,
    dimensions=[
        "total_bill",
        "tip",
        "size"
    ],
    color="sex",
    title="Interactive Pair Plot Using Plotly"
)
# Display Plot
fig.show()
```
