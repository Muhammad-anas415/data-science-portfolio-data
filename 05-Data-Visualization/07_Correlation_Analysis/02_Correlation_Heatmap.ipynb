#### Correlation Heatmap

The following example demonstrates how to create a **Correlation Heatmap** to visualize the relationship between numerical variables using colors.

A correlation heatmap displays the correlation matrix in a visual format, making it easier to identify positive, negative, and weak relationships between features.


**Purpose:**
- Understand relationships between numerical variables.
- Identify highly correlated features.
- Detect multicollinearity before Machine Learning.
- Support feature selection decisions.
- Quickly analyze large numbers of variables.


**Correlation Value Range:**

- **+1** → Strong positive relationship.
- **0** → No linear relationship.
- **-1** → Strong negative relationship.


**Color Representation:**

- Strong positive correlation → Similar color intensity.
- Strong negative correlation → Opposite color intensity.
- Values near zero → Weak relationship.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Load Dataset
tips = sns.load_dataset("tips")
# Select Numerical Features
corr = tips[
    [
        "total_bill",
        "tip",
        "size"
    ]
].corr()
# Create Figure
plt.figure(figsize=(7,5))
# Display Correlation Matrix
plt.imshow(
    corr,
    cmap="coolwarm"
)
# Add Color Bar
plt.colorbar()
# Add Labels
plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=45
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)
# Title
plt.title(
    "Correlation Heatmap Using Matplotlib"
)
plt.show()
```

#### seaborn

```python
# Calculate Correlation
corr = tips[
    [
        "total_bill",
        "tip",
        "size"
    ]
].corr()
# Create Figure
plt.figure(figsize=(7,5))
# Correlation Heatmap
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
# Title
plt.title(
    "Correlation Heatmap Using Seaborn"
)
plt.show()
```

#### Plotly

```python
import plotly.express as px
import seaborn as sns
# Load Dataset
tips = sns.load_dataset("tips")
# Calculate Correlation
corr = tips[
    [
        "total_bill",
        "tip",
        "size"
    ]
].corr()
# Create Interactive Heatmap
fig = px.imshow(
    corr,
    text_auto=True,
    color_continuous_scale="RdBu",
    title="Interactive Correlation Heatmap Using Plotly"
)
# Display Plot
fig.show()
```
