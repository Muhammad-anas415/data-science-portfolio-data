#### Covariance Visualization

The following example demonstrates how to visualize **Covariance** between numerical variables.

Covariance measures the direction of the relationship between two variables and shows whether they increase or decrease together.


**Purpose:**
- Understand how two variables change together.
- Identify positive or negative relationships.
- Compare the movement of numerical features.
- Support feature analysis during Exploratory Data Analysis (EDA).


**Covariance Value Interpretation:**

- **Positive Covariance (+)**:
  - Both variables increase or decrease together.

- **Negative Covariance (-)**:
  - One variable increases while the other decreases.

- **Covariance near 0**:
  - Variables have little or no linear relationship.

#### Matplotlib

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Load Dataset
tips = sns.load_dataset("tips")
# Select Numerical Columns
data = tips[
    [
        "total_bill",
        "tip",
        "size"
    ]
]
# Calculate Covariance Matrix
cov_matrix = data.cov()
# Create Figure
plt.figure(figsize=(7,5))
# Display Covariance Matrix
plt.imshow(
    cov_matrix,
    cmap="coolwarm"
)
# Color Bar
plt.colorbar()
# Labels
plt.xticks(
    range(len(cov_matrix.columns)),
    cov_matrix.columns,
    rotation=45
)
plt.yticks(
    range(len(cov_matrix.columns)),
    cov_matrix.columns
)
# Title
plt.title(
    "Covariance Visualization Using Matplotlib"
)
plt.show()
```

#### Seaborn

```python
# Create Figure
plt.figure(figsize=(7,5))
# Covariance Heatmap
sns.heatmap(
    cov_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
# Title
plt.title(
    "Covariance Heatmap Using Seaborn"
)
plt.show()
```

#### Plotly

```python
import plotly.express as px
# Interactive Heatmap
fig = px.imshow(
    cov_matrix,
    text_auto=True,
    color_continuous_scale="RdBu",
    title="Interactive Covariance Heatmap Using Plotly"
)
# Display Plot
fig.show()
```
