### Data Cleaning

Data Cleaning is the process of identifying and correcting inaccurate, incomplete, duplicate, or inconsistent data within a dataset. It is one of the most critical steps in the Data Science workflow because the quality of the data directly impacts the accuracy of analysis and machine learning models.

Real-world datasets often contain missing values, duplicate records, incorrect data types, inconsistent formatting, and invalid entries. If these issues are not addressed, they can lead to misleading insights, poor model performance, and incorrect business decisions.

#### Importing required libraries for data cleaning


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

#### Load the Dataset

The Titanic dataset is one of the most widely used datasets for learning Data Science and Machine Learning. It contains information about passengers aboard the RMS Titanic, including demographic details, ticket information, travel class, fare, and survival status.

After loading the dataset, we display the first few rows to verify that the data has been loaded successfully.

```python
# Load Titanic Dataset
df = sns.load_dataset("titanic")
# Display first five rows
df.head()
```

#### Create a Copy of the Dataset

Before performing any cleaning operations, it is recommended to create a copy of the original dataset. This ensures that the original data remains unchanged and can be used for comparison or recovery if needed.

```python
# Create a copy of the dataset
df = df.copy()
df.head()
```

#### Check Missing Values

Missing values represent unavailable or unknown information within a dataset. They may occur due to human error, equipment failure, incomplete surveys, or data collection issues.

Identifying missing values is an essential step because they can affect statistical analysis and machine learning performance.

The `isnull()` function detects missing values, while the `sum()` function counts the number of missing values in each column.

```python
# Check missing values
df.isnull().sum()
```

####  Handle Missing Values

After identifying missing values, we need to decide how to handle them. Common approaches include:

- Removing rows containing missing values.
- Removing columns with excessive missing values.
- Filling missing values using the mean.
- Filling missing values using the median.
- Filling missing values using the mode.

The appropriate method depends on the nature of the dataset and the feature being cleaned.

```python
# Fill missing Age values with the median
df["age"] = df["age"].fillna(df["age"].median())

# Fill missing Embarked values with the mode
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

# Drop the Deck column because it contains many missing values
df.drop(columns="deck", inplace=True)

# Verify remaining missing values
df.isnull().sum()
```

#### Check Duplicate Records

Duplicate records occur when the same observation appears more than once in a dataset. These duplicates can distort statistical analysis and negatively affect machine learning models.

Before removing duplicates, it is important to determine how many duplicate rows exist.

```python
# Count duplicate rows
df.duplicated().sum()
```

####  Remove Duplicate Records

If duplicate records are present, they should be removed to improve data quality and ensure that each observation represents unique information.

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Verify duplicates
df.duplicated().sum()
```

#### Check Data Types

Every column in a dataset has a specific data type, such as integer, float, object, boolean, or category.

Checking data types helps ensure that each feature is stored correctly and allows appropriate statistical operations to be performed.

```python
# Display data types
df.dtypes

# More detailed information
df.info()
```

####  Convert Data Types

Sometimes a feature is stored using an incorrect data type. Converting columns to the appropriate type improves memory efficiency and ensures accurate analysis.

```python
# Convert Survived column to integer
df["survived"] = df["survived"].astype("int")

df.dtypes
```

####  Rename Columns

Renaming columns improves readability and makes feature names more descriptive, especially when sharing notebooks or collaborating with others.

```python
# Rename selected columns
df.rename(
    columns={
        "sibsp": "siblings_spouses",
        "parch": "parents_children"
    },
    inplace=True
)

df.columns
```

####  Handle Inconsistent Values

Text data may contain inconsistent formatting, such as extra spaces or different letter cases. Cleaning these values ensures consistency across the dataset.

```python
# Remove extra spaces
df["sex"] = df["sex"].str.strip()

# Convert to lowercase
df["sex"] = df["sex"].str.lower()

# Display unique values
df["sex"].unique()
```

#### Final Data Quality Check

After completing the cleaning process, a final inspection is performed to ensure that the dataset is free from major quality issues and is ready for Exploratory Data Analysis and Machine Learning.

```python
# Dataset Information
df.info()

# Missing Values
print(df.isnull().sum())

# Duplicate Records
print(df.duplicated().sum())

# Display cleaned dataset
df.head()
```

```python
# Save the cleaned dataset as a CSV file
df.to_csv("clean_titanic.csv", index=False)
print("✅Clean Titanic dataset saved successfully!")
```
