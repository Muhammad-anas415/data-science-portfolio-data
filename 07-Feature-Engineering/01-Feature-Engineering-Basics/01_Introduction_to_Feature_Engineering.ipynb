# Introduction to Feature Engineering

Feature Engineering is the process of transforming raw data into meaningful features that can be used effectively by machine learning algorithms.

Good feature engineering can help a machine learning model:

- Learn meaningful patterns
- Improve predictive performance
- Reduce noise
- Handle different types of data
- Make the dataset suitable for machine learning algorithms

#### What is a Feature?

A feature is an individual measurable property or characteristic of an observation.

For example, consider a dataset containing information about houses:

| Feature | Example |
|---|---:|
| Area | 1500 |
| Bedrooms | 3 |
| Bathrooms | 2 |
| Location | Lahore |
| Price | 25,000,000 |

Here, `Area`, `Bedrooms`, `Bathrooms`, and `Location` are features.

The `Price` column is the target variable that we want our model to predict.

```python
import pandas as pd
import numpy as np
s
data = {
    "area": [1200, 1500, 1800, 2200, 2500],
    "bedrooms": [2, 3, 3, 4, 5],
    "bathrooms": [1, 2, 2, 3, 3],
    "location": ["Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad"],
    "price": [12000000, 18000000, 20000000, 25000000, 30000000]
}

df = pd.DataFrame(data)

df
```

#### Types of Features

##### 1. Numerical Features

Numerical features contain numerical values.

Examples:

- Age
- Salary
- Height
- Weight
- House Area

They can be further divided into:

- Discrete features
- Continuous features

##### 2. Categorical Features

Categorical features represent groups or categories.

Examples:

- Gender
- City
- Product Category
- Department

For example:

`Lahore`, `Karachi`, and `Islamabad` are categorical values.

##### 3. Nominal Features

Nominal features contain categories that do not have a natural order.

Example:

`City = Lahore, Karachi, Islamabad`

There is no concept of one city being "greater" or "less" than another.

##### 4. Ordinal Features

Ordinal features have a meaningful order.

Example:

`Education = School < Bachelor's < Master's < PhD`

or

`Size = Small < Medium < Large`

#### Common Feature Engineering Techniques

Feature engineering includes several techniques for improving the representation of data.

##### 1. Handling Missing Values

Missing values can be handled using techniques such as:

- Mean imputation
- Median imputation
- Mode imputation
- KNN imputation
- Forward fill
- Backward fill

##### 2. Encoding Categorical Variables

Machine learning algorithms often require numerical input.

Common encoding techniques include:

- Label Encoding
- One-Hot Encoding
- Ordinal Encoding
- Frequency Encoding
- Target Encoding

##### 3. Feature Scaling

Scaling puts numerical features on comparable scales.

Common techniques:

- Standardization
- Min-Max Scaling
- Robust Scaling

### Key Takeaways

- Feature Engineering transforms raw data into useful features.
- Features are the input variables used by machine learning models.
- Features can be numerical, categorical, ordinal, nominal, date/time, or text.
- Feature Engineering can involve creating, transforming, encoding, scaling, or selecting features.
- Good features can make it easier for a machine learning model to learn meaningful patterns.
- Feature Engineering should be performed carefully to avoid data leakage.
- The best technique depends on the dataset and the machine learning algorithm.
