# Feature Types

Feature types describe the kind of information stored in a dataset's columns.

Understanding feature types is an important step in data preprocessing and feature engineering because different types of features require different preprocessing techniques.

For example:

- Numerical features may require scaling.
- Categorical features may require encoding.
- Date-time features can be transformed into useful components.
- Text features can be converted into numerical representations.

#### 1. Numerical Features

Numerical features contain numbers and represent measurable quantities.

Examples:

- Age
- Salary
- Height
- Weight
- House Price
- Number of Rooms

Numerical features are mainly divided into two types:

1. Discrete Features
2. Continuous Features

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "age": [19, 22, 25, 30, 35],
    "salary": [30000, 45000, 55000, 70000, 85000],
    "height": [165.5, 172.3, 180.2, 168.7, 175.4]
})

df
```

#### 2. Discrete Features

Discrete features contain countable numerical values.

Examples:
- Number of children
- Number of bedrooms
- Number of employees
- Number of products

Discrete values are usually whole numbers.

```python
df = pd.DataFrame({
    "number_of_bedrooms": [1, 2, 3, 4, 5],
    "number_of_children": [0, 1, 2, 3, 4],
    "number_of_products": [2, 5, 3, 8, 10]
})

df
```

#### 3. Continuous Features

Continuous features can take any numerical value within a range.

Examples:
- Height
- Weight
- Temperature
- Distance
- Salary

Continuous features usually represent measurements.

```python
df = pd.DataFrame({
    "height": [165.5, 172.3, 180.7, 168.9, 175.4],
    "weight": [60.5, 68.7, 82.4, 64.8, 75.2],
    "temperature": [36.5, 37.2, 36.8, 37.0, 36.6]
})

df
```

#### 4. Categorical Features

Categorical features represent groups or categories rather than measurable quantities.

Examples:
- Gender
- City
- Department
- Product Category
- Education Level

Categorical features are mainly divided into:
- Nominal
- Ordinal

```python
df = pd.DataFrame({
    "city": ["Lahore", "Karachi", "Islamabad", "Lahore"],
    "department": ["IT", "HR", "Finance", "IT"],
    "category": ["A", "B", "A", "C"]
})

df
```

#### 5. Nominal Features

Nominal features contain categories that do not have a meaningful order or ranking.

Examples:
- City
- Country
- Eye Color
- Blood Group
- Product Category

Example:

Lahore, Karachi, Islamabad

There is no natural order between these categories.

```python
df = pd.DataFrame({
    "city": ["Lahore", "Karachi", "Islamabad", "Lahore"],
    "blood_group": ["A+", "B+", "O+", "AB+"]
})

df
```

#### 6. Ordinal Features

Ordinal features contain categories with a meaningful order or ranking.

Examples:
- Education Level
- Satisfaction Level
- Product Size
- Rating

Example:

Small < Medium < Large

The order contains useful information.

```python
df = pd.DataFrame({
    "size": ["Small", "Medium", "Large", "Medium"],
    "satisfaction": ["Low", "Medium", "High", "High"]
})

df
```

#### 7. Binary Features

Binary features contain only two possible values.

Examples:
- Yes / No
- True / False
- 0 / 1
- Spam / Not Spam
- Pass / Fail

```python
df = pd.DataFrame({
    "is_student": [1, 0, 1, 1, 0],
    "is_employed": [True, False, True, True, False],
    "is_spam": ["Yes", "No", "No", "Yes", "No"]
})

df
```

#### 8. Date & Time Features

Date and time features contain temporal information.

Examples:
- Date of Birth
- Purchase Date
- Order Date
- Login Time
- Transaction Time

Date and time features can be transformed into:
- Year
- Month
- Day
- Day of Week
- Hour
- Weekend/Weekday

```python
df = pd.DataFrame({
    "purchase_date": [
        "2026-01-10",
        "2026-02-15",
        "2026-03-20",
        "2026-04-05"
    ]
})

df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df
```

```python
df["year"] = df["purchase_date"].dt.year
df["month"] = df["purchase_date"].dt.month
df["day"] = df["purchase_date"].dt.day
df["day_of_week"] = df["purchase_date"].dt.day_name()

df
```

#### 9. Text Features

Text features contain natural language.

Examples:
- Customer Reviews
- Emails
- Tweets
- Comments
- Product Descriptions

Raw text cannot usually be directly used by traditional machine learning models.

It can be converted into numerical features using:
- Bag of Words
- TF-IDF
- Word Embeddings
- Tokenization

```python
df = pd.DataFrame({
    "review": [
        "This product is excellent",
        "Very good quality",
        "I did not like this product",
        "Amazing experience"
    ]
})

df
```

#### 10. Target Feature

The target feature is the variable that a machine learning model tries to predict.

Examples:

House Price Prediction:
- Features → Area, Bedrooms, Location
- Target → Price

Spam Detection:
- Features → Email Text
- Target → Spam / Not Spam

```python
df = pd.DataFrame({
    "age": [20, 25, 30, 35],
    "salary": [30000, 45000, 60000, 75000],
    "experience": [1, 3, 6, 10],
    "purchased": [0, 1, 1, 1]
})

X = df.drop("purchased", axis=1)
y = df["purchased"]

X
```
