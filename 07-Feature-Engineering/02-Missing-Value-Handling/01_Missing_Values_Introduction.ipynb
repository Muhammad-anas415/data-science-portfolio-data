# Introduction to Missing Values

Missing values occur when some observations in a dataset do not contain a value for a particular feature.

Missing data is common in real-world datasets and can occur because of:

- Data entry errors
- Unanswered questions
- Sensor failures
- Data collection problems
- Merged datasets
- Information not being available

Before training a machine learning model, missing values should be identified and handled appropriately.

## Why Are Missing Values Important?

Missing values can cause several problems:

- Some machine learning algorithms cannot handle missing values.
- Statistical calculations may become inaccurate.
- Missing data can reduce the quality of the dataset.
- Removing too many rows can result in loss of important information.
- Poor handling of missing values can introduce bias.

Therefore, identifying missing values is an important step in data preprocessing.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha"],
    "Age": [20, 25, np.nan, 30, 22],
    "Salary": [30000, np.nan, 45000, 50000, np.nan],
    "City": ["Lahore", "Karachi", np.nan, "Islamabad", "Lahore"]
})

df
```

#### Detecting Missing Values

Pandas provides several useful functions for identifying missing values.

The most commonly used function is:

`isnull()`

It returns `True` when a value is missing and `False` when a value is present.

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

#### Missing Value Percentage

Counting missing values is useful, but we can also calculate the percentage of missing values in each column.

This helps us understand how much data is missing from each feature.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage
```

#### Different Representations of Missing Data

Missing data can appear in different forms depending on the dataset.

Common representations include:

- `NaN`
- `None`
- `NaT`
- Empty strings
- `?`
- `NA`
- `N/A`
- `null`

Pandas recognizes several of these automatically, while others may need to be converted into proper missing values.

```python
example = pd.DataFrame({
    "numeric": [10, np.nan, 30],
    "object": ["A", None, "C"],
    "date": [pd.Timestamp("2026-01-01"), pd.NaT, pd.Timestamp("2026-03-01")]
})

example
```

```python
example.isnull().sum()
```

#### Visualizing Missing Values

Visualization can help us quickly understand the distribution of missing data.

A simple bar chart can show the number of missing values in each column.

```python
import matplotlib.pyplot as plt

missing = df.isnull().sum()

missing[missing > 0].plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Features")
plt.ylabel("Number of Missing Values")
plt.show()
```
