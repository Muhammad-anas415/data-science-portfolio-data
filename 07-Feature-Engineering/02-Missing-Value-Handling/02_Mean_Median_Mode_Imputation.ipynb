# Mean, Median & Mode Imputation

Imputation is the process of replacing missing values with suitable values.

Three simple and commonly used imputation techniques are:

- Mean Imputation
- Median Imputation
- Mode Imputation

The choice depends on the type and distribution of the feature.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 25, np.nan, 30, 35, np.nan, 40],
    "Salary": [30000, 45000, 50000, np.nan, 70000, 80000, np.nan],
    "City": ["Lahore", "Karachi", "Lahore", np.nan, "Islamabad", "Lahore", "Karachi"]
})

df
```

#### Identifying Missing Values

Before applying imputation, we first need to identify which columns contain missing values.

```python
df.isnull().sum()
```

#### 1. Mean Imputation

Mean imputation replaces missing numerical values with the average of the available values.

### Formula

Mean = Sum of all values / Number of values

Mean imputation is generally suitable when:

- The feature is numerical.
- The data is approximately normally distributed.
- There are no significant outliers.

Mean imputation can be strongly affected by extreme values.

```python
df_mean = df.copy()

mean_age = df_mean["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

df_mean
```

```python
df_mean["Age"].isnull().sum()
```

#### 2. Median Imputation

Median imputation replaces missing numerical values with the median of the available values.

The median is the middle value after sorting the data.

Median imputation is generally useful when:

- The feature is numerical.
- The data is skewed.
- The feature contains outliers.

The median is less affected by extreme values than the mean.

```python
df_median = df.copy()

median_salary = df_median["Salary"].median()

df_median["Salary"] = df_median["Salary"].fillna(median_salary)

df_median
```

```python
df_median["Salary"].isnull().sum()
```

```python
print("Mean Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
```

#### 3. Mode Imputation

Mode imputation replaces missing values with the most frequently occurring value.

Mode imputation is commonly used for categorical features.

For example, if:

Lahore → 3 times  
Karachi → 2 times  
Islamabad → 1 time

then the mode is `Lahore`.

```python
df_mode = df.copy()

mode_city = df_mode["City"].mode()[0]

df_mode["City"] = df_mode["City"].fillna(mode_city)

df_mode
```

```python
df_mode["City"].isnull().sum()
```

#### Using Scikit-Learn SimpleImputer

Scikit-learn provides `SimpleImputer` for applying basic imputation strategies.

Available strategies include:

- `mean`
- `median`
- `most_frequent`
- `constant`

Using `SimpleImputer` is especially useful when building machine learning preprocessing pipelines.

```python
from sklearn.impute import SimpleImputer
```

```python
mean_imputer = SimpleImputer(strategy="mean")

df["Age"] = mean_imputer.fit_transform(
    df[["Age"]]
).ravel()

df
```

```python
median_imputer = SimpleImputer(strategy="median")

df["Salary"] = median_imputer.fit_transform(
    df[["Salary"]]
).ravel()

df
```

```python
mode_imputer = SimpleImputer(strategy="most_frequent")

df["City"] = mode_imputer.fit_transform(
    df[["City"]]
).ravel()

df
```
