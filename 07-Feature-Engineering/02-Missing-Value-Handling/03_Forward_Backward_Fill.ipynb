### Forward Fill & Backward Fill

Forward Fill and Backward Fill are techniques used to replace missing values using nearby observations.

- **Forward Fill (ffill):** Uses the previous available value.
- **Backward Fill (bfill):** Uses the next available value.

These techniques are especially useful for ordered or time-series data.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5, 6, 7],
    "Temperature": [30, np.nan, np.nan, 34, np.nan, 36, np.nan],
    "Sales": [100, 120, np.nan, np.nan, 180, np.nan, 220]
})

df
```

#### Identifying Missing Values

Before filling missing values, we first check which columns contain missing observations.

```python
df.isnull().sum()
```

#### 1. Forward Fill

Forward Fill (`ffill`) replaces a missing value with the most recent available value before it.

For example:

Before:

10  
20  
NaN  
NaN  
50

After Forward Fill:

10  
20  
20  
20  
50

Forward Fill is useful when the previous observation is a reasonable estimate of the missing value.

```python
df_ffill = df.copy()

df_ffill["Temperature"] = df_ffill["Temperature"].ffill()

df_ffill
```

```python
df_ffill = df.copy()

df_ffill["Sales"] = df_ffill["Sales"].ffill()

df_ffill
```

#### 2. Backward Fill

Backward Fill (`bfill`) replaces a missing value with the next available value after it.

For example:

Before:

10  
20  
NaN  
NaN  
50

After Backward Fill:

10  
20  
50  
50  
50

Backward Fill is useful when the next observation is a reasonable estimate of the missing value.

```python
df_bfill = df.copy()

df_bfill["Temperature"] = df_bfill["Temperature"].bfill()

df_bfill
```

```python
df_bfill = df.copy()

df_bfill["Sales"] = df_bfill["Sales"].bfill()

df_bfill
```

#### Forward Fill with Time-Series Data

Forward Fill is commonly used with time-series data because observations are naturally ordered by time.

For example, suppose a sensor records temperature throughout the day, but some readings are missing.

```python
dates = pd.date_range(
    start="2026-01-01",
    periods=7,
    freq="D"
)

temperature_df = pd.DataFrame({
    "Date": dates,
    "Temperature": [30, 31, np.nan, np.nan, 34, np.nan, 36]
})

temperature_df
```

```python
temperature_df["Temperature"] = (
    temperature_df["Temperature"].ffill()
)

temperature_df
```

```python
temperature_df = pd.DataFrame({
    "Date": dates,
    "Temperature": [30, 31, np.nan, np.nan, 34, np.nan, 36]
})

temperature_df["Temperature"] = (
    temperature_df["Temperature"].bfill()
)

temperature_df
```

#### Important Limitations

Forward Fill and Backward Fill should not be used blindly.

Potential problems include:

- A value may remain unchanged for too long.
- Large gaps can produce unrealistic values.
- The method may not be appropriate for every dataset.
- The order of observations is important.
- Forward Fill cannot fill missing values at the beginning if no previous value exists.
- Backward Fill cannot fill missing values at the end if no next value exists.

Always consider the meaning and structure of the data before choosing an imputation method.
