# KNN Imputation

KNN (K-Nearest Neighbors) Imputation is a missing-value technique that uses similar observations to estimate missing values.

Instead of replacing a missing value with a single statistic such as the mean or median, KNN Imputation looks at the nearest observations and uses their values to calculate the replacement.

The similarity between observations is usually determined using a distance metric such as Euclidean distance.

## How Does KNN Imputation Work?

The basic process is:

1. Find observations with similar feature values.
2. Identify the `K` nearest observations.
3. Look at the values of the feature containing the missing value.
4. Calculate an appropriate value from those neighbors.
5. Replace the missing value.

For example, if `K = 3`, the algorithm uses the 3 nearest observations to estimate the missing value.

A smaller `K` considers fewer neighbors, while a larger `K` considers more neighbors.

```python
import pandas as pd
import numpy as np

from sklearn.impute import KNNImputer
```

```python
df = pd.DataFrame({
    "Age": [20, 22, 25, np.nan, 30, 32, 35],
    "Salary": [30000, 35000, np.nan, 50000, 55000, np.nan, 70000],
    "Experience": [1, 2, 3, 5, np.nan, 8, 10]
})

df
```

```python
df.isnull().sum()
```

#### Applying KNN Imputation

Scikit-learn provides the `KNNImputer` class for KNN-based missing-value imputation.

The `n_neighbors` parameter determines how many neighboring observations are used.

For example:

`n_neighbors=3`

means the algorithm considers the 3 nearest observations.

```python
imputer = KNNImputer(n_neighbors=3)

df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

df_imputed
```

#### Before vs After Imputation

Let's compare the original dataset with the dataset after KNN imputation.

```python
print("Original Dataset:")
display(df)

print("\nKNN Imputed Dataset:")
display(df_imputed)
```
