# Iterative Imputation

Iterative Imputation is an advanced technique for handling missing values.

Instead of replacing missing values using only the mean or median, Iterative Imputation uses relationships between features to estimate missing values.

The process works iteratively:

1. Initialize missing values with an initial estimate.
2. Select one feature containing missing values.
3. Use the other features to predict its missing values.
4. Replace the missing values with the predictions.
5. Repeat the process for other features.
6. Continue for several iterations until the estimates become stable.

Iterative Imputation is also known as **Multivariate Imputation**.

#### How Does Iterative Imputation Work?

Suppose we have:

- Age
- Salary
- Experience

If Salary is missing, we can use Age and Experience to estimate Salary.

If Age is missing, we can use Salary and Experience to estimate Age.

The algorithm repeatedly updates these estimates using relationships between the features.

This makes Iterative Imputation different from simple imputation methods such as mean and median imputation.

```python
import pandas as pd
import numpy as np

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = pd.DataFrame({
    "Age": [20, 22, np.nan, 28, 30, np.nan, 35],
    "Salary": [30000, np.nan, 42000, 50000, np.nan, 65000, 70000],
    "Experience": [1, 2, 3, np.nan, 6, 8, 10]
})

df
```

```python
df.isnull().sum()
```

#### Applying Iterative Imputation

Scikit-learn provides `IterativeImputer` for multivariate iterative imputation.

By default, it uses a Bayesian Ridge regression model to estimate missing values.

The estimator can also be changed to another machine learning model.

```python
imputer = IterativeImputer(
    random_state=42
)

df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

df_imputed
```

```python
print("Original Dataset:")
display(df)

print("\nIteratively Imputed Dataset:")
display(df_imputed)
```
