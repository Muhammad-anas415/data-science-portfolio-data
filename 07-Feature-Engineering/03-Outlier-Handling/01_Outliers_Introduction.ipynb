# Introduction to Outliers

An **outlier** is an observation that is unusually far from the other observations in a dataset.

Outliers can occur because of:

- Measurement errors
- Data entry errors
- Sensor problems
- Unusual events
- Natural variation
- Rare but valid observations

Outliers are important because they can strongly affect statistical analysis and machine learning models.

#### Example

Consider the following ages:

`20, 21, 22, 23, 24, 25, 26, 90`

Most observations are between 20 and 26.

The value `90` is unusually far from the rest and may be considered an outlier.

However, an outlier is not automatically an error. It may represent a real observation.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 21, 22, 23, 24, 25, 26, 90],
    "Salary": [30000, 32000, 35000, 37000, 40000, 42000, 45000, 200000]
})

df
```

#### Understanding the Data

Before detecting outliers, we should understand the distribution of our numerical features.

Useful descriptive statistics include:

- Mean
- Median
- Minimum
- Maximum
- Standard deviation
- Quartiles

```python
df.describe()
```

#### Visualizing Outliers

Visualization is one of the easiest ways to identify potential outliers.

Common visualizations include:

- Box Plot
- Histogram
- Scatter Plot

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Age"])

plt.title("Age Distribution")
plt.ylabel("Age")

plt.show()
```

```python
plt.boxplot(df["Salary"])

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

```python
plt.hist(df["Age"], bins=8)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["Age"], df["Salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
