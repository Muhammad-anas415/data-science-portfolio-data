# Label Encoding

Label Encoding is a technique used to convert categorical values into numerical values.

Machine learning algorithms generally work with numerical data, so categorical features often need to be transformed.

For example:

Male → 1
Female → 0

Label Encoding assigns a unique integer to each category.

#### Why Do We Need Encoding?

Machine learning models generally cannot directly process text categories such as:

- Male
- Female
- Lahore
- Karachi
- Bachelor
- Master

Therefore, categorical values need to be converted into numerical representations.

Encoding is an important step in preprocessing categorical features.

```python
import pandas as pd

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Female", "Male", "Female"],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"]
})

df
```

#### LabelEncoder

Scikit-learn provides `LabelEncoder` for converting categorical labels into integers.

Each unique category receives a numerical label.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()

df["Gender_Encoded"] = encoder.fit_transform(df["Gender"])

df
```

```python
print("Classes:", encoder.classes_)
```

```python
mapping = dict(
    zip(
        encoder.classes_,
        encoder.transform(encoder.classes_)
    )
)

mapping
```

```python
city_encoder = LabelEncoder()

df["City_Encoded"] = city_encoder.fit_transform(df["City"])

df
```

```python
city_mapping = dict(
    zip(
        city_encoder.classes_,
        city_encoder.transform(city_encoder.classes_)
    )
)

city_mapping
```
