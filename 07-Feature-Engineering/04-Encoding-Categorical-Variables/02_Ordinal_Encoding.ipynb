# Ordinal Encoding

Ordinal Encoding converts categorical features into numerical values while preserving the meaningful order between categories.

Ordinal features have a natural ranking.

Examples:

- Small < Medium < Large
- Low < Medium < High
- Beginner < Intermediate < Advanced
- Bachelor's < Master's < PhD

For example:

Small → 0
Medium → 1
Large → 2

The numerical values represent the order of the categories.

```python
import pandas as pd

df = pd.DataFrame({
    "Size": [
        "Small",
        "Medium",
        "Large",
        "Medium",
        "Small",
        "Large"
    ]
})

df
```

```python
df["Size"].unique()
```

#### Manual Ordinal Encoding

We can manually define a mapping when we know the correct order of the categories.

For the Size feature:

Small → 0
Medium → 1
Large → 2

This approach gives us complete control over the ordering.

```python
size_mapping = {
    "Small": 0,
    "Medium": 1,
    "Large": 2
}

df["Size_Encoded"] = df["Size"].map(size_mapping)

df
```

```python
df[["Size", "Size_Encoded"]]
```

#### Education Level Example

Education level is an ordinal feature because the categories have a meaningful order.

For example:

High School < Bachelor's < Master's < PhD

```python
education_df = pd.DataFrame({
    "Education": [
        "High School",
        "Bachelor's",
        "Master's",
        "PhD",
        "Bachelor's"
    ]
})

education_df
```

```python
education_mapping = {
    "High School": 0,
    "Bachelor's": 1,
    "Master's": 2,
    "PhD": 3
}

education_df["Education_Encoded"] = (
    education_df["Education"].map(education_mapping)
)

education_df
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder(
    categories=[["Small", "Medium", "Large"]]
)

df["Size_Ordinal"] = encoder.fit_transform(
    df[["Size"]]
)

df
```

```python
df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Medium"],
    "Satisfaction": ["Low", "High", "Medium", "High"]
})

df
```

```python
encoder = OrdinalEncoder(
    categories=[
        ["Small", "Medium", "Large"],
        ["Low", "Medium", "High"]
    ]
)

encoded = encoder.fit_transform(df)

encoded_df = pd.DataFrame(
    encoded,
    columns=["Size_Encoded", "Satisfaction_Encoded"]
)

encoded_df
```

```python
result = pd.concat(
    [df, encoded_df],
    axis=1
)

result
```
