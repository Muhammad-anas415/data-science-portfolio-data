# Ordinal Encoding

Ordinal Encoding converts categorical features into numerical values while preserving the meaningful order between categories.

Ordinal features have a natural ranking.

Examples:

- Small < Medium < Large
- Low < Medium < High
- Beginner < Intermediate < Advanced
- Bachelor's < Master's < PhD

For example:

Small → 0
Medium → 1
Large → 2

The numerical values represent the order of the categories.

```python
import pandas as pd

df = pd.DataFrame({
    "Size": [
        "Small",
        "Medium",
        "Large",
        "Medium",
        "Small",
        "Large"
    ]
})

df
```

```python
df["Size"].unique()
```

#### Manual Ordinal Encoding

We can manually define a mapping when we know the correct order of the categories.

For the Size feature:

Small → 0
Medium → 1
Large → 2

This approach gives us complete control over the ordering.

```python
size_mapping = {
    "Small": 0,
    "Medium": 1,
    "Large": 2
}

df["Size_Encoded"] = df["Size"].map(size_mapping)

df
```

#### Using Scikit-Learn OrdinalEncoder

Scikit-learn provides `OrdinalEncoder` for encoding categorical input features.

We can explicitly specify the order of the categories so that the encoding represents the correct ranking.

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder(
    categories=[["Small", "Medium", "Large"]]
)

df["Size_Ordinal"] = encoder.fit_transform(
    df[["Size"]]
)

df
```

```python
df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Medium"],
    "Satisfaction": ["Low", "High", "Medium", "High"]
})

df
```

```python
encoder = OrdinalEncoder(
    categories=[
        ["Small", "Medium", "Large"],
        ["Low", "Medium", "High"]
    ]
)

encoded = encoder.fit_transform(df)

encoded_df = pd.DataFrame(
    encoded,
    columns=["Size_Encoded", "Satisfaction_Encoded"]
)

encoded_df
```

```python
result = pd.concat(
    [df, encoded_df],
    axis=1
)

result
```
