# Frequency Encoding

Frequency Encoding is a categorical encoding technique that replaces each category with the number of times it appears in the dataset.

For example:

Lahore → 3
Karachi → 2
Islamabad → 1

Frequency Encoding converts categorical values into numerical values based on their frequency.

It is especially useful for categorical features with many unique categories.

```python
import pandas as pd

df = pd.DataFrame({
    "City": [
        "Lahore",
        "Karachi",
        "Lahore",
        "Islamabad",
        "Karachi",
        "Lahore",
        "Lahore",
        "Islamabad"
    ]
})

df
```

#### Calculating Category Frequency

First, we count how many times each category appears.

Pandas provides the `value_counts()` method for this purpose.

```python
frequency = df["City"].value_counts()

frequency
```

#### Applying Frequency Encoding

We can replace each category with its corresponding frequency.

For example:

Lahore → 4
Karachi → 2
Islamabad → 2

Every occurrence of the same category receives the same numerical value.

```python
df["City_Frequency"] = df["City"].map(frequency)

df
```

#### Frequency as a Percentage

Instead of using the raw count, we can represent frequency as a proportion of the total observations.

Formula:

Frequency Ratio = Category Count / Total Number of Observations

```python
frequency_ratio = df["City"].value_counts(normalize=True)

frequency_ratio
```

```python
df["City_Frequency_Ratio"] = df["City"].map(
    frequency_ratio
)

df
```

```python
df[[
    "City",
    "City_Frequency",
    "City_Frequency_Ratio"
]]
```

#### Creating a Reusable Function

We can create a function to perform frequency encoding on any categorical column.

```python
def frequency_encode(data, column):
    frequency = data[column].value_counts()
    
    data[column + "_Frequency"] = (
        data[column].map(frequency)
    )
    
    return data
```

```python
df = pd.DataFrame({
    "City": [
        "Lahore",
        "Karachi",
        "Lahore",
        "Islamabad",
        "Karachi",
        "Lahore"
    ]
})

df = frequency_encode(df, "City")

df
```

#### Frequency Encoding Multiple Features

Frequency Encoding can also be applied to multiple categorical features.

For example:

- City
- Department
- Product Category

```python
df = pd.DataFrame({
    "City": [
        "Lahore", "Karachi", "Lahore",
        "Islamabad", "Karachi", "Lahore"
    ],
    "Department": [
        "IT", "HR", "IT",
        "Finance", "HR", "IT"
    ]
})

df
```

```python
for column in ["City", "Department"]:
    
    frequency = df[column].value_counts()
    
    df[column + "_Frequency"] = (
        df[column].map(frequency)
    )

df
```

#### High-Cardinality Features

Frequency Encoding is useful when a categorical feature has many unique categories.

For example:

- ZIP Code
- Product ID
- Customer ID
- Store ID
- Website Domain

One-Hot Encoding could create hundreds or thousands of columns.

Frequency Encoding represents each category using only one numerical feature.

```python
df = pd.DataFrame({
    "Product": [
        "A", "A", "A",
        "B", "B",
        "C",
        "D", "D", "D", "D"
    ]
})

frequency = df["Product"].value_counts()

df["Product_Frequency"] = df["Product"].map(
    frequency
)

df
```
