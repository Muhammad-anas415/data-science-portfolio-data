# Standardization

Standardization is a feature scaling technique that transforms numerical features so that they have:

- Mean ≈ 0
- Standard Deviation ≈ 1

It is also known as **Z-score normalization**.

The formula is:

z = (x - μ) / σ

Where:

- x = Original value
- μ = Mean of the feature
- σ = Standard deviation

Standardization does not force values into a fixed range such as 0 to 1.

## Why Do We Need Standardization?

Consider two features:

Age:
20, 25, 30, 35, 40

Salary:
25000, 40000, 60000, 90000, 150000

Salary has a much larger numerical scale than Age.

Algorithms that depend on distance, magnitude, or gradients may be affected by these different scales.

Standardization transforms the features to a comparable scale.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [25000, 40000, 60000, 90000, 150000],
    "Experience": [1, 3, 5, 8, 12]
})

df
```

```python
df.describe()
```

## Manual Standardization

We can standardize a feature using the Z-score formula:

z = (x - μ) / σ

Steps:

1. Calculate the mean.
2. Calculate the standard deviation.
3. Subtract the mean from each value.
4. Divide by the standard deviation.

```python
mean = df["Age"].mean()
std = df["Age"].std()

df["Age_Standardized"] = (
    (df["Age"] - mean) / std
)

df
```

```python
print("Mean:", df["Age_Standardized"].mean())
print("Standard Deviation:", df["Age_Standardized"].std())
```

## Standardizing Multiple Features

In a real machine learning project, we usually standardize multiple numerical features at the same time.

Scikit-learn provides `StandardScaler` for this purpose.

```python
from sklearn.preprocessing import StandardScaler
```

```python
features = [
    "Age",
    "Salary",
    "Experience"
]

scaler = StandardScaler()

df_scaled = pd.DataFrame(
    scaler.fit_transform(df[features]),
    columns=features
)

df_scaled
```

```python
print("Means:")
print(df_scaled.mean())

print("\nStandard Deviations:")
print(df_scaled.std())
```

```python
print(df_scaled.mean())
print(df_scaled.std(ddof=0))
```

## Before vs After Standardization

Before standardization:

- Features have different numerical ranges.
- Salary has much larger values than Age.

After standardization:

- Features are centered around zero.
- Features have comparable scales.

```python
print("Original Data:")
display(df[features])

print("\nStandardized Data:")
display(df_scaled)
```
