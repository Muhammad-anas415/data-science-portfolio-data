# Square Root Transformation

Square Root Transformation is a mathematical transformation used to reduce the skewness of numerical data.

It is generally less aggressive than Log Transformation.

The formula is:

x_transformed = √x

Square Root Transformation is commonly useful for:

- Right-skewed data
- Count data
- Moderate skewness
- Data containing zero values

## Why Do We Need Square Root Transformation?

Many real-world numerical features are right-skewed.

Examples include:

- Number of purchases
- Number of website visits
- Number of calls
- Number of customers
- Count of events

Square Root Transformation compresses larger values more than smaller values.

This can help reduce right skewness and make the distribution more balanced.

```python
import pandas as pd

df = pd.DataFrame({
    "Visits": [
        0, 1, 2, 4, 6,
        10, 15, 25, 50, 100
    ]
})

df
```

```python
import matplotlib.pyplot as plt

plt.hist(df["Visits"], bins=10)

plt.title("Before Square Root Transformation")
plt.xlabel("Visits")
plt.ylabel("Frequency")

plt.show()
```

```python
print("Original Skewness:", df["Visits"].skew())
```

## Manual Square Root Transformation

We can apply the square root transformation using:

√x

For example:

√100 = 10

√25 = 5

√4 = 2

The transformation compresses large values while keeping zero as zero.

```python
import numpy as np

df["Visits_Sqrt"] = np.sqrt(df["Visits"])

df
```

```python
df[["Visits", "Visits_Sqrt"]]
```

```python
print("Original Skewness:")
print(df["Visits"].skew())

print("\nSquare Root Skewness:")
print(df["Visits_Sqrt"].skew())
```

```python
plt.hist(df["Visits_Sqrt"], bins=10)

plt.title("After Square Root Transformation")
plt.xlabel("Square Root Transformed Visits")
plt.ylabel("Frequency")

plt.show()
```
