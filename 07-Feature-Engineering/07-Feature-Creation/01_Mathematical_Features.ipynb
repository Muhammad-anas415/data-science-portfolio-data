# Mathematical Features

Mathematical Feature Engineering is the process of creating new features by applying mathematical operations to existing numerical features.

These transformations can help machine learning models discover relationships that may not be obvious from the original features.

Common mathematical operations include:

- Addition
- Subtraction
- Multiplication
- Division
- Power
- Absolute Value
- Logarithm
- Square Root

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 200, 300, 400, 500],
    "Quantity": [2, 3, 4, 5, 6],
    "Discount": [10, 20, 30, 40, 50]
})

df
```

## Addition Features

We can combine two numerical features using addition.

For example:

Total = Price + Discount

Mathematical operations can create new information from existing features.

```python
df["Price_Discount_Total"] = (
    df["Price"] + df["Discount"]
)

df
```

## Subtraction Features

Subtraction can be used to calculate differences between numerical features.

For example:

Price Difference = Price - Discount

```python
df["Price_After_Discount"] = (
    df["Price"] - df["Discount"]
)

df
```

## Multiplication Features

Multiplication can be used to create meaningful features from two numerical variables.

For example:

Total Cost = Price × Quantity

```python
df["Total_Cost"] = (
    df["Price"] * df["Quantity"]
)

df
```

## Division Features

Division can be used to create ratios between features.

For example:

Price Per Unit = Total Cost / Quantity

```python
df["Price_Per_Unit"] = (
    df["Total_Cost"] / df["Quantity"]
)

df
```

## Ratio Features

Ratios are commonly used in feature engineering because they describe the relationship between two numerical features.

Example:

Discount Ratio = Discount / Price

A ratio can sometimes provide more useful information than the original features.

```python
df["Discount_Ratio"] = (
    df["Discount"] / df["Price"]
)

df
```

## Power Features

We can create new features by raising an existing feature to a power.

Examples:

x²
x³
√x

Polynomial features can help models capture non-linear relationships.

```python
df["Price_Squared"] = df["Price"] ** 2

df["Price_Cubed"] = df["Price"] ** 3

df
```

## Absolute Value Features

The absolute value removes the sign of a number.

For example:

|-10| = 10

This can be useful when only the magnitude of a difference is important.

```python
df["Price_Difference"] = (
    df["Price"] - df["Discount"]
)

df["Absolute_Difference"] = (
    df["Price_Difference"].abs()
)

df
```

## Logarithmic Features

Logarithmic transformation can compress large numerical values.

It is particularly useful for highly right-skewed features.

For non-negative data, we can use:

log(1 + x)

```python
import numpy as np

df["Price_Log"] = np.log1p(df["Price"])

df
```

## Square Root Features

Square root transformation can be useful for reducing moderate right skewness.

Formula:

√x

```python
df["Price_Sqrt"] = np.sqrt(df["Price"])

df
```
