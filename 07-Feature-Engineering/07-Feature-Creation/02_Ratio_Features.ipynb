# Ratio Features

Ratio Feature Engineering creates a new feature by dividing one numerical feature by another.

A ratio describes the relationship between two variables.

General formula:

Ratio = Feature A / Feature B

Ratio features are useful when the relationship between two variables is more meaningful than their individual values.

## Why Do We Create Ratio Features?

Two individual features may not fully describe a relationship.

For example:

- Income
- Expenses

Instead of using only these two features, we can create:

Savings Ratio = Expenses / Income

This can provide the model with additional information about the relationship between income and expenses.

```python
import pandas as pd

df = pd.DataFrame({
    "Income": [30000, 40000, 50000, 60000, 80000],
    "Expenses": [15000, 20000, 30000, 30000, 40000],
    "Age": [20, 25, 30, 35, 40]
})

df
```

## Creating a Basic Ratio

We can calculate the ratio between two numerical features.

For example:

Expense Ratio = Expenses / Income

This tells us how large the expenses are relative to income.

```python
df["Expense_Ratio"] = (
    df["Expenses"] / df["Income"]
)

df
```

## Ratio as a Percentage

A ratio can also be converted into a percentage.

Formula:

Percentage = Ratio × 100

For example:

Expense Ratio = 0.5

Expense Percentage = 0.5 × 100 = 50%

This means expenses are 50% of income.

```python
df["Expense_Percentage"] = (
    df["Expense_Ratio"] * 100
)

df
```

## Price Per Unit

A common real-world ratio feature is calculating the value per unit.

Formula:

Price Per Unit = Total Price / Quantity

```python
df = pd.DataFrame({
    "Total_Price": [1000, 2000, 3000, 5000, 8000],
    "Quantity": [10, 20, 30, 25, 40]
})

df["Price_Per_Unit"] = (
    df["Total_Price"] / df["Quantity"]
)

df
```

## Business Ratio Example

Financial datasets commonly contain useful ratio features.

Examples:

- Profit Margin = Profit / Revenue
- Debt Ratio = Debt / Assets
- Expense Ratio = Expenses / Income
- Return on Investment = Profit / Investment

These ratios can provide more meaningful information than raw values alone.

```python
df = pd.DataFrame({
    "Revenue": [100000, 150000, 200000, 300000],
    "Profit": [10000, 30000, 50000, 90000]
})

df["Profit_Margin"] = (
    df["Profit"] / df["Revenue"]
)

df
```

## Handling Division by Zero

Ratio features can create problems when the denominator contains zero.

For example:

Ratio = A / B

If:

B = 0

the result becomes undefined or infinite.

Therefore, zero denominators must be handled carefully.

```python
import numpy as np

df = pd.DataFrame({
    "Income": [30000, 40000, 0, 60000],
    "Expenses": [15000, 20000, 10000, 30000]
})

df["Expense_Ratio"] = (
    df["Expenses"] /
    df["Income"].replace(0, np.nan)
)

df
```

```python
df["Expense_Ratio"] = (
    df["Expense_Ratio"]
    .replace([np.inf, -np.inf], np.nan)
)

df
```
