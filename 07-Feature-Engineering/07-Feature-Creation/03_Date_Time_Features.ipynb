# DateTime Features

DateTime Feature Engineering is the process of extracting useful information from date and time values.

A single datetime column can contain many useful features, such as:

- Year
- Month
- Day
- Day of Week
- Hour
- Minute
- Second
- Quarter
- Week of Year
- Weekend
- Time of Day

These features can help machine learning models understand patterns related to time.

```python
import pandas as pd

df = pd.DataFrame({
    "Date": [
        "2026-01-15 08:30:00",
        "2026-02-20 14:45:00",
        "2026-03-25 18:15:00",
        "2026-04-10 21:00:00",
        "2026-05-05 10:20:00"
    ]
})

df
```

## Converting Data to Datetime

Before extracting datetime features, the date column should be converted into pandas datetime format.

We can use:

`pd.to_datetime()`

This allows us to easily extract different components from the date.

```python
df["Date"] = pd.to_datetime(df["Date"])

df.dtypes
```

```python
df["Year"] = df["Date"].dt.year

df
```

```python
df["Month"] = df["Date"].dt.month

df
```

```python
df["Day"] = df["Date"].dt.day

df
```

## Day of Week

The day of the week can be useful because many real-world activities depend on weekdays and weekends.

Examples:

- Monday = 0
- Tuesday = 1
- Wednesday = 2
- ...
- Sunday = 6

```python
df["Day_of_Week"] = df["Date"].dt.dayofweek

df
```

```python
df["Day_Name"] = df["Date"].dt.day_name()

df
```

## Weekend Feature

We can create a binary feature that identifies whether a date falls on a weekend.

Saturday and Sunday → 1

Monday to Friday → 0

```python
df["Is_Weekend"] = (
    df["Date"].dt.dayofweek >= 5
).astype(int)

df
```

```python
df["Hour"] = df["Date"].dt.hour

df
```

```python
df["Minute"] = df["Date"].dt.minute
df["Second"] = df["Date"].dt.second

df
```

## Quarter

A year is divided into four quarters:

- Q1 → January to March
- Q2 → April to June
- Q3 → July to September
- Q4 → October to December

Quarter information can be useful for business and seasonal analysis.

```python
df["Quarter"] = df["Date"].dt.quarter

df
```

```python
df["Week_of_Year"] = (
    df["Date"].dt.isocalendar().week
)

df
```

```python
df["Day_of_Year"] = df["Date"].dt.dayofyear

df
```

## Time of Day

We can convert the hour into meaningful categories.

Example:

- Morning
- Afternoon
- Evening
- Night

This can be useful for applications such as:

- Sales prediction
- Traffic prediction
- Website activity
- Customer behavior

```python
def time_of_day(hour):
    
    if 5 <= hour < 12:
        return "Morning"
    
    elif 12 <= hour < 17:
        return "Afternoon"
    
    elif 17 <= hour < 21:
        return "Evening"
    
    else:
        return "Night"


df["Time_of_Day"] = df["Hour"].apply(time_of_day)

df
```

```python
df["Month_Name"] = df["Date"].dt.month_name()

df
```

```python
df["Month_Start"] = df["Date"].dt.is_month_start
df["Month_End"] = df["Date"].dt.is_month_end

df
```

## Date Difference

We can calculate the number of days between two dates.

This is useful for features such as:

- Customer age
- Days since purchase
- Days since last login
- Subscription duration
- Delivery time

```python
df_dates = pd.DataFrame({
    "Order_Date": pd.to_datetime([
        "2026-01-01",
        "2026-01-10",
        "2026-02-01"
    ]),
    
    "Delivery_Date": pd.to_datetime([
        "2026-01-05",
        "2026-01-15",
        "2026-02-08"
    ])
})

df_dates["Delivery_Days"] = (
    df_dates["Delivery_Date"] -
    df_dates["Order_Date"]
).dt.days

df_dates
```

## Creating Age from Date of Birth

DateTime features can also be used to calculate age.

A common approach is:

Age = Current Date - Date of Birth

For precise age calculations, the year/month/day relationship should be considered rather than simply dividing days by 365.

```python
import pandas as pd

today = pd.Timestamp("2026-08-13")

df_age = pd.DataFrame({
    "Date_of_Birth": pd.to_datetime([
        "2000-05-10",
        "1995-08-20",
        "2005-01-15"
    ])
})

df_age["Age"] = (
    today.year - df_age["Date_of_Birth"].dt.year
    - (
        (df_age["Date_of_Birth"].dt.month > today.month)
        |
        (
            (df_age["Date_of_Birth"].dt.month == today.month)
            &
            (df_age["Date_of_Birth"].dt.day > today.day)
        )
    ).astype(int)
)

df_age
```

## Cyclical DateTime Features

Some datetime features are cyclical.

For example:

Hour 23 → Hour 0

These values are actually close together, but a normal numerical representation makes them appear far apart.

We can represent cyclical features using:

sin()

and

cos()

This is useful for:

- Hour
- Day of week
- Month
- Day of year

```python
import numpy as np

df["Hour_Sin"] = np.sin(
    2 * np.pi * df["Hour"] / 24
)

df["Hour_Cos"] = np.cos(
    2 * np.pi * df["Hour"] / 24
)

df
```

```python
datetime_features = pd.DataFrame({
    "Year": df["Date"].dt.year,
    "Month": df["Date"].dt.month,
    "Day": df["Date"].dt.day,
    "Day_of_Week": df["Date"].dt.dayofweek,
    "Hour": df["Date"].dt.hour,
    "Minute": df["Date"].dt.minute,
    "Quarter": df["Date"].dt.quarter,
    "Week_of_Year": df["Date"].dt.isocalendar().week,
    "Day_of_Year": df["Date"].dt.dayofyear,
    "Is_Weekend": (
        df["Date"].dt.dayofweek >= 5
    ).astype(int)
})

datetime_features
```
