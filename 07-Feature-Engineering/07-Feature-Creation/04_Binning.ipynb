# Binning

Binning is a feature engineering technique that converts continuous numerical values into a smaller number of groups or intervals called **bins**.

Instead of using exact numerical values, we group similar values together.

For example:

Age:

0–18   → Child
19–35  → Young Adult
36–60  → Adult
61+    → Senior

Binning can simplify numerical data and make patterns easier for a machine learning model to learn.

## Why Do We Use Binning?

Binning can be useful when:

- A numerical feature has a non-linear relationship with the target.
- We want to reduce the effect of small variations.
- We want to convert continuous values into meaningful groups.
- We want to make numerical data easier to interpret.

Examples:

- Age groups
- Income groups
- Temperature ranges
- Exam score categories
- Customer spending groups

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [5, 12, 18, 22, 27, 35, 42, 50, 65, 80]
})

df
```

## Equal-Width Binning

Equal-width binning divides the range of values into intervals having approximately equal numerical widths.

For example:

0–20
20–40
40–60
60–80
80–100

Each interval has the same width.

```python
df["Age_Bin"] = pd.cut(
    df["Age"],
    bins=4
)

df
```

## Custom Binning

We can manually define meaningful boundaries based on domain knowledge.

For example:

- 0–18 → Child
- 19–35 → Young Adult
- 36–60 → Adult
- 61+ → Senior

```python
bins = [0, 18, 35, 60, 100]

labels = [
    "Child",
    "Young Adult",
    "Adult",
    "Senior"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df
```

## Equal-Frequency Binning

Equal-frequency binning attempts to place approximately the same number of observations into each bin.

Pandas provides `qcut()` for this purpose.

It divides the data according to quantiles.

```python
df["Age_Quantile_Bin"] = pd.qcut(
    df["Age"],
    q=4
)

df
```

```python
df["Age_Quantile_Group"] = pd.qcut(
    df["Age"],
    q=4,
    labels=[
        "Q1",
        "Q2",
        "Q3",
        "Q4"
    ]
)

df
```

## Example: Exam Score Binning

Binning can convert numerical exam scores into meaningful performance categories.

Example:

0–49   → Fail
50–59  → Pass
60–69  → Good
70–79  → Very Good
80–100 → Excellent

```python
df_scores = pd.DataFrame({
    "Score": [35, 48, 55, 62, 68, 73, 81, 90, 96]
})

bins = [0, 49, 59, 69, 79, 100]

labels = [
    "Fail",
    "Pass",
    "Good",
    "Very Good",
    "Excellent"
]

df_scores["Performance"] = pd.cut(
    df_scores["Score"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df_scores
```

## Example: Income Binning

Income can also be converted into meaningful groups.

For example:

Low Income
Middle Income
High Income
Very High Income

The boundaries should be selected according to the dataset and domain.

```python
df_income = pd.DataFrame({
    "Income": [
        20000,
        30000,
        50000,
        75000,
        100000,
        150000,
        250000
    ]
})

bins = [
    0,
    40000,
    80000,
    150000,
    float("inf")
]

labels = [
    "Low",
    "Middle",
    "High",
    "Very High"
]

df_income["Income_Group"] = pd.cut(
    df_income["Income"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df_income
```

## Converting Bins to Numerical Values

Some machine learning algorithms require numerical input.

Therefore, categorical bins can be converted into numerical labels.

For example:

Low → 0
Medium → 1
High → 2

```python
df_income["Income_Group_Code"] = (
    df_income["Income_Group"]
    .cat.codes
)

df_income
```
