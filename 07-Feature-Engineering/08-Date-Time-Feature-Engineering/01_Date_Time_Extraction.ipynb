# DateTime Extraction

DateTime Extraction is the process of extracting useful information from a date or time column.

A single datetime value can contain many useful features, such as:

- Year
- Month
- Day
- Hour
- Minute
- Second
- Day of Week
- Week of Year
- Quarter
- Day of Year

These extracted features can be used in Exploratory Data Analysis and Machine Learning.

```python
import pandas as pd

df = pd.DataFrame({
    "Date": [
        "2026-01-15 08:30:45",
        "2026-02-20 14:45:30",
        "2026-03-25 18:15:20",
        "2026-04-10 21:00:10",
        "2026-05-05 10:20:55"
    ]
})

df
```

## Converting a Column to Datetime

Before extracting datetime components, we first convert the column into Pandas datetime format.

We use:

`pd.to_datetime()`

This allows us to use the `.dt` accessor for datetime extraction.

```python
df["Date"] = pd.to_datetime(df["Date"])

df
```

```python
df.dtypes
```

## Extracting Year

The year can be extracted using:

`dt.year`

Example:

2026-01-15 → 2026

```python
df["Year"] = df["Date"].dt.year

df
```

## Extracting Month

The month can be extracted using:

`dt.month`

The result is a number from:

1 → January

12 → December

```python
df["Month"] = df["Date"].dt.month

df
```

```python
df["Month_Name"] = df["Date"].dt.month_name()

df
```

## Extracting Day

The day of the month can be extracted using:

`dt.day`

For example:

2026-05-05 → 5

```python
df["Day"] = df["Date"].dt.day

df
```

## Extracting Day of Week

Pandas represents:

Monday → 0
Tuesday → 1
Wednesday → 2
Thursday → 3
Friday → 4
Saturday → 5
Sunday → 6

This can be useful for identifying weekday and weekend patterns.

```python
df["Day_of_Week"] = df["Date"].dt.dayofweek

df
```

```python
df["Day_Name"] = df["Date"].dt.day_name()

df
```

## Extracting Hour

The hour can be extracted using:

`dt.hour`

For example:

08:30:45 → 8

14:45:30 → 14

21:00:10 → 21

```python
df["Hour"] = df["Date"].dt.hour

df
```

```python
df["Minute"] = df["Date"].dt.minute

df
```

```python
df["Second"] = df["Date"].dt.second

df
```
