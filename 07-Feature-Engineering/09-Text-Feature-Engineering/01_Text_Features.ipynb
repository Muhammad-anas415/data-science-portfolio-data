# Text Features

Text Feature Engineering is the process of converting raw text into
structured numerical features that machine learning models can understand.

Machine learning models cannot directly work with raw text.

Therefore, text needs to be transformed into numerical representations.

Common text features include:

- Text Length
- Word Count
- Character Count
- Sentence Count
- Average Word Length
- Number of Digits
- Number of Special Characters
- Uppercase Count
- Punctuation Count
- Bag of Words
- TF-IDF

```python
import pandas as pd

df = pd.DataFrame({
    "Text": [
        "I love machine learning!",
        "Python is amazing for data science.",
        "This is my first NLP project.",
        "Machine learning is powerful."
    ]
})

df
```

## Text Length

Text length represents the total number of characters in a text.

It can be useful for identifying differences between short and long messages.

```python
df["Text_Length"] = df["Text"].str.len()

df
```

```python
df["Character_Count"] = df["Text"].str.len()

df
```

## Word Count

Word Count represents the number of words present in a text.

Formula:

Word Count = Number of words in the text

```python
df["Word_Count"] = df["Text"].str.split().str.len()

df
```

## Sentence Count

Sentence Count represents the number of sentences in a text.

This can provide information about the structure and complexity of the text.

```python
import re

df["Sentence_Count"] = df["Text"].apply(
    lambda x: len(re.findall(r"[.!?]+", x))
)

df
```

## Average Word Length

Average Word Length represents the average number of characters per word.

Formula:

Average Word Length =
Total Characters / Total Words

```python
df["Average_Word_Length"] = (
    df["Text_Length"] / df["Word_Count"]
)

df
```

```python
df["Uppercase_Count"] = df["Text"].apply(
    lambda x: sum(char.isupper() for char in x)
)

df
```

## Digit Count

Digit Count represents the number of numerical digits present in the text.

For example:

"Order 12345"

contains 5 digits.

```python
df["Digit_Count"] = df["Text"].apply(
    lambda x: sum(char.isdigit() for char in x)
)

df
```

```python
df["Special_Character_Count"] = df["Text"].apply(
    lambda x: sum(
        not char.isalnum() and not char.isspace()
        for char in x
    )
)

df
```

## Punctuation Count

Punctuation can sometimes provide useful information about text.

Examples:

- !
- ?
- .
- ,
- :
- ;

```python
import string

df["Punctuation_Count"] = df["Text"].apply(
    lambda x: sum(
        char in string.punctuation
        for char in x
    )
)

df
```

```python
df["Whitespace_Count"] = df["Text"].apply(
    lambda x: sum(char.isspace() for char in x)
)

df
```

## Text Normalization

Converting text to lowercase is a common preprocessing step.

For example:

"Python Is Amazing"

becomes:

"python is amazing"

This helps reduce unnecessary differences between words.

```python
df["Lowercase_Text"] = df["Text"].str.lower()

df
```

## URL Count

The number of URLs in a text can be useful for tasks such as:

- Spam detection
- Phishing detection
- Social media analysis
- Email classification

```python
url_pattern = r"https?://\S+|www\.\S+"

df["URL_Count"] = df["Text"].apply(
    lambda x: len(re.findall(url_pattern, x))
)

df
```

## Keyword Features

We can create binary features based on the presence of specific keywords.

For example, in spam detection:

- free
- winner
- offer
- prize
- urgent

A feature can indicate whether a particular keyword appears in the text.

```python
df["Contains_Machine"] = (
    df["Text"]
    .str.lower()
    .str.contains("machine")
    .astype(int)
)

df
```

## Bag of Words

Bag of Words converts text into numerical features based on word frequency.

Each unique word becomes a feature.

For example:

Text 1: "I love Python"

Text 2: "I love machine learning"

The vocabulary may contain:

I, love, Python, machine, learning

Each document is represented using the frequency of these words.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X_bow = vectorizer.fit_transform(df["Text"])

X_bow.toarray()
```

```python
print(vectorizer.get_feature_names_out())
```

## TF-IDF

TF-IDF stands for:

Term Frequency - Inverse Document Frequency

It assigns higher importance to words that are important in a document
but less common across the entire collection of documents.

TF-IDF is widely used for:

- Text classification
- Spam detection
- Sentiment analysis
- Document similarity
- Search systems

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(df["Text"])

X_tfidf.toarray()
```
