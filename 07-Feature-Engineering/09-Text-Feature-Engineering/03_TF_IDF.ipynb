# TF-IDF

TF-IDF stands for **Term Frequency - Inverse Document Frequency**.

TF-IDF is a text feature engineering technique that converts text into
numerical features based on the importance of words.

The main idea is:

- A word is important if it appears frequently in a document.
- A word is less important if it appears in many documents.

TF-IDF gives higher values to words that are important to a particular
document and lower values to very common words.

## Why Do We Need TF-IDF?

Bag of Words mainly counts how frequently words occur.

However, some words may occur frequently across almost every document.

For example:

- the
- is
- and
- data
- machine

A word that appears in almost every document may not be useful for
distinguishing between documents.

TF-IDF reduces the importance of common words and gives more importance
to words that are more specific to individual documents.

```python
documents = [
    "machine learning is powerful",
    "machine learning is useful",
    "deep learning is powerful"
]

documents
```

## Mathematical Formula

TF-IDF is calculated using two main components:

### 1. Term Frequency (TF)

TF measures how frequently a term appears in a document.

A simple formula is:

TF = Number of times term appears in document
     / Total number of terms in document

### 2. Inverse Document Frequency (IDF)

IDF reduces the importance of words that appear in many documents.

A common formula is:

IDF = log(N / DF)

Where:

N = Total number of documents

DF = Number of documents containing the term

### Final Formula

TF-IDF = TF × IDF

## TF-IDF Using Scikit-Learn

Scikit-learn provides `TfidfVectorizer` to automatically calculate
TF-IDF features.

It performs several steps:

1. Tokenizes the text.
2. Creates a vocabulary.
3. Calculates term frequency.
4. Calculates inverse document frequency.
5. Produces TF-IDF vectors.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

X
```

```python
vectorizer.get_feature_names_out()
```

```python
import pandas as pd

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

tfidf_df
```

## Understanding the TF-IDF Matrix

Each row represents a document.

Each column represents a word from the vocabulary.

Each value represents the TF-IDF score of that word in that document.

Higher value:

The word is more important for that document.

Lower value:

The word is common or less important for that document.

## Removing Stop Words

Stop words are very common words that may not provide useful information
for some NLP tasks.

Examples:

- the
- is
- a
- an
- and
- of

We can remove English stop words using:

`stop_words="english"`

```python
vectorizer = TfidfVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

## N-Grams with TF-IDF

By default, TF-IDF uses individual words called unigrams.

We can also include combinations of words.

Example:

"machine learning"

Bigram:

"machine learning"

Using n-grams can help preserve some information about word relationships.

```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2)
)

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

```python
df = pd.DataFrame({
    "Text": [
        "I love machine learning",
        "Python is great for machine learning",
        "I love Python",
        "Machine learning is powerful"
    ]
})

df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["Text"])

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

tfidf_df
```

## Limiting the Number of Features

Large text datasets can contain thousands or millions of unique words.

We can limit the number of features using:

`max_features`

This keeps only the most important features according to the vectorizer's
feature selection process.

```python
vectorizer = TfidfVectorizer(
    max_features=10
)

X = vectorizer.fit_transform(df["Text"])

print(vectorizer.get_feature_names_out())
```
