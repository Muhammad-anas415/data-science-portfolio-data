# Feature Selection

Feature Selection is the process of selecting the most useful features
from a dataset for building a machine learning model.

A dataset may contain many features, but not every feature is useful.

Feature selection helps us:

- Remove irrelevant features.
- Remove redundant features.
- Reduce the dimensionality of the dataset.
- Improve model performance.
- Reduce training time.
- Make models easier to interpret.
- Reduce the risk of overfitting.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(18, 60, 100),
    "Income": np.random.randint(20000, 100000, 100),
    "Experience": np.random.randint(0, 20, 100),
    "Random_Feature": np.random.randn(100),
    "Customer_ID": range(1, 101),
    "Target": np.random.randint(0, 2, 100)
})

df.head()
```

## Why Do We Need Feature Selection?

Using every available feature is not always a good idea.

Some features may be:

- Irrelevant
- Redundant
- Highly correlated
- Noisy
- Duplicated
- Constant
- Weakly related to the target

Keeping unnecessary features can increase model complexity and training
time.

Feature selection attempts to keep the most informative features.

## Main Types of Feature Selection

There are three major approaches:

### 1. Filter Methods

Features are selected using statistical properties of the data.

Examples:

- Correlation
- Chi-Square
- ANOVA
- Mutual Information
- Variance Threshold

### 2. Wrapper Methods

Different subsets of features are evaluated using a machine learning model.

Examples:

- Forward Selection
- Backward Elimination
- Recursive Feature Elimination (RFE)

### 3. Embedded Methods

Feature selection happens during model training.

Examples:

- Lasso Regression
- Decision Trees
- Random Forest
- Gradient Boosting

## Filter Methods

Filter methods select features based on statistical relationships between
features and the target.

They are generally fast because they do not require repeatedly training
a machine learning model.

Common methods:

- Variance Threshold
- Correlation
- Chi-Square Test
- ANOVA
- Mutual Information

## Wrapper Methods

Wrapper methods evaluate different combinations or subsets of features
using a machine learning model.

The model performance is used to determine which features are useful.

Common methods:

- Forward Selection
- Backward Elimination
- Recursive Feature Elimination (RFE)

Wrapper methods can provide good feature subsets but may be computationally
expensive.

## Embedded Methods

Embedded methods perform feature selection as part of the model training
process.

Examples:

### Lasso Regression

Lasso can shrink some feature coefficients to zero.

### Decision Trees

Decision trees can measure feature importance based on how useful a feature
is for splitting the data.

Embedded methods are often more efficient than wrapper methods.
