# Scikit-Learn Pipeline

A Pipeline in Scikit-Learn is used to combine multiple data preprocessing
and machine learning steps into a single workflow.

A pipeline can include:

- Missing value imputation
- Feature scaling
- Encoding
- Feature selection
- Dimensionality reduction
- Machine learning model

The main goal is to make the machine learning workflow:

- Clean
- Reproducible
- Easier to manage
- Less error-prone
- Resistant to data leakage

## Why Do We Need Pipelines?

Without a pipeline, we may have to manually perform:

1. Imputation
2. Scaling
3. Feature selection
4. Model training

This can lead to repeated code and mistakes.

A Pipeline combines these steps into one object.

Example:

Raw Data
   ↓
Imputation
   ↓
Scaling
   ↓
Feature Selection
   ↓
Model

```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Age": [20, 25, np.nan, 35, 40, 45, 50, np.nan],
    "Income": [30000, 40000, 50000, np.nan, 70000, 80000, 90000, 100000],
    "Experience": [1, 3, 5, 7, np.nan, 12, 15, 18],
    "Target": [0, 0, 0, 1, 1, 1, 1, 1]
})

df
```

```python
X = df.drop(columns="Target")
y = df["Target"]
```

## Train-Test Split

Before creating the pipeline, we split the dataset into training and
testing sets.

The test data must remain unseen during preprocessing and model training.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

## Simple Pipeline

A simple pipeline can contain:

1. StandardScaler
2. LogisticRegression

The pipeline automatically applies scaling before training the model.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])
```

```python
pipeline.fit(
    X_train,
    y_train
)
```

```python
y_pred = pipeline.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
```

## Pipeline with Missing Value Imputation

Real-world datasets often contain missing values.

We can add `SimpleImputer` to the pipeline.

Workflow:

Missing Values
      ↓
Imputation
      ↓
Scaling
      ↓
Model

```python
from sklearn.impute import SimpleImputer

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression())
])
```

```python
pipeline.fit(
    X_train,
    y_train
)
```

```python
y_pred = pipeline.predict(X_test)

y_pred
```

## Pipeline Steps

Each step in a Pipeline has:

1. A name
2. A transformer or estimator

Example:

("imputer", SimpleImputer())

"imputer" → Name of the step

SimpleImputer() → Transformation

The final step is usually the machine learning model.

```python
pipeline.named_steps
```

```python
pipeline.named_steps["imputer"]
```

```python
pipeline.named_steps["scaler"]
```

## Pipeline with Feature Selection

Feature selection can also be included inside a pipeline.

Example workflow:

Imputation
    ↓
Scaling
    ↓
Feature Selection
    ↓
Model

```python
from sklearn.feature_selection import SelectKBest, f_classif

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("selector", SelectKBest(
        score_func=f_classif,
        k=2
    )),
    ("model", LogisticRegression())
])
```

```python
pipeline.fit(
    X_train,
    y_train
)
```

```python
y_pred = pipeline.predict(X_test)

accuracy_score(
    y_test,
    y_pred
)
```

## Pipeline with PCA

PCA can also be included in a pipeline.

Workflow:

Imputation
    ↓
Scaling
    ↓
PCA
    ↓
Model

```python
from sklearn.decomposition import PCA

pca_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=2)),
    ("model", LogisticRegression())
])
```

## Pipeline with Cross-Validation

Pipelines work very well with cross-validation.

The preprocessing steps are fitted separately within each training fold.

This helps prevent data leakage during model evaluation.

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    pipeline,
    X,
    y,
    cv=5,
    scoring="accuracy"
)

scores
```

```python
print("Mean Accuracy:", scores.mean())
```
