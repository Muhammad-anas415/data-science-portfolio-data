# Cross-Validation

Cross-validation is a model evaluation technique used to estimate how well a machine learning model will perform on unseen data.

Instead of relying on only one train-test split, cross-validation divides the training data into multiple parts called **folds**.

The model is trained and validated multiple times.

### Main Benefits

- Provides a more reliable estimate of model performance
- Reduces dependency on a single train-test split
- Helps detect overfitting
- Useful for comparing different models
- Commonly used during hyperparameter tuning

# K-Fold Cross-Validation

K-Fold Cross-Validation divides the dataset into **K equal or approximately equal folds**.

For example, with:

```text
K = 5

````python
import numpy as np
import pandas as pd

from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LogisticRegression
````

````python
import seaborn as sns

df = sns.load_dataset("iris")

df.head()
````

````python
# Features
X = df.drop("species", axis=1)

# Target
y = df["species"].map({
    "setosa": 0,
    "versicolor": 1,
    "virginica": 2
})

print("X shape:", X.shape)
print("y shape:", y.shape)
````

````python
kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print(kf)
````

````python
for fold, (train_index, validation_index) in enumerate(
    kf.split(X),
    start=1
):

    print(f"Fold {fold}")

    print("Training samples:", len(train_index))
    print("Validation samples:", len(validation_index))

    print("-" * 40)
````

# Using cross_val_score()

Scikit-learn provides `cross_val_score()` to perform cross-validation more easily.

Instead of manually creating folds and training the model, we can use:

```python
cross_val_score()

````python
from sklearn.model_selection import cross_val_score

model = LogisticRegression(
    max_iter=1000
)

scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="accuracy"
)

print("Cross-Validation Scores:")
print(scores)
````

````python
print(
    "Mean Accuracy:",
    scores.mean()
)

print(
    "Standard Deviation:",
    scores.std()
)
````
