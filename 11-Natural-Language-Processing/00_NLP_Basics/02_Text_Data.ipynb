#### Character Count

The number of characters in a text can provide a basic measure of its length.

Longer emails may contain more information, while shorter messages may contain only a few words.

We can calculate the number of characters in each email using Python's `len()` function.

```python
# Calculate character count
df["character_count"] = df["text"].str.len()
df[["text", "character_count"]].head()
```

####  Word Count

Word count measures the number of words contained in an email.

It provides a simple way to understand the length and structure of text.

Word count can later be useful for exploratory analysis and feature engineering.

```python
# Calculate word count
df["word_count"] = df["text"].str.split().str.len()
df[["text", "word_count"]].head()
```

#### Sentence Count

An email may contain one or multiple sentences.

Sentence count provides another way of measuring the structure of text.

A simple sentence-counting approach can be created by splitting text around common sentence-ending punctuation such as:

- `.`
- `!`
- `?`

More advanced NLP libraries provide specialized sentence tokenizers, which we will explore later.

```python
# Simple sentence count
df["sentence_count"] = (
    df["text"]
    .str.split(r"[.!?]+")
    .str.len()
)

df[["text", "sentence_count"]].head()
```
