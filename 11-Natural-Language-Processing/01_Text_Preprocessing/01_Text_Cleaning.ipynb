# Text Cleaning — Email Spam Detection

Text cleaning is one of the first steps in Natural Language Processing (NLP).

Raw text often contains unnecessary or inconsistent elements such as:

- Uppercase and lowercase letters
- HTML tags
- URLs
- Email addresses
- Punctuation
- Special characters
- Numbers
- Extra whitespace
- Repeated characters

These elements can make text analysis more difficult.

#### Workflow

```text
Raw Email
    ↓
Lowercase
    ↓
Remove HTML
    ↓
Remove URLs
    ↓
Remove Email Addresses
    ↓
Remove Punctuation
    ↓
Remove Special Characters
    ↓
Handle Numbers
    ↓
Remove Extra Whitespace
    ↓
Clean Email

#### 1. Convert Text to Lowercase

Computers treat uppercase and lowercase characters as different characters.

For example:

```text
"FREE"
"Free"
"free"

````python
# Convert text to lowercase
df["clean_text"] = df["text"].str.lower()
df[["text", "clean_text"]].head()
````

#### 2. Remove HTML Tags

Some emails may contain HTML formatting.

For example:

```html
<div>Hello <b>World</b></div>

````python
# Remove HTML tags

def remove_html(text):
    return re.sub(r"<.*?>", " ", text)

df["clean_text"] = df["clean_text"].apply(remove_html)

df[["text", "clean_text"]].head()
````

#### 3. Remove URLs

Emails can contain URLs such as:

```text
https://example.com
www.example.com

````python
# Remove URLs

def remove_urls(text):
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    return text

df["clean_text"] = df["clean_text"].apply(remove_urls)

df[["text", "clean_text"]].head()
````

#### 4. Remove Email Addresses

Email messages can contain email addresses such as:

```text
support@example.com
user@gmail.com

````python
# Remove email addresses

def remove_emails(text):
    return re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )

df["clean_text"] = df["clean_text"].apply(remove_emails)

df[["text", "clean_text"]].head()
````

#### 5. Remove Punctuation

Punctuation includes characters such as:

```text
. , ! ? : ; ' " ( ) [ ] { }

````python
# Remove punctuation

def remove_punctuation(text):
    return text.translate(str.maketrans("", "", string.punctuation))

df["clean_text"] = df["clean_text"].apply(remove_punctuation)

df[["text", "clean_text"]].head()
````

#### 6. Remove Special Characters

Text may contain unnecessary symbols such as:

```text
@ # $ % & * + = ~ ^

````python
# Remove special characters

def remove_special_characters(text):
    return re.sub(r"[^a-zA-Z0-9\s]", " ", text)

df["clean_text"] = df["clean_text"].apply(remove_special_characters)

df[["text", "clean_text"]].head()
````

#### 7. Handle Numbers

Emails may contain numbers such as:

```text
"Win $1000"
"Call 123456789"
"Get 50% discount"

````python
# Remove numbers
# Use this only if numbers are not important for your task.

def remove_numbers(text):
    return re.sub(r"\d+", " ", text)

df["clean_text"] = df["clean_text"].apply(remove_numbers)
````

#### 8. Remove Extra Whitespace

After removing URLs, HTML, punctuation, and other elements, multiple spaces may remain.

For example:

```text
"hello     world"

````python
# Remove extra whitespace

def remove_extra_whitespace(text):
    return re.sub(r"\s+", " ", text).strip()

df["clean_text"] = df["clean_text"].apply(remove_extra_whitespace)

df[["text", "clean_text"]].head()
````
