#### Stemming in NLP

Stemming is a Natural Language Processing (NLP) technique used to reduce words to a simpler base form called a **stem**.

The main idea is to remove prefixes or suffixes from words so that different forms of a word can be treated as related.

For example:

```text
playing   → play
played    → play
plays     → play

#### Stemming vs Lemmatization

Stemming and lemmatization both attempt to reduce words to a simpler form, but they work differently.

#### Stemming

Uses rules to remove parts of words.

```text
studies → studi
playing → play

#### Import nltk

````python
import nltk
````

####  Porter Stemmer

The **Porter Stemmer** is one of the most commonly used traditional stemming algorithms.

It applies a series of linguistic rules to remove common word suffixes.

For example:

```text
playing    → play
played     → play
studies    → studi
connection → connect

````python
from nltk.stem import PorterStemmer

# Create stemmer
porter = PorterStemmer()

words = [
    "play",
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connection",
    "connected"
]

for word in words:
    print(f"{word:15} → {porter.stem(word)}")
````

####  Snowball Stemmer

The **Snowball Stemmer** is an improved stemming framework that supports multiple languages.

For English, it is commonly used as an alternative to the Porter Stemmer.

It generally provides more consistent stemming results while remaining relatively fast.

````python
from nltk.stem import SnowballStemmer

# Create English Snowball Stemmer
snowball = SnowballStemmer("english")

for word in words:
    print(f"{word:15} → {snowball.stem(word)}")
````

#### Compare Stemming Algorithms

Different stemming algorithms can produce different stems for the same word.

Therefore, it is useful to compare their results.

We will compare:

- Porter Stemmer
- Snowball Stemmer

````python
# Compare Porter and Snowball

for word in words:
    porter_result = porter.stem(word)
    snowball_result = snowball.stem(word)

    print(
        f"{word:15} → "
        f"Porter: {porter_result:12} | "
        f"Snowball: {snowball_result}"
    )
````

#### Apply Stemming to a Sentence

Stemming is usually performed on individual tokens rather than directly on an entire sentence.

For example:

```text
"Customers are purchasing products"

````python
sentence = "Customers are purchasing products and receiving discounts"

tokens = sentence.lower().split()

stemmed_tokens = [
    porter.stem(word)
    for word in tokens
]

print("Original Tokens:")
print(tokens)

print("\nStemmed Tokens:")
print(stemmed_tokens)
````

####  Apply Stemming to Email Data

Now we will apply stemming to our Email Spam Detection dataset.

We will use the cleaned text created in the previous text-cleaning notebook.

The process will be:

```text
Clean Email
     ↓
Tokenization
     ↓
Porter Stemmer
     ↓
Stemmed Email

````python
import pandas as pd
import re
import string
# Load dataset
df = pd.read_csv(
    r"F:\Anas\DATA_SCIENCE_POTPOLIO\12-Natural-Language-Processing\00_NLP_Basics\combined_data.csv"
)

# Basic cleaning function from previous notebook
def clean_text(text):
    text = text.lower()
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Clean emails
df["clean_text"] = df["text"].apply(clean_text)
````

````python
# Display result
df[["text", "clean_text"]].head()
````

#### Creating a stemming function

````python
# Create stemming function

def stem_text(text):
    tokens = text.split()

    stemmed_tokens = [
        porter.stem(word)
        for word in tokens
    ]

    return " ".join(stemmed_tokens)
````

````python
df["stemmed_text"] = df["clean_text"].apply(stem_text)

df[["text", "clean_text", "stemmed_text"]].head()
````

````python
# Display examples

for i in range(5):

    print("ORIGINAL:")
    print(df["clean_text"].iloc[i])

    print("\nSTEMMED:")
    print(df["stemmed_text"].iloc[i])

    print("\n" + "=" * 80 + "\n")
````

````python
# Compare vocabulary size

original_vocab = set(
    " ".join(df["clean_text"]).split()
)

stemmed_vocab = set(
    " ".join(df["stemmed_text"]).split()
)

print("Original Vocabulary Size:", len(original_vocab))
print("Stemmed Vocabulary Size:", len(stemmed_vocab))
````
