#  Word Tokenization in NLP

Word Tokenization is the process of splitting text into smaller units called **tokens**.

A token can be a:

- Word
- Number
- Punctuation mark
- Special symbol

#### Import Libraries

````python
import nltk
import pandas as pd

from nltk.tokenize import word_tokenize
````

##### Basic Word Tokenization

Let's tokenize a simple sentence.

For example:

```text
"I am learning NLP."

````python
text = "I am learning NLP."

tokens = word_tokenize(text)

print(tokens)
````

#### Display Individual Tokens

We can display each token separately and assign it an index.

This makes it easier to understand how tokenization works.

````python
for i, token in enumerate(tokens, start=1):
    print(f"Token {i}: {token}")
````

#### Count the Number of Tokens

After tokenization, we can easily calculate the total number of tokens in a sentence.

````python
print("Number of tokens:", len(tokens))
````

#### Tokenization with Punctuation

Word tokenization does not necessarily mean extracting only words.

Depending on the tokenizer, punctuation marks can also become individual tokens.

For example:

```text
"Hello! How are you?"

````python
text = "Hello! How are you? I am learning NLP."

tokens = word_tokenize(text)

print(tokens)
````

#### Tokenization with Numbers

Numbers can also appear as separate tokens.

For example:

```text
"Get $1000 today!"

````python
text = "Get $1000 today! You can save 50%."

tokens = word_tokenize(text)

print(tokens)
````

#### Tokenize an Email

Now let's apply word tokenization to an email.

This is especially useful for our **Email Spam Detection** project because the machine learning model will eventually need to work with individual text tokens.

````python
email = """
Congratulations! You have won $1000.
Click the link now to claim your prize.
"""

tokens = word_tokenize(email)

for token in tokens:
    print(token)
````
