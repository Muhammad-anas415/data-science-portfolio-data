# Bag of Words (BoW)

Bag of Words (BoW) is a text representation technique that converts text into a numerical representation based on the frequency of words.

Machine Learning algorithms cannot directly understand raw text.

####  What is Bag of Words?

Bag of Words represents a document using the words that appear in the document.

It mainly considers:

- Which words are present?
- How frequently does each word appear?

It does **not** consider the order of the words.

#### Creating a Vocabulary Manually

````python
documents = [
    "I love NLP",
    "I love Python",
    "NLP is interesting"
]

# Create vocabulary
vocabulary = set()

for document in documents:
    words = document.lower().split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)

print("Vocabulary:")
print(vocabulary)
````

####  Creating a Bag of Words Manually

For each document, we count how many times every vocabulary word appears.

If a word does not appear in the document, its value is `0`.

If it appears once, its value is `1`.

If it appears multiple times, the value represents its frequency.

````python
documents = [
    "I love NLP",
    "I love Python",
    "NLP is interesting"
]

vocabulary = sorted(set(
    word
    for document in documents
    for word in document.lower().split()
))

bow_vectors = []

for document in documents:

    words = document.lower().split()

    vector = [
        words.count(word)
        for word in vocabulary
    ]

    bow_vectors.append(vector)

print("Vocabulary:")
print(vocabulary)

print("\nBag of Words:")
for vector in bow_vectors:
    print(vector)
````

#### Bag of Words Using Scikit-learn

Instead of manually creating the vocabulary and counting words, we can use Scikit-learn's `CountVectorizer`.

`CountVectorizer` automatically:

1. Creates a vocabulary.
2. Counts word occurrences.
3. Converts documents into numerical vectors.
4. Creates a document-term matrix.

This is the most practical way to implement Bag of Words for Machine Learning projects.

````python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love NLP",
    "I love Python",
    "NLP is interesting"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

print(X)
````

````python
X_array = X.toarray()

print(X_array)
````

#### Vocabulary Created by CountVectorizer

The vectorizer creates a vocabulary and assigns each word a column in the numerical matrix.

We can inspect this vocabulary using `get_feature_names_out()`.

````python
feature_names = vectorizer.get_feature_names_out()

print(feature_names)
````

````python
import pandas as pd

bow_df = pd.DataFrame(
    X_array,
    columns=feature_names
)

bow_df
````

#### Sparse Matrix

Text datasets usually contain thousands of vocabulary words.

However, an individual email contains only a small percentage of those words.

Therefore, most values in the Bag of Words matrix are `0`.

This creates a **sparse matrix**.

Scikit-learn's `CountVectorizer` stores the result as a sparse matrix to save memory.
