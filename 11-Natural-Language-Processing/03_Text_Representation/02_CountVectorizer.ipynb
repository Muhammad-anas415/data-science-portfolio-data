# CountVectorizer

CountVectorizer is a text representation technique provided by `scikit-learn` that converts a collection of text documents into a numerical matrix based on word frequency.

Machine learning models cannot directly understand raw text. CountVectorizer solves this problem by:

1. Building a vocabulary from the text.
2. Assigning an index to each unique word.
3. Counting how many times each word appears in each document.
4. Representing each document as a numerical vector.

####  Import CountVectorizer

We use `CountVectorizer` from `sklearn.feature_extraction.text` to convert text documents into numerical feature vectors.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

####  Sample Text Data


```python
documents = [
    "free offer today",
    "free money offer",
    "win free money",
    "hello how are you"
]

documents
```

#### Creating the CountVectorizer

We create a `CountVectorizer` object and use `fit_transform()` to:

- Learn the vocabulary from the documents.
- Convert the documents into a numerical representation.

The resulting matrix contains the frequency of each vocabulary word in every document.

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

X
```

#### Vocabulary

The `vocabulary_` attribute contains the words learned by CountVectorizer and their corresponding numerical indices.

Each unique word receives an index that determines its position in the feature matrix.

```python
vectorizer.vocabulary_
```

#### Feature Names

We can retrieve the vocabulary as an ordered list using `get_feature_names_out()`.

These feature names correspond to the columns of the numerical matrix.

```python
feature_names = vectorizer.get_feature_names_out()

print(feature_names)
```
