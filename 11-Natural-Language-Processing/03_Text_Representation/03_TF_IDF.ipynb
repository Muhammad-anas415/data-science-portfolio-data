# TF-IDF (Term Frequency-Inverse Document Frequency)

TF-IDF is a text representation technique used in Natural Language Processing (NLP) to convert text data into numerical features.

TF-IDF gives higher importance to words that are frequent in a particular document but less common across the entire collection of documents.

It consists of two main components:

- **TF (Term Frequency):** Measures how frequently a word appears in a document.
- **IDF (Inverse Document Frequency):** Measures how rare a word is across all documents.

The final TF-IDF score is calculated by combining Term Frequency and Inverse Document Frequency.

### Formula

TF-IDF = TF × IDF

#### Import library

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

#### Creating Sample Text Data

```python
documents = [
    "free offer today",
    "free money offer",
    "win free money",
    "hello how are you"
]
documents
```

#### Creating the TF-IDF Vectorizer

We create a `TfidfVectorizer` object from Scikit-learn.

The `fit_transform()` method:

1. Learns the vocabulary from the documents.
2. Calculates the TF-IDF scores.
3. Converts the text into numerical features.

```python
tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(documents)

X_tfidf
```

##### Extracting Feature Names

The `get_feature_names_out()` method returns all the words learned by the TF-IDF vectorizer.

Each word represents a feature in the resulting numerical matrix.

```python
feature_names = tfidf.get_feature_names_out()

print(feature_names)
```
