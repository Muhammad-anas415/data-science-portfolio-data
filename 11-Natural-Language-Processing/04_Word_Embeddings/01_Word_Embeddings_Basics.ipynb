# Word Embeddings

Word Embedding is a Natural Language Processing (NLP) technique that represents words as numerical vectors in a continuous vector space.

Unlike CountVectorizer and TF-IDF, which represent words mainly based on their frequency, word embeddings can capture relationships and semantic similarities between words.

For example:

- `king`
- `queen`
- `man`
- `woman`

Words with similar meanings or similar usage patterns tend to have similar vector representations.

### Main characteristics of Word Embeddings

- Convert words into numerical vectors.
- Represent words using dense vectors.
- Capture relationships between words.
- Capture semantic and syntactic information.
- Can be used as input to neural networks.

### Example

A word can be represented as:

`king → [0.21, -0.43, 0.87, 0.15, ...]`

The vector contains many numerical dimensions that represent learned characteristics of the word.

### Common Word Embedding Techniques

1. Word2Vec
2. GloVe
3. FastText
4. Embedding Layer

Word embeddings are widely used in NLP tasks such as:

- Text Classification
- Sentiment Analysis
- Spam Detection
- Machine Translation
- Named Entity Recognition
- Question Answering
- Text Generation

## Why Do We Need Word Embeddings?

Traditional text representation techniques such as Bag of Words and TF-IDF represent text using individual word features.

However, they have some limitations:

- They create very large feature spaces.
- They usually produce sparse representations.
- They do not naturally capture semantic similarity between words.

Word embeddings solve some of these limitations by representing words as dense numerical vectors.

For example, words such as:

`happy`, `joyful`, `glad`

may have similar vector representations because they are used in similar contexts.

## Dense Vector Representation

Word embeddings represent each word using a dense vector.

For example:

`hello → [0.21, 0.45, -0.18, 0.72]`

`hi → [0.19, 0.43, -0.16, 0.70]`

The vectors contain numerical values learned from text data.

Words that have similar meanings or appear in similar contexts can have similar vector representations.
