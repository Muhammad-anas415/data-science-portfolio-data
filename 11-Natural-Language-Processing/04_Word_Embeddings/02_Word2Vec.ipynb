# Word2Vec

Word2Vec is a word embedding technique used in Natural Language Processing (NLP) to represent words as dense numerical vectors.

The main idea behind Word2Vec is:

> Words that appear in similar contexts tend to have similar meanings.

Instead of representing a word as a simple count or frequency, Word2Vec learns a vector representation for each word from the surrounding words in a large collection of text.

For example:

`king → [0.21, -0.43, 0.87, ...]`

`queen → [0.19, -0.40, 0.82, ...]`

Words with similar meanings or similar usage patterns can have similar vector representations.

### Main Word2Vec Architectures

1. **CBOW (Continuous Bag of Words)**
2. **Skip-Gram**

#### Importing Word2Vec

We use the `Word2Vec` implementation provided by Gensim.

Gensim is a popular Python library for working with word embeddings and topic modeling.

```python
from gensim.models import Word2Vec
```

#### Creating Sample Text

```python
sentences = [
    ["free", "money", "offer"],
    ["free", "offer", "today"],
    ["win", "free", "money"],
    ["special", "offer", "today"],
    ["win", "special", "prize"]
]

sentences
```

#### Training the Word2Vec Model

We create a Word2Vec model using our tokenized sentences.

Important parameters:

- `vector_size` → Number of dimensions for each word vector.
- `window` → Number of surrounding words considered as context.
- `min_count` → Minimum number of times a word must appear.
- `workers` → Number of CPU threads used for training.

The model learns word representations from the context of the words.

```python
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)
```

#### Word2Vec Vocabulary

The trained model contains a vocabulary of words that were learned from the training sentences.

We can access the vocabulary using `model.wv.key_to_index`.

```python
print(model.wv.key_to_index)
```

#### Getting a Word Vector

Each word in the Word2Vec vocabulary is represented by a dense numerical vector.

We can retrieve the vector of a specific word using:

`model.wv["word"]`

The vector contains the numerical representation learned by Word2Vec.

```python
vector = model.wv["free"]

print(vector)
print("Vector Shape:", vector.shape)
```

#### Finding Similar Words

Word2Vec can calculate which words have similar vector representations.

The `most_similar()` method returns words that are closest to a given word in the learned vector space.

Similarity is usually calculated using cosine similarity.

```python
similar_words = model.wv.most_similar("free")

print(similar_words)
```
