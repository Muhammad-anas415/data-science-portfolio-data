#### NLP Workflow — Email Spam Detection

Natural Language Processing (NLP) involves a series of steps that transform raw human language into useful information that can be processed by Machine Learning and Deep Learning models.


The objective is to transform raw email messages into numerical features and use those features to classify emails as:

- **Spam**
- **Not Spam (Ham)**

#### NLP Workflow

The general workflow for our project is:

```text
Raw Email
    ↓
Data Collection
    ↓
Text Exploration
    ↓
Text Preprocessing
    ↓
Tokenization
    ↓
Text Representation
    ↓
Feature Engineering
    ↓
Train/Test Split
    ↓
Model Training
    ↓
Model Evaluation
    ↓
Prediction
    ↓
Deployment
