# Lemmatization in NLP

Lemmatization is a Natural Language Processing (NLP) technique used to reduce a word to its meaningful **base or dictionary form**, known as a **lemma**.

Unlike stemming, lemmatization considers the linguistic meaning and grammatical structure of a word.

### Examples

```text
running   → run
studies   → study
better    → good
children  → child

#### Import Nltk

````python
import nltk
# Download resources required for lemmatization
nltk.download("wordnet")
nltk.download("omw-1.4")
````

#### WordNet Lemmatizer

NLTK provides the `WordNetLemmatizer`, which uses the **WordNet lexical database** to find the lemma of a word.

It can convert different forms of a word into a meaningful base form.

For example:

```text
cars       → car
studies    → study
children   → child

````python
from nltk.stem import WordNetLemmatizer

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

words = [
    "cars",
    "studies",
    "children",
    "running",
    "played"
]

for word in words:
    print(f"{word:12} → {lemmatizer.lemmatize(word)}")
````

#### Part-of-Speech (POS) and Lemmatization

Lemmatization can produce different results depending on the grammatical role of a word.

For example, the word:

```text
"better"

````python
# Lemmatization with different POS tags

word = "better"

print("Noun:", lemmatizer.lemmatize(word, pos="n"))
print("Verb:", lemmatizer.lemmatize(word, pos="v"))
print("Adjective:", lemmatizer.lemmatize(word, pos="a"))
print("Adverb:", lemmatizer.lemmatize(word, pos="r"))
````

#### Stemming vs Lemmatization Example

Let's compare the output of a stemmer and a lemmatizer.

This demonstrates an important difference:

```text
Stemming
    ↓
Rule-based reduction
    ↓
May produce invalid words

Lemmatization
    ↓
Linguistic analysis
    ↓
Attempts to produce meaningful dictionary words

````python
from nltk.stem import PorterStemmer
porter = PorterStemmer()

words = [
    "studies",
    "studying",
    "connected",
    "connections",
    "children",
    "cars"
]

for word in words:

    stem = porter.stem(word)
    lemma = lemmatizer.lemmatize(word)

    print(
        f"{word:15} "
        f"→ Stem: {stem:12} "
        f"→ Lemma: {lemma}"
    )
````

#### Load the data 

````python
import pandas as pd

# Load dataset
df = pd.read_csv(r"F:\Anas\DATA_SCIENCE_PROJECTS\DATASETS\combined_data.csv")

df.head()
````

#### Clean the text

````python
import re
import string

def clean_text(text):

    text = text.lower()

    # Remove HTML
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Remove email addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["clean_text"] = df["text"].apply(clean_text)

df[["text", "clean_text"]].head()
````

#### Create a Lemmatization Function

We will create a reusable function that:

1. Splits the cleaned text into individual words.
2. Applies the WordNet Lemmatizer to each word.
3. Combines the lemmatized words back into a sentence.

````python
def lemmatize_text(text):

    # Tokenize text
    tokens = text.split()

    # Lemmatize each token
    lemmatized_tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    # Join tokens back into text
    return " ".join(lemmatized_tokens)
````

````python
# Apply lemmatization

df["lemmatized_text"] = df["clean_text"].apply(
    lemmatize_text
)

df[[
    "clean_text",
    "lemmatized_text"
]].head()
````
