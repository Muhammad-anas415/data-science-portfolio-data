### Sentence Tokenization in NLP

Sentence Tokenization is the process of splitting a paragraph or document into individual sentences.

#### Why Do We Need Sentence Tokenization?

A document or email can contain multiple sentences.

Instead of processing the entire document as one large piece of text, we can split it into individual sentences.

#### Download NLTK Sentence Tokenization Resource

NLTK provides a pre-trained sentence tokenizer that can identify sentence boundaries.

We will use the `punkt` tokenizer.

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

#### Basic Sentence Tokenization

We can use NLTK's `sent_tokenize()` function to split text into individual sentences.

```python
from nltk.tokenize import word_tokenize

text = "Hello brother. I am learning NLP."

tokens = word_tokenize(text)

print(tokens)
```

```python
from nltk.tokenize import sent_tokenize

text = "I love Python. I am learning NLP. NLP is interesting!"

sentences = sent_tokenize(text)

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")
```

#### Count the Number of Sentences

After tokenization, we can easily calculate how many sentences are present in a piece of text.

```python
print("Number of sentences:", len(sentences))
```

#### Create a Sentence Tokenization Function

We can create a reusable function that takes an email and returns a list of sentences.

This allows us to apply the same process to every email in the dataset.

```python
def sentence_tokenize(text):
    return sent_tokenize(str(text))
```

```python
df["sentences"] = df["text"].apply(sentence_tokenize)

df[["text", "sentences"]].head()
```

#### Inspect Individual Sentences

Let's inspect one email and display each sentence separately.

```python
email = df["text"].iloc[0]

sentences = sent_tokenize(str(email))

print("Original Email:")
print(email)

print("\n" + "=" * 60)

print("Individual Sentences:")

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")
```
