#  Subword Tokenization in NLP

Subword Tokenization is a technique that breaks words into smaller meaningful or frequently occurring pieces called **subwords**.

Instead of treating every complete word as a single token, a tokenizer can divide an unknown or rare word into smaller parts.

####  Main Subword Algorithms

##### 1. BPE — Byte Pair Encoding
BPE starts with smaller units and repeatedly combines frequent pairs.

Used in several modern NLP systems.

##### 2. WordPiece

WordPiece builds a vocabulary of subword units.

It is famously used by BERT-style models.

##### 3. Unigram

The Unigram model starts with a large vocabulary and removes tokens according to their probability.

##### 4. SentencePiece

SentencePiece is a tokenizer framework that can implement algorithms such as BPE and Unigram.

#### importing libraries

````python
from transformers import AutoTokenizer
````

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)
````

#### Basic Subword Tokenization

Let's tokenize a simple sentence using the BERT tokenizer.

Unlike normal word tokenization, the tokenizer can split a word into multiple subword pieces.

````python
text = "I am learning natural language processing."

tokens = tokenizer.tokenize(text)

print(tokens)
````

#### Tokenizing Complex Words

The real advantage of subword tokenization becomes clearer with uncommon or complex words.

A word that is not present as a complete vocabulary item can be divided into smaller pieces.

For example:

```text
unhappiness

````python
words = [
    "unhappiness",
    "playing",
    "beautiful",
    "unknownword"
]

for word in words:
    print(word)
    print(tokenizer.tokenize(word))
    print()
````

#### Convert Subwords into Token IDs

Machine Learning models cannot directly process text tokens.

The tokenizer therefore converts each token into a numerical **Token ID**.

For example:

```text
Text
 ↓
Tokens
 ↓
Token IDs

````python
text = "I am learning NLP."

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````
