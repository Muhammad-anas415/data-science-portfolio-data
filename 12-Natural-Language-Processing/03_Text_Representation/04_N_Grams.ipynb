# N-Grams

N-Grams are a text representation technique in Natural Language Processing (NLP) that groups consecutive words together to capture relationships between words.

The value of `N` determines how many consecutive words are included in each group.

### Types of N-Grams

- **Unigram (N=1):** One word at a time.
- **Bigram (N=2):** Two consecutive words.
- **Trigram (N=3):** Three consecutive words.
- **4-Gram (N=4):** Four consecutive words.

#### Import Libraries

We use `CountVectorizer` from Scikit-learn to generate N-Gram features from our text data.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

#### Sample Text Data

We first create a small collection of sentences to understand how different N-Grams are generated.

```python
documents = [
    "free money offer",
    "free offer today",
    "win free money",
    "special offer today"
]
documents
```

#### Unigrams

A unigram contains exactly one word.

For example:

"free money offer"

becomes:

- free
- money
- offer

Unigrams capture individual word information but do not capture relationships between neighboring words.

```python
vectorizer = CountVectorizer(ngram_range=(1, 1))

X_unigram = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

#### Bigrams

A bigram contains two consecutive words.

For example:

"free money offer"

produces:

- free money
- money offer

Bigrams provide more context than individual words because they capture relationships between two neighboring words.

```python
vectorizer = CountVectorizer(ngram_range=(2, 2))

X_bigram = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

#### Trigrams

A trigram contains three consecutive words.

For example:

"free money offer"

produces:

- free money offer

Trigrams can capture more context than unigrams and bigrams, but they also produce more features.

```python
vectorizer = CountVectorizer(ngram_range=(3, 3))

X_trigram = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
```

#### N-Grams with TF-IDF

N-Grams can also be combined with TF-IDF.

Using `ngram_range=(1, 2)` allows TF-IDF to represent both individual words and pairs of words.

This can be useful for email spam detection because phrases such as:

- "free offer"
- "claim prize"
- "click here"
- "win money"

can contain more useful information than individual words alone.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    ngram_range=(1, 2)
)

X_tfidf = tfidf.fit_transform(documents)

print("Shape:", X_tfidf.shape)
print("Features:")
print(tfidf.get_feature_names_out())
```
