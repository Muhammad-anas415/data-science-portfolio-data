# CBOW (Continuous Bag of Words)

CBOW is one of the two main architectures used by Word2Vec to learn word embeddings.

CBOW stands for:

**Continuous Bag of Words**

The main idea of CBOW is:

> Use surrounding context words to predict the target word.

#### Sample data

```python
sentences = [
    ["the", "cat", "is", "sleeping"],
    ["the", "dog", "is", "running"],
    ["the", "cat", "is", "running"],
    ["the", "dog", "is", "sleeping"]
]

sentences
```

#### Context Window

The `window` determines how many words around the target word are considered as context.

For example, with:

`window = 2`

Sentence:

`the cat is sleeping`

If the target word is:

`is`

The context can contain:

`the`, `cat`, `sleeping`

The model uses these surrounding words to learn the representation of the target word.

#### Creating CBOW Training Examples

To understand CBOW, we can manually create training examples.

Consider:

`the cat is sleeping`

Using a context window of 2, we can create examples where surrounding words are used to predict the center word.

For example:

Context:

`the`, `is`

Target:

`cat`

Therefore:

`[the, is] → cat`

```python
sentence = ["the", "cat", "is", "sleeping"]

context = ["the", "is"]
target = "cat"

print("Context:", context)
print("Target:", target)
```

#### Training CBOW Using Gensim

Gensim provides a Word2Vec implementation that can train both CBOW and Skip-Gram models.

By default:

`sg=0`

selects the **CBOW architecture**.

Therefore, we can train a CBOW model by setting:

`sg=0`

```python
from gensim.models import Word2Vec

cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=2,
    min_count=1,
    workers=4,
    sg=0
)
```

#### CBOW Vocabulary

After training, the model contains a vocabulary of the words that were learned from the training sentences.

We can inspect this vocabulary using `key_to_index`.

```python
print(cbow_model.wv.key_to_index)
```

#### Getting Word Vectors

The CBOW model learns a dense vector representation for each word.

We can retrieve a word vector using:

`model.wv["word"]`

The size of the vector depends on the `vector_size` parameter.

```python
word_vector = cbow_model.wv["cat"]

print("Vector:", word_vector)
print("Shape:", word_vector.shape)
```

#### Finding Similar Words

After training, we can find words that have similar vector representations.

The `most_similar()` method compares the learned word vectors and returns the most similar words.

```python
similar_words = cbow_model.wv.most_similar("cat")

print(similar_words)
```
