# Skip-Gram

Skip-Gram is one of the two main architectures used by Word2Vec to learn word embeddings.

The main idea of Skip-Gram is:

> Use a target word to predict its surrounding context words.

#### Sample data

```python
sentences = [
    ["the", "cat", "is", "sleeping"],
    ["the", "dog", "is", "running"],
    ["the", "cat", "is", "running"],
    ["the", "dog", "is", "sleeping"]
]
sentences
```

#### Context Window

The `window` parameter determines how many words around the target word are considered as context.

#### Creating Skip-Gram Training Pairs

Skip-Gram creates training examples by pairing a target word with each surrounding context word.

For example:

Sentence:

`the cat is sleeping`

Target:

`cat`

Context:

`the`, `is`

Training pairs:

`cat → the`

`cat → is`

This allows the model to learn relationships between words based on their context.

```python
sentence = ["the", "cat", "is", "sleeping"]

target = "cat"
context = ["the", "is"]

for word in context:
    print("Target:", target, "→ Context:", word)
```

#### Training Skip-Gram Using Gensim

Gensim provides the `Word2Vec` implementation for training Word2Vec models.

We select Skip-Gram using:

`sg=1`

Important parameters:

- `vector_size` → Size of each word vector.
- `window` → Number of surrounding words used as context.
- `min_count` → Minimum frequency required for a word.
- `workers` → Number of CPU threads.
- `sg=1` → Selects Skip-Gram.

```python
from gensim.models import Word2Vec

skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=2,
    min_count=1,
    workers=4,
    sg=1
)
```

####  Skip-Gram Vocabulary

After training, the model contains the vocabulary learned from the training sentences.

We can inspect the vocabulary using:

`model.wv.key_to_index`

```python
print(skipgram_model.wv.key_to_index)
```

#### Getting a Word Vector

Skip-Gram learns a dense numerical vector for each word in the vocabulary.

We can retrieve a word vector using:

`model.wv["word"]`

The size of the vector depends on the `vector_size` parameter.

```python
word_vector = skipgram_model.wv["cat"]

print("Vector:", word_vector)
print("Vector Shape:", word_vector.shape)
```

#### Finding Similar Words

After training the Skip-Gram model, we can find words with similar vector representations.

The `most_similar()` method compares word vectors and returns the most similar words.

```python
similar_words = skipgram_model.wv.most_similar(
    "cat",
    topn=5
)

print(similar_words)
```
