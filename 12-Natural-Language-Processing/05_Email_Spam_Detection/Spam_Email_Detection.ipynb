Data Preprocessing

```python
data[~data["text"].str.islower()]
```

```python
#convering the data into lowe case
data.text=data.text.str.lower()
data.head()
```

```python
#removes the html tages
data.text=data.text.apply(lambda x: BeautifulSoup(x, 'html.parser').get_text())
data.head()
```

```python
#removes the numbers
data.text=data.text.apply(lambda x: re.sub(r'\d+', '', x))
data.head()
```

```python
#removes the punctuations
data.text=data.text.apply(lambda x: x.translate(str.maketrans('', '', string.punctuation)))
data.head()
```

```python
#removes the urls
data.text=data.text.apply(lambda x: re.sub(r'http\S+', '', x))
data.head()
```

```python
#removes the extra spaces
data.text=data.text.apply(lambda x: " ".join(x.split()))
data.head()
```

```python
#cehcking for the urls
url_rows = data[data['text'].str.contains(r'http\S+|www\S+', regex=True, na=False)]
url_rows
```

```python
#converting the text into tokens by using nltk library
from nltk.tokenize import word_tokenize
data["text"] = data["text"].apply(lambda x: word_tokenize(x))
data.head()
```

```python
stop_words = set(stopwords.words("english"))
data["text"] = data["text"].apply(
    lambda words: [word for word in words if word.lower() not in stop_words]
)

data.head()
```

```python
lemmatizer = WordNetLemmatizer()
data["text"] = data["text"].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)
data.head()
```

```python
print(type(data["text"].iloc[0]))
print(data["text"].iloc[0])
```

```python
data["text"] = data["text"].apply(lambda x: " ".join(x))
data.head()
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(data["text"])
X
```

```python
y=data["label"] 
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
naive_bayes = MultinomialNB()

svm = SVC()

logistic_regression = LogisticRegression(max_iter=1000)
```

```python
naive_bayes.fit(X_train, y_train)

svm.fit(X_train, y_train)

logistic_regression.fit(X_train, y_train)
```

```python
nb_prediction = naive_bayes.predict(X_test)

svm_prediction = svm.predict(X_test)

lr_prediction = logistic_regression.predict(X_test)
```

```python
nb_accuracy = accuracy_score(y_test, nb_prediction)

svm_accuracy = accuracy_score(y_test, svm_prediction)

lr_accuracy = accuracy_score(y_test, lr_prediction)
print("Naive Bayes Accuracy:", nb_accuracy)

print("SVM Accuracy:", svm_accuracy)

print("Logistic Regression Accuracy:", lr_accuracy)
```

```python
results = {
    "Naive Bayes": nb_accuracy,
    "SVM": svm_accuracy,
    "Logistic Regression": lr_accuracy
}

for model, accuracy in results.items():
    print(f"{model}: {accuracy:.4f}")
```

```python
best_model = max(results, key=results.get)

print("Best Model:", best_model)

print("Best Accuracy:", results[best_model])
```

```python
def predict_email():

    # Take user input
    email = input("Enter your email message:\n")

    # 1. Lowercase
    email = email.lower()

    # 2. Remove HTML
    email = BeautifulSoup(email, "html.parser").get_text()

    # 3. Remove URLs
    email = re.sub(r'http\S+|www\S+', '', email)

    # 4. Remove punctuation
    email = email.translate(str.maketrans('', '', string.punctuation))

    # 5. Remove numbers
    email = re.sub(r'\d+', '', email)

    # 6. Remove extra spaces
    email = " ".join(email.split())

    # 7. Tokenization
    words = word_tokenize(email)

    # 8. Remove stop words
    words = [word for word in words if word not in stop_words]

    # 9. Lemmatization
    words = [lemmatizer.lemmatize(word) for word in words]

    # 10. Join words
    email = " ".join(words)

    # 11. Bag of Words
    email_vector = vectorizer.transform([email])

    # 12. Prediction
    prediction = logistic_regression.predict(email_vector)

    # 13. Probability
    probability = logistic_regression.predict_proba(email_vector)

    # 14. Result
    if prediction[0] == 1:
        print("\n🚨 Spam Email")
    else:
        print("\n✅ Not Spam (Ham)")

    print(f"Confidence: {max(probability[0]) * 100:.2f}%")
predict_email()    
```
