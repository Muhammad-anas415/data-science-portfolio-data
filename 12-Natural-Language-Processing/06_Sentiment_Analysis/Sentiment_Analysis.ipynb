### Text Preprocessing

```python
#convering into x and y
X=data['review']
y=data['sentiment']
```

```python
#checking the shape of x and y
X.shape,y.shape
```

```python
X[0]
```

```python
#Converting into lower case 
X = X.str.lower()
X[0]
```

```python
#removing html tags
X = X.apply(lambda x: BeautifulSoup(x, 'html.parser').get_text())
X[0]
```

```python
#removing punctuations
X = X.str.replace('[^\w\s]', '')
X[0]
```

```python
#removing urls
X = X.apply(lambda x: ' '.join([w for w in x.split() if 'http' not in w]))
X[0]
```

```python
#removing extra spaces
X = X.str.replace('\s+', ' ')
X[0]
```

Tokenization

```python
X = X.apply(word_tokenize)
```

```python
X[0]
```

```python
#removing punctuations
X = X.apply(lambda x: [item for item in x if item not in string.punctuation])
X[0]
```

```python
#removing the stop words
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
X = X.apply(lambda x: [item for item in x if item not in stop_words])
X[0]
```

```python
#lemmitization
lemmatizer = WordNetLemmatizer()

X = X.apply(lambda words: [lemmatizer.lemmatize(word) for word in words])
```

```python
#cehcking the lemmitization of first review
X[0]
```

```python
#joining the words for  applying the bag of words vectorization
X = X.apply(lambda words: " ".join(words))
X[0]
```

```python
#applying bag of words
vectorizer = CountVectorizer()

X_features = vectorizer.fit_transform(X)
X_features
```

Encoding the Target Variable 

```python
#encoding the target variable byusing the label encoder
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y = encoder.fit_transform(y)
y
```

Train Test Split The Data

```python
#traing test split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_features,
    y,
    test_size=0.2,
    random_state=42
)
```

checking the shape of the training and testing data 

```python
#printing the shape of the train and test data
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

selecting models

```python
#Naive Bayes
naive_bayes = MultinomialNB()
#Logistic Regression
logistic_regression = LogisticRegression(max_iter=1000)
```

traning naive bayes algorithim

```python
#Training Naive Bayes 
naive_bayes.fit(X_train, y_train)
```

training logistic regression algorithim

```python
#Training Logistic Regression
logistic_regression.fit(X_train, y_train)
```

predictions

```python
#Naive Bayes and Logistic Regression Prediction
nb_prediction = naive_bayes.predict(X_test)
lr_prediction = logistic_regression.predict(X_test)
```

Accuarcy

```python
#Accuracy of Naive Bayes and Logistic Regression
nb_accuracy = accuracy_score(y_test, nb_prediction)

print("Naive Bayes Accuracy :", nb_accuracy)
lr_accuracy = accuracy_score(y_test, lr_prediction)

print("Logistic Regression Accuracy :", lr_accuracy)
```

Results

```python
#Comparing the Accuracy
results = pd.DataFrame({
    "Model": ["Naive Bayes", "Logistic Regression"],
    "Accuracy": [nb_accuracy, lr_accuracy]
})

results
```

User Input for more reviews.

```python
from better_profanity import profanity
# Load profanity words
profanity.load_censor_words()

# Stopwords
stop_words = set(stopwords.words('english'))

# Keep important negation words
stop_words.discard("not")
stop_words.discard("no")
stop_words.discard("never")

# Lemmatizer
lemmatizer = WordNetLemmatizer()

def predict_sentiment():

    # ===========================
    # Step 1: Take User Input
    # ===========================
    review = input("Enter your movie review: ")

    # ===========================
    # Step 2: Check for Abusive Words
    # ===========================
    if profanity.contains_profanity(review):
        print("\n⚠️ Your review contains abusive or offensive language.")
        print("Please use respectful language and try again.")
        return

    # ===========================
    # Step 3: Lowercase
    # ===========================
    review = review.lower()

    # ===========================
    # Step 4: Remove HTML Tags
    # ===========================
    review = BeautifulSoup(review, "html.parser").get_text()

    # ===========================
    # Step 5: Remove URLs
    # ===========================
    review = re.sub(r'http\S+|www\S+', '', review)

    # ===========================
    # Step 6: Remove Punctuation
    # ===========================
    review = review.translate(str.maketrans('', '', string.punctuation))

    # ===========================
    # Step 7: Remove Numbers
    # ===========================
    review = re.sub(r'\d+', '', review)

    # ===========================
    # Step 8: Remove Extra Spaces
    # ===========================
    review = " ".join(review.split())

    # ===========================
    # Step 9: Tokenization
    # ===========================
    words = review.split()

    # ===========================
    # Step 10: Remove Stopwords
    # ===========================
    words = [word for word in words if word not in stop_words]

    # ===========================
    # Step 11: Lemmatization
    # ===========================
    words = [lemmatizer.lemmatize(word) for word in words]

    # ===========================
    # Step 12: Join Words
    # ===========================
    review = " ".join(words)

    # ===========================
    # Step 13: Vectorization
    # ===========================
    review_vector = vectorizer.transform([review])

    # ===========================
    # Step 14: Prediction
    # ===========================
    prediction = logistic_regression.predict(review_vector)

    # ===========================
    # Step 15: Display Result
    # ===========================
    if prediction[0] == 1:
        print("\nPrediction: 😊 Positive")
    else:
        print("\nPrediction: 😞 Negative")


# Run the function
predict_sentiment()
```
